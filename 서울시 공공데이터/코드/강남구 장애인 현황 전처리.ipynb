```python
resident_filename = list(resident_uploaded.keys())[0]

resident = pd.read_csv(
    resident_filename,
    encoding="utf-8-sig"
)

print("크기:", resident.shape)
print("컬럼:")
print(resident.columns.tolist())

print("\n분기:")
print(resident["quarter"].unique())

print("\n행정동 수:")
print(resident["dong_name"].nunique())
```

```python
# 2025 Q3/Q4 평균 등록인구
resident_mean = (
    resident
    .groupby("dong_name", as_index=False)
    .agg(
        resident_mean=("resident_total", "mean")
    )
)

print("크기:", resident_mean.shape)
display(resident_mean.head())
```

```python
disabled_total = (
    gangnam_disabled[
        (gangnam_disabled["disability_type"] == "합계") &
        (gangnam_disabled["sex"] == "계")
    ][["dong_name", "disabled_count"]]
    .rename(columns={"disabled_count": "disabled_total"})
    .copy()
)

print("크기:", disabled_total.shape)
display(disabled_total.head())
```

```python
disabled_scale = disabled_total.merge(
    resident_mean,
    on="dong_name",
    how="left"
)

disabled_scale["disabled_per_1000"] = (
    disabled_scale["disabled_total"]
    / disabled_scale["resident_mean"]
    * 1000
)

print("크기:", disabled_scale.shape)

print("\n결측치:")
print(disabled_scale.isna().sum())

print("\ndisabled_per_1000 분포:")
print(
    disabled_scale["disabled_per_1000"]
    .describe()
)

print("\n인구 1,000명당 등록장애인 규모:")
display(
    disabled_scale[
        ["dong_name", "disabled_total",
         "resident_mean", "disabled_per_1000"]
    ]
    .sort_values(
        "disabled_per_1000",
        ascending=False
    )
    .round(2)
)
```

```python
type_profile = (
    gangnam_disabled[
        (gangnam_disabled["sex"] == "계") &
        (gangnam_disabled["disability_type"] != "합계")
    ]
    .groupby("disability_type")
    .agg(
        total_count=("disabled_count", "sum"),
        mean_count=("disabled_count", "mean"),
        median_count=("disabled_count", "median"),
        zero_dongs=("disabled_count", lambda x: (x == 0).sum()),
        nonzero_dongs=("disabled_count", lambda x: (x > 0).sum())
    )
    .reset_index()
)

# 강남구 전체 등록장애인 15,517명 대비 유형별 비율
gangnam_disabled_total = disabled_total["disabled_total"].sum()

type_profile["share_pct"] = (
    type_profile["total_count"]
    / gangnam_disabled_total
    * 100
)

type_profile = type_profile.sort_values(
    "total_count",
    ascending=False
)

display(type_profile.round(2))
```

```python
print(
    "15개 장애유형 총합:",
    type_profile["total_count"].sum()
)

print(
    "강남구 등록장애인 총합:",
    gangnam_disabled_total
)

print(
    "장애유형 비율 합:",
    type_profile["share_pct"].sum()
)
```

```python
# 성별 '계', 전체 '합계' 제외
type_detail = gangnam_disabled[
    (gangnam_disabled["sex"] == "계") &
    (gangnam_disabled["disability_type"] != "합계")
].copy()

# 행정동별 전체 장애인 수 결합
type_detail = type_detail.merge(
    disabled_total,
    on="dong_name",
    how="left"
)

# 해당 동 등록장애인 중 각 장애유형의 비율
type_detail["type_ratio"] = (
    type_detail["disabled_count"]
    / type_detail["disabled_total"]
)

print("크기:", type_detail.shape)

print("\n행정동별 구성비 합 검증:")
ratio_check = (
    type_detail.groupby("dong_name")["type_ratio"]
    .sum()
)

print(ratio_check.describe())
print(
    "1과의 최대 오차:",
    (ratio_check - 1).abs().max()
)
```

```python
main_types = [
    "지체",
    "청각",
    "뇌병변",
    "시각",
    "지적장애",
    "신장장애",
    "자폐성",
    "정신장애"
]

main_type_feature = (
    type_detail[
        type_detail["disability_type"].isin(main_types)
    ]
    .pivot(
        index="dong_name",
        columns="disability_type",
        values="type_ratio"
    )
    .reset_index()
)

main_type_feature = main_type_feature.rename(
    columns={
        "지체": "physical_ratio",
        "청각": "hearing_ratio",
        "뇌병변": "brain_lesion_ratio",
        "시각": "visual_ratio",
        "지적장애": "intellectual_ratio",
        "신장장애": "kidney_ratio",
        "자폐성": "autism_ratio",
        "정신장애": "mental_ratio"
    }
)

# 나머지 7개 유형을 other로 통합
other_feature = (
    type_detail[
        ~type_detail["disability_type"].isin(main_types)
    ]
    .groupby("dong_name", as_index=False)
    .agg(other_disability_ratio=("type_ratio", "sum"))
)

disability_type_feature = main_type_feature.merge(
    other_feature,
    on="dong_name",
    how="left"
)

print("크기:", disability_type_feature.shape)
print("\n결측치:")
print(disability_type_feature.isna().sum())

display(disability_type_feature.head().round(4))
```

```python
ratio_cols = [
    "physical_ratio",
    "hearing_ratio",
    "brain_lesion_ratio",
    "visual_ratio",
    "intellectual_ratio",
    "kidney_ratio",
    "autism_ratio",
    "mental_ratio",
    "other_disability_ratio"
]

disability_type_feature["ratio_sum"] = (
    disability_type_feature[ratio_cols].sum(axis=1)
)

print(
    disability_type_feature["ratio_sum"].describe()
)

print(
    "1과의 최대 오차:",
    (disability_type_feature["ratio_sum"] - 1)
    .abs()
    .max()
)
```

```python
gender_disabled = gangnam_disabled[
    gangnam_disabled["disability_type"] == "합계"
].copy()

gender_pivot = (
    gender_disabled[
        gender_disabled["sex"].isin(["남자", "여자"])
    ]
    .pivot(
        index="dong_name",
        columns="sex",
        values="disabled_count"
    )
    .reset_index()
)

gender_pivot = gender_pivot.rename(
    columns={
        "남자": "disabled_male",
        "여자": "disabled_female"
    }
)

# 전체 장애인 수 결합
gender_pivot = gender_pivot.merge(
    disabled_total,
    on="dong_name",
    how="left"
)

# 성별 구성비
gender_pivot["disabled_male_ratio"] = (
    gender_pivot["disabled_male"]
    / gender_pivot["disabled_total"]
)

gender_pivot["disabled_female_ratio"] = (
    gender_pivot["disabled_female"]
    / gender_pivot["disabled_total"]
)

print("크기:", gender_pivot.shape)

print("\n결측치:")
print(gender_pivot.isna().sum())

display(gender_pivot.head().round(4))
```

```python
gender_pivot["gender_count_check"] = (
    gender_pivot["disabled_male"]
    + gender_pivot["disabled_female"]
    - gender_pivot["disabled_total"]
)

gender_pivot["gender_ratio_check"] = (
    gender_pivot["disabled_male_ratio"]
    + gender_pivot["disabled_female_ratio"]
)

print(
    "성별 인원 합계 불일치 동:",
    (gender_pivot["gender_count_check"] != 0).sum()
)

print(
    "성별 구성비 1과의 최대 오차:",
    (gender_pivot["gender_ratio_check"] - 1)
    .abs()
    .max()
)

print("\n성별 구성비 분포:")
print(
    gender_pivot[
        ["disabled_male_ratio", "disabled_female_ratio"]
    ].describe()
)
```

```python
# 규모 Feature
scale_feature = disabled_scale[
    ["dong_name", "disabled_total", "disabled_per_1000"]
].copy()

# 성별 Feature
gender_feature = gender_pivot[
    [
        "dong_name",
        "disabled_male_ratio",
        "disabled_female_ratio"
    ]
].copy()

# 최종 결합
disabled_feature = (
    scale_feature
    .merge(
        gender_feature,
        on="dong_name",
        how="left"
    )
    .merge(
        disability_type_feature.drop(columns=["ratio_sum"]),
        on="dong_name",
        how="left"
    )
)

print("최종 크기:", disabled_feature.shape)

print("\n결측치:")
print(disabled_feature.isna().sum())

print("\n중복 행정동:")
print(disabled_feature["dong_name"].duplicated().sum())

display(disabled_feature.round(4))
```

```python
# 성별 구성비
gender_sum = (
    disabled_feature["disabled_male_ratio"]
    + disabled_feature["disabled_female_ratio"]
)

# 장애유형 구성비
type_cols = [
    "physical_ratio",
    "hearing_ratio",
    "brain_lesion_ratio",
    "visual_ratio",
    "intellectual_ratio",
    "kidney_ratio",
    "autism_ratio",
    "mental_ratio",
    "other_disability_ratio"
]

type_sum = disabled_feature[type_cols].sum(axis=1)

print(
    "성별 구성비 최대 오차:",
    (gender_sum - 1).abs().max()
)

print(
    "장애유형 구성비 최대 오차:",
    (type_sum - 1).abs().max()
)
```

```python
gangnam = disabled_age_raw[
    disabled_age_raw["자치구별(2)"] == "강남구"
].iloc[0]

gangnam_disability_background = pd.DataFrame([{
    "region": "강남구",
    "year": 2025,

    # 전체
    "disabled_total": gangnam["2025"],
    "disabled_male": gangnam["2025.1"],
    "disabled_female": gangnam["2025.2"],

    # 장애정도
    "severe_total": gangnam["2025.3"],
    "severe_male": gangnam["2025.4"],
    "severe_female": gangnam["2025.5"],

    "nonsevere_total": gangnam["2025.6"],
    "nonsevere_male": gangnam["2025.7"],
    "nonsevere_female": gangnam["2025.8"],

    # 연령
    "age_0_9": gangnam["2025.9"],
    "age_10_19": gangnam["2025.12"],
    "age_20_29": gangnam["2025.15"],
    "age_30_39": gangnam["2025.18"],
    "age_40_49": gangnam["2025.21"],
    "age_50_59": gangnam["2025.24"],
    "age_60_69": gangnam["2025.27"],
    "age_70_79": gangnam["2025.30"],
    "age_80_plus": gangnam["2025.33"]
}])

# 숫자형 변환
numeric_cols = gangnam_disability_background.columns.difference(
    ["region"]
)

gangnam_disability_background[numeric_cols] = (
    gangnam_disability_background[numeric_cols]
    .apply(pd.to_numeric, errors="coerce")
)

display(gangnam_disability_background)
```

```python
bg = gangnam_disability_background.iloc[0]

# ① 남 + 여 = 전체
gender_diff = (
    bg["disabled_male"]
    + bg["disabled_female"]
    - bg["disabled_total"]
)

# ② 심한 장애 + 심하지 않은 장애 = 전체
severity_diff = (
    bg["severe_total"]
    + bg["nonsevere_total"]
    - bg["disabled_total"]
)

# ③ 9개 연령구간 합 = 전체
age_cols = [
    "age_0_9",
    "age_10_19",
    "age_20_29",
    "age_30_39",
    "age_40_49",
    "age_50_59",
    "age_60_69",
    "age_70_79",
    "age_80_plus"
]

age_sum = bg[age_cols].sum()
age_diff = age_sum - bg["disabled_total"]

print("전체 장애인:", bg["disabled_total"])

print("\n남 + 여 차이:", gender_diff)
print("장애정도 합계 차이:", severity_diff)
print("연령구간 합계:", age_sum)
print("연령구간 합계 차이:", age_diff)

print("\n결측치:")
print(gangnam_disability_background.isna().sum())
```

```python
background_feature = gangnam_disability_background.copy()

# 전체 성별 구성
background_feature["disabled_male_ratio"] = (
    background_feature["disabled_male"]
    / background_feature["disabled_total"]
)

background_feature["disabled_female_ratio"] = (
    background_feature["disabled_female"]
    / background_feature["disabled_total"]
)

# 장애정도 구성
background_feature["severe_ratio"] = (
    background_feature["severe_total"]
    / background_feature["disabled_total"]
)

background_feature["nonsevere_ratio"] = (
    background_feature["nonsevere_total"]
    / background_feature["disabled_total"]
)
```

```python
age_cols = [
    "age_0_9",
    "age_10_19",
    "age_20_29",
    "age_30_39",
    "age_40_49",
    "age_50_59",
    "age_60_69",
    "age_70_79",
    "age_80_plus"
]

for col in age_cols:
    background_feature[f"{col}_ratio"] = (
        background_feature[col]
        / background_feature["disabled_total"]
    )
```

```python
age_ratio_cols = [
    f"{col}_ratio"
    for col in age_cols
]

gender_ratio_sum = (
    background_feature["disabled_male_ratio"]
    + background_feature["disabled_female_ratio"]
)

severity_ratio_sum = (
    background_feature["severe_ratio"]
    + background_feature["nonsevere_ratio"]
)

age_ratio_sum = (
    background_feature[age_ratio_cols]
    .sum(axis=1)
)

print(
    "성별 구성비 합:",
    gender_ratio_sum.iloc[0]
)

print(
    "장애정도 구성비 합:",
    severity_ratio_sum.iloc[0]
)

print(
    "연령 구성비 합:",
    age_ratio_sum.iloc[0]
)

print("\n주요 구성비:")

print(
    "심한 장애:",
    round(background_feature["severe_ratio"].iloc[0] * 100, 2),
    "%"
)

print(
    "심하지 않은 장애:",
    round(background_feature["nonsevere_ratio"].iloc[0] * 100, 2),
    "%"
)

for col in age_ratio_cols:
    print(
        col,
        round(background_feature[col].iloc[0] * 100, 2),
        "%"
    )
```

```python
output_file = "gangnam_disability_background_2025.csv"

background_feature.round(6).to_csv(
    output_file,
    index=False,
    encoding="utf-8-sig"
)

print("저장 완료:", output_file)
print("크기:", background_feature.shape)
print("\n컬럼:")
print(background_feature.columns.tolist())
```
