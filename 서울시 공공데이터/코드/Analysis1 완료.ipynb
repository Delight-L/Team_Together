```python
import pandas as pd

# 1. Dong-Month Behavior Mart
behavior = pd.read_csv(
    "gangnam_db1_dong_month_behavior_mart_2025_07_12.csv",
    encoding="utf-8-sig"
)

# 2. 주민등록인구
resident = pd.read_csv(
    "gangnam_resident_population_features_2025Q3_Q4.csv",
    encoding="utf-8-sig"
)

# 3. 1인가구
single = pd.read_csv(
    "gangnam_single_household_features_2025.csv",
    encoding="utf-8-sig"
)

# 4. 경제취약
welfare = pd.read_csv(
    "gangnam_welfare_economic_features_2024_05.csv",
    encoding="utf-8-sig"
)

# 5. 장애인
disability = pd.read_csv(
    "gangnam_disability_features_2025.csv",
    encoding="utf-8-sig"
)

print("[Behavior]", behavior.shape)
print("[Resident]", resident.shape)
print("[Single Household]", single.shape)
print("[Welfare]", welfare.shape)
print("[Disability]", disability.shape)
```

```python
datasets = {
    "Behavior": behavior,
    "Resident": resident,
    "Single Household": single,
    "Welfare": welfare,
    "Disability": disability
}

for name, df in datasets.items():
    print(f"\n===== {name} =====")
    print("shape:", df.shape)

    for i, col in enumerate(df.columns, 1):
        print(f"{i:2d}. {col}")
```

```python
# ==========================================
# 1. Behavior에서 22개 행정동 마스터 생성
# ==========================================

dong_master = (
    behavior[
        ["행정동코드", "행정동명"]
    ]
    .drop_duplicates()
    .sort_values("행정동코드")
    .reset_index(drop=True)
)

print("[Dong Master]")
print("행정동 수:", len(dong_master))
print(
    "코드 중복:",
    dong_master["행정동코드"].duplicated().sum()
)
print(
    "동명 중복:",
    dong_master["행정동명"].duplicated().sum()
)

display(dong_master)


# ==========================================
# 2. 각 데이터의 행정동 목록
# ==========================================

dong_sets = {
    "Behavior": set(behavior["행정동명"].unique()),
    "Resident": set(resident["dong_name"].unique()),
    "Single Household": set(single["행정동명"].unique()),
    "Welfare": set(welfare["dong_name"].unique()),
    "Disability": set(disability["dong_name"].unique())
}

print("\n[행정동 수]")
for name, dong_set in dong_sets.items():
    print(name, len(dong_set))


# ==========================================
# 3. Behavior를 기준으로 차이 확인
# ==========================================

base = dong_sets["Behavior"]

for name, dong_set in dong_sets.items():

    if name == "Behavior":
        continue

    print(f"\n===== {name} =====")

    print(
        "Behavior에는 있고 해당 데이터에는 없음:",
        sorted(base - dong_set)
    )

    print(
        "해당 데이터에는 있고 Behavior에는 없음:",
        sorted(dong_set - base)
    )
```

```python
# 확인할 주민등록인구 구조 변수
resident_check_cols = [
    "resident_total",
    "foreigner_ratio",
    "resident_age_10_ratio_10plus",
    "resident_age_20_ratio_10plus",
    "resident_age_30_ratio_10plus",
    "resident_age_40_ratio_10plus",
    "resident_age_50_ratio_10plus",
    "resident_age_60_plus_ratio_10plus"
]

# Q3 / Q4를 가로로 펼치기
resident_compare = resident.pivot(
    index="dong_name",
    columns="quarter",
    values=resident_check_cols
)

print("quarter 값:", resident["quarter"].unique())
print("비교 테이블 shape:", resident_compare.shape)

# 분기별 기술통계
for col in resident_check_cols:
    print(f"\n===== {col} =====")

    temp = resident.pivot(
        index="dong_name",
        columns="quarter",
        values=col
    )

    q3 = temp.iloc[:, 0]
    q4 = temp.iloc[:, 1]

    diff = q4 - q3

    print("Q3 평균:", round(q3.mean(), 6))
    print("Q4 평균:", round(q4.mean(), 6))
    print("평균 변화:", round(diff.mean(), 6))
    print("평균 절대 변화:", round(diff.abs().mean(), 6))
    print("최대 절대 변화:", round(diff.abs().max(), 6))
```

```python
resident_feature_cols = [
    "resident_total",
    "foreigner_ratio",
    "resident_age_10_ratio_10plus",
    "resident_age_20_ratio_10plus",
    "resident_age_30_ratio_10plus",
    "resident_age_40_ratio_10plus",
    "resident_age_50_ratio_10plus",
    "resident_age_60_plus_ratio_10plus"
]

resident_dong = (
    resident
    .groupby("dong_name")[resident_feature_cols]
    .mean()
    .reset_index()
)

# 이름 정리
resident_dong = resident_dong.rename(
    columns={
        "dong_name": "행정동명",
        "resident_total": "resident_total_mean",
        "foreigner_ratio": "foreigner_ratio_mean",
        "resident_age_10_ratio_10plus": "resident_age_10_ratio",
        "resident_age_20_ratio_10plus": "resident_age_20_ratio",
        "resident_age_30_ratio_10plus": "resident_age_30_ratio",
        "resident_age_40_ratio_10plus": "resident_age_40_ratio",
        "resident_age_50_ratio_10plus": "resident_age_50_ratio",
        "resident_age_60_plus_ratio_10plus": "resident_age_60_plus_ratio"
    }
)

# Behavior의 Dong Master에서 행정동코드 부여
resident_dong = (
    dong_master
    .merge(
        resident_dong,
        on="행정동명",
        how="left",
        validate="one_to_one"
    )
)

print("Resident Dong Feature shape:", resident_dong.shape)
print("행정동 수:", resident_dong["행정동코드"].nunique())
print("결측값:", resident_dong.isna().sum().sum())
print("행정동코드 중복:", resident_dong["행정동코드"].duplicated().sum())

display(resident_dong.head())
```

```python
import numpy as np

# ------------------------------------------
# 1. 연령 gap 합
# ------------------------------------------

age_gap_cols = [
    "single_age_20_gap",
    "single_age_30_gap",
    "single_age_40_gap",
    "single_age_50_gap",
    "single_age_60_plus_gap"
]

single["age_gap_sum"] = single[age_gap_cols].sum(axis=1)

print("[연령 gap 합]")
print(single["age_gap_sum"].describe())


# ------------------------------------------
# 2. 성별 비율 합
# ------------------------------------------

single["gender_ratio_sum"] = (
    single["single_gender_1_ratio"]
    + single["single_gender_2_ratio"]
)

print("\n[성별 비율 합]")
print(single["gender_ratio_sum"].describe())


# ------------------------------------------
# 3. 규모/변화 변수 상관관계
# ------------------------------------------

single_scale_cols = [
    "single_household_mean",
    "2025_Q3_1인세대",
    "2025_Q4_1인세대",
    "official_single_household_mean",
    "official_single_household_change",
    "official_single_household_change_rate"
]

print("\n[규모·변화 변수 상관관계]")
display(
    single[single_scale_cols]
    .corr()
    .round(3)
)


# ------------------------------------------
# 4. 기본 검증
# ------------------------------------------

print("\n[기본 검증]")
print("행정동 수:", single["행정동코드"].nunique())
print(
    "key 중복:",
    single["행정동코드"].duplicated().sum()
)
print(
    "원본 결측값:",
    single.drop(
        columns=["age_gap_sum", "gender_ratio_sum"]
    ).isna().sum().sum()
)
```

```python
single_feature_cols = [
    # 연령구조 gap
    "single_age_20_gap",
    "single_age_30_gap",
    "single_age_40_gap",
    "single_age_50_gap",
    "single_age_60_plus_gap",

    # 성별구조
    "single_gender_1_ratio",
    "single_gender_2_ratio",

    # 규모
    "single_household_mean",
    "official_single_household_mean",

    # 변화
    "official_single_household_change_rate"
]

single_dong = single[
    [
        "행정동코드",
        "행정동명"
    ] + single_feature_cols
].copy()

print("Single Household Feature shape:", single_dong.shape)
print("Feature 수:", len(single_feature_cols))
print("행정동 수:", single_dong["행정동코드"].nunique())
print("결측값:", single_dong.isna().sum().sum())
print(
    "행정동코드 중복:",
    single_dong["행정동코드"].duplicated().sum()
)

display(single_dong.head())
```

```python
welfare_cols = [
    "livelihood_per_1000",
    "medical_per_1000",
    "housing_per_1000",
    "elderly_livelihood_per_1000",
    "elderly_medical_per_1000",
    "elderly_housing_per_1000"
]

print("[기본 구조]")
print("shape:", welfare.shape)
print("행정동 수:", welfare["dong_name"].nunique())
print("중복:", welfare["dong_name"].duplicated().sum())

print("\n[컬럼별 결측값]")
print(welfare[welfare_cols].isna().sum())

print("\n[상관관계]")
display(
    welfare[welfare_cols]
    .corr()
    .round(3)
)

# |r| >= 0.8인 조합 추출
corr = welfare[welfare_cols].corr()

high_corr = []

for i in range(len(welfare_cols)):
    for j in range(i + 1, len(welfare_cols)):
        r = corr.iloc[i, j]

        if abs(r) >= 0.8:
            high_corr.append({
                "Feature_1": welfare_cols[i],
                "Feature_2": welfare_cols[j],
                "상관계수": r
            })

high_corr = pd.DataFrame(high_corr)

if len(high_corr) > 0:
    high_corr["절대상관"] = high_corr["상관계수"].abs()
    high_corr = (
        high_corr
        .sort_values("절대상관", ascending=False)
        .drop(columns="절대상관")
        .reset_index(drop=True)
    )

print("\n[|r| >= 0.8]")
print("관계 수:", len(high_corr))
display(high_corr)
```

```python
welfare_dong = (
    dong_master
    .merge(
        welfare.rename(
            columns={"dong_name": "행정동명"}
        ),
        on="행정동명",
        how="left",
        validate="one_to_one"
    )
)

print("Welfare Dong Feature shape:", welfare_dong.shape)
print("행정동 수:", welfare_dong["행정동코드"].nunique())
print(
    "행정동코드 중복:",
    welfare_dong["행정동코드"].duplicated().sum()
)

print("\n[결측값]")
print(welfare_dong.isna().sum())

display(welfare_dong.head())
```

```python
disability_type_cols = [
    "brain_lesion_ratio",
    "visual_ratio",
    "kidney_ratio",
    "autism_ratio",
    "mental_ratio",
    "intellectual_ratio",
    "physical_ratio",
    "hearing_ratio",
    "other_disability_ratio"
]

# ==========================================
# 1. 성별 구성비 합
# ==========================================

disability["gender_ratio_sum"] = (
    disability["disabled_male_ratio"]
    + disability["disabled_female_ratio"]
)

print("[장애인 성별 비율 합]")
print(
    disability["gender_ratio_sum"]
    .describe()
)


# ==========================================
# 2. 장애유형 구성비 합
# ==========================================

disability["type_ratio_sum"] = (
    disability[disability_type_cols]
    .sum(axis=1)
)

print("\n[장애유형 비율 합]")
print(
    disability["type_ratio_sum"]
    .describe()
)


# ==========================================
# 3. 전체 Feature 상관관계
# ==========================================

disability_feature_cols = [
    "disabled_total",
    "disabled_per_1000",
    "disabled_male_ratio",
    "disabled_female_ratio"
] + disability_type_cols

corr = disability[
    disability_feature_cols
].corr()

high_corr = []

for i in range(len(disability_feature_cols)):
    for j in range(i + 1, len(disability_feature_cols)):

        r = corr.iloc[i, j]

        if abs(r) >= 0.8:
            high_corr.append({
                "Feature_1": disability_feature_cols[i],
                "Feature_2": disability_feature_cols[j],
                "상관계수": r
            })

high_corr = pd.DataFrame(high_corr)

if len(high_corr) > 0:
    high_corr["절대상관"] = (
        high_corr["상관계수"].abs()
    )

    high_corr = (
        high_corr
        .sort_values(
            "절대상관",
            ascending=False
        )
        .drop(columns="절대상관")
        .reset_index(drop=True)
    )

print("\n[|r| >= 0.8]")
print("관계 수:", len(high_corr))
display(high_corr.round(3))


# ==========================================
# 4. 기본 검증
# ==========================================

print("\n[기본 검증]")
print(
    "행정동 수:",
    disability["dong_name"].nunique()
)
print(
    "중복:",
    disability["dong_name"].duplicated().sum()
)
print(
    "원본 결측값:",
    disability[
        ["dong_name"] + disability_feature_cols
    ].isna().sum().sum()
)
```

```python
disability_feature_cols = [
    "disabled_total",
    "disabled_per_1000",
    "disabled_male_ratio",
    "disabled_female_ratio",
    "brain_lesion_ratio",
    "visual_ratio",
    "kidney_ratio",
    "autism_ratio",
    "mental_ratio",
    "intellectual_ratio",
    "physical_ratio",
    "hearing_ratio",
    "other_disability_ratio"
]

disability_dong = (
    dong_master
    .merge(
        disability[
            ["dong_name"] + disability_feature_cols
        ].rename(
            columns={"dong_name": "행정동명"}
        ),
        on="행정동명",
        how="left",
        validate="one_to_one"
    )
)

print(
    "Disability Dong Feature shape:",
    disability_dong.shape
)

print(
    "행정동 수:",
    disability_dong["행정동코드"].nunique()
)

print(
    "행정동코드 중복:",
    disability_dong["행정동코드"].duplicated().sum()
)

print(
    "결측값:",
    disability_dong.isna().sum().sum()
)

display(disability_dong.head())
```

```python
# ID를 제외한 Behavior Feature
behavior_cols = [
    col for col in behavior.columns
    if col not in [
        "행정동코드",
        "행정동명",
        "기준연월"
    ]
]

print("Behavior Feature 수:", len(behavior_cols))


# ==========================================
# 6개월 평균
# ==========================================

behavior_dong = (
    behavior
    .groupby(
        ["행정동코드", "행정동명"]
    )[behavior_cols]
    .mean()
    .reset_index()
)

# 출처/의미가 명확하도록 suffix 부여
behavior_dong = behavior_dong.rename(
    columns={
        col: f"{col}_mean"
        for col in behavior_cols
    }
)

print(
    "Behavior Dong Feature shape:",
    behavior_dong.shape
)

print(
    "행정동 수:",
    behavior_dong["행정동코드"].nunique()
)

print(
    "행정동코드 중복:",
    behavior_dong["행정동코드"]
    .duplicated()
    .sum()
)

print(
    "결측값:",
    behavior_dong.isna().sum().sum()
)

display(behavior_dong.head())
```

```python
# ==========================================
# DB1 Dong Feature Mart 생성
# ==========================================

dong_feature_mart = (
    behavior_dong

    .merge(
        resident_dong,
        on=["행정동코드", "행정동명"],
        how="left",
        validate="one_to_one"
    )

    .merge(
        single_dong,
        on=["행정동코드", "행정동명"],
        how="left",
        validate="one_to_one"
    )

    .merge(
        welfare_dong,
        on=["행정동코드", "행정동명"],
        how="left",
        validate="one_to_one"
    )

    .merge(
        disability_dong,
        on=["행정동코드", "행정동명"],
        how="left",
        validate="one_to_one"
    )
)


# ==========================================
# 최종 구조 검증
# ==========================================

print(
    "Dong Feature Mart shape:",
    dong_feature_mart.shape
)

print(
    "행정동 수:",
    dong_feature_mart["행정동코드"].nunique()
)

print(
    "행정동코드 중복:",
    dong_feature_mart["행정동코드"]
    .duplicated()
    .sum()
)

print(
    "전체 Feature 수:",
    dong_feature_mart.shape[1] - 2
)

print(
    "\n[결측값이 있는 컬럼]"
)

missing = (
    dong_feature_mart
    .isna()
    .sum()
)

print(
    missing[missing > 0]
)

display(dong_feature_mart.head())
```

```python
import numpy as np
import pandas as pd

# ==========================================
# 1. ML 후보 Feature
# ==========================================

id_cols = [
    "행정동코드",
    "행정동명"
]

feature_cols = [
    col for col in dong_feature_mart.columns
    if col not in id_cols
]

X = dong_feature_mart[feature_cols].copy()

print("[기본 구조]")
print("행정동 수:", len(X))
print("Feature 수:", X.shape[1])


# ==========================================
# 2. 결측값
# ==========================================

missing = X.isna().sum()
missing = missing[missing > 0]

print("\n[결측값]")
print(missing)


# ==========================================
# 3. 분산 확인
# ==========================================

std_values = X.std()

zero_variance = std_values[
    np.isclose(std_values, 0)
]

print("\n[분산이 0에 가까운 Feature]")
print("개수:", len(zero_variance))

if len(zero_variance) > 0:
    display(zero_variance)


# ==========================================
# 4. 전체 상관관계
# ==========================================

corr = X.corr()

corr_pairs = []

for i in range(len(feature_cols)):
    for j in range(i + 1, len(feature_cols)):

        r = corr.iloc[i, j]

        if pd.notna(r) and abs(r) >= 0.8:
            corr_pairs.append({
                "Feature_1": feature_cols[i],
                "Feature_2": feature_cols[j],
                "상관계수": r,
                "절대상관": abs(r)
            })

corr_pairs = pd.DataFrame(corr_pairs)

if len(corr_pairs) > 0:
    corr_pairs = (
        corr_pairs
        .sort_values(
            "절대상관",
            ascending=False
        )
        .reset_index(drop=True)
    )


# ==========================================
# 5. 강도별 요약
# ==========================================

print("\n[고상관 관계 요약]")

if len(corr_pairs) == 0:
    print("|r| >= 0.8 관계 없음")

else:
    print(
        "|r| >= 0.8 :",
        (corr_pairs["절대상관"] >= 0.8).sum()
    )

    print(
        "|r| >= 0.9 :",
        (corr_pairs["절대상관"] >= 0.9).sum()
    )

    print(
        "|r| >= 0.95:",
        (corr_pairs["절대상관"] >= 0.95).sum()
    )

    print("\n[|r| >= 0.9 관계]")

    display(
        corr_pairs[
            corr_pairs["절대상관"] >= 0.9
        ][
            ["Feature_1", "Feature_2", "상관계수"]
        ].round(3)
    )
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
import numpy as np
import pandas as pd

# ==========================================
# V4 기준 Feature Group 수정
# ==========================================

feature_groups_v4 = {
    group: [
        col for col in cols
        if col in ml_feature_v4.columns
    ]
    for group, cols in feature_groups.items()
}

# Welfare는 이미 PCA 완료 → 이번 진단 제외
feature_groups_v4.pop(
    "J_복지경제취약",
    None
)


# ==========================================
# 도메인별 PCA 설명분산 진단
# ==========================================

pca_summary = []

for group_name, cols in feature_groups_v4.items():

    X_group = ml_feature_v4[cols].copy()

    # 표준화
    scaler = StandardScaler()
    X_scaled = scaler.fit_transform(X_group)

    # 전체 PCA
    pca = PCA()
    pca.fit(X_scaled)

    explained = pca.explained_variance_ratio_
    cumulative = np.cumsum(explained)

    # 80%, 90%, 95% 설명에 필요한 PC 수
    n80 = np.argmax(cumulative >= 0.80) + 1
    n90 = np.argmax(cumulative >= 0.90) + 1
    n95 = np.argmax(cumulative >= 0.95) + 1

    pca_summary.append({
        "도메인": group_name,
        "원본_Feature수": len(cols),
        "PC1_설명분산": explained[0],
        "PC1_PC2_누적": (
            cumulative[1]
            if len(cumulative) >= 2
            else cumulative[0]
        ),
        "80%_PC수": n80,
        "90%_PC수": n90,
        "95%_PC수": n95
    })


pca_domain_summary = pd.DataFrame(
    pca_summary
)

display(
    pca_domain_summary.round(3)
)
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
import pandas as pd
import numpy as np

# ==========================================
# 도메인별 사용할 PC 개수
# ==========================================

domain_n_components = {
    "A_유동활동규모": 1,
    "B_유동인구구조": 2,
    "C_시간요일활동": 2,
    "D_시민생활관심집단": 2,
    "E_이동체류": 2,
    "F_통신생활서비스금융": 3,
    "G_주민인구구조": 3,
    "H_1인가구구조": 3,
    "I_장애관련구조": 4
}

domain_pc_parts = []

# ==========================================
# A~I 도메인별 PCA
# ==========================================

for group_name, n_pc in domain_n_components.items():

    cols = feature_groups_v4[group_name]

    X_group = ml_feature_v4[cols].copy()

    scaler = StandardScaler()
    X_scaled = scaler.fit_transform(X_group)

    pca = PCA(n_components=n_pc)
    X_pca = pca.fit_transform(X_scaled)

    pc_cols = [
        f"{group_name}_PC{i+1}"
        for i in range(n_pc)
    ]

    pc_df = pd.DataFrame(
        X_pca,
        columns=pc_cols,
        index=ml_feature_v4.index
    )

    domain_pc_parts.append(pc_df)


# ==========================================
# A~I 결합
# ==========================================

domain_pc_matrix = pd.concat(
    domain_pc_parts,
    axis=1
)


# ==========================================
# J. 기존 Welfare PC1 추가
# ==========================================

domain_pc_matrix[
    "J_복지경제취약_PC1"
] = ml_feature_v4[
    "welfare_vulnerability_pc1"
].values


# ==========================================
# ID 추가
# ==========================================

domain_feature_matrix = pd.concat(
    [
        ml_feature_v4[
            ["행정동코드", "행정동명"]
        ].reset_index(drop=True),

        domain_pc_matrix.reset_index(drop=True)
    ],
    axis=1
)


# ==========================================
# 검증
# ==========================================

print("[Domain Feature Matrix]")
print(
    "shape:",
    domain_feature_matrix.shape
)

print(
    "Feature 수:",
    domain_feature_matrix.shape[1] - 2
)

print(
    "결측값:",
    domain_feature_matrix.isna().sum().sum()
)

print("\n[도메인별 PC 수]")
for group, n_pc in domain_n_components.items():
    print(group, ":", n_pc)

print(
    "J_복지경제취약 : 1"
)

display(
    domain_feature_matrix.head()
)
```

```python
# ==========================================
# Domain Feature의 실제 분산 기여도 확인
# ==========================================

domain_prefix_map = {
    "A": "A_유동활동규모",
    "B": "B_유동인구구조",
    "C": "C_시간요일활동",
    "D": "D_시민생활관심집단",
    "E": "E_이동체류",
    "F": "F_통신생활서비스금융",
    "G": "G_주민인구구조",
    "H": "H_1인가구구조",
    "I": "I_장애관련구조",
    "J": "J_복지경제취약"
}

pc_cols = [
    col for col in domain_feature_matrix.columns
    if col not in ["행정동코드", "행정동명"]
]

rows = []

for code, domain_name in domain_prefix_map.items():

    cols = [
        col for col in pc_cols
        if col.startswith(f"{code}_")
    ]

    variances = (
        domain_feature_matrix[cols]
        .var(ddof=0)
    )

    rows.append({
        "도메인": domain_name,
        "PC수": len(cols),
        "분산합": variances.sum(),
        "PC당_평균분산": variances.mean()
    })

domain_variance = pd.DataFrame(rows)

domain_variance["전체분산_기여율"] = (
    domain_variance["분산합"]
    / domain_variance["분산합"].sum()
)

print("[도메인별 현재 거리 기여 구조]")

display(
    domain_variance.round(4)
)
```

```python
# ==========================================
# Domain-balanced Feature Matrix 생성
# 각 도메인의 총 분산 = 1
# ==========================================

balanced_pc_matrix = domain_pc_matrix.copy()

balance_info = []

for code, domain_name in domain_prefix_map.items():

    cols = [
        col for col in balanced_pc_matrix.columns
        if col.startswith(f"{code}_")
    ]

    # 현재 도메인의 총 분산
    current_variance = (
        balanced_pc_matrix[cols]
        .var(ddof=0)
        .sum()
    )

    # 총 분산을 1로 맞추기 위한 가중치
    weight = 1 / np.sqrt(current_variance)

    # 해당 도메인의 모든 PC에 동일한 가중치 적용
    balanced_pc_matrix[cols] = (
        balanced_pc_matrix[cols] * weight
    )

    new_variance = (
        balanced_pc_matrix[cols]
        .var(ddof=0)
        .sum()
    )

    balance_info.append({
        "도메인": domain_name,
        "PC수": len(cols),
        "기존_분산합": current_variance,
        "가중치": weight,
        "조정후_분산합": new_variance
    })


balance_info = pd.DataFrame(balance_info)


# ==========================================
# ID 결합
# ==========================================

balanced_domain_feature_matrix = pd.concat(
    [
        domain_feature_matrix[
            ["행정동코드", "행정동명"]
        ].reset_index(drop=True),

        balanced_pc_matrix.reset_index(drop=True)
    ],
    axis=1
)


# ==========================================
# 검증
# ==========================================

print("[Domain Balance 결과]")

display(
    balance_info.round(4)
)

print(
    "\nshape:",
    balanced_domain_feature_matrix.shape
)

print(
    "Feature 수:",
    balanced_domain_feature_matrix.shape[1] - 2
)

print(
    "결측값:",
    balanced_domain_feature_matrix.isna().sum().sum()
)


# ==========================================
# 실제 조정 후 기여율 확인
# ==========================================

check_rows = []

for code, domain_name in domain_prefix_map.items():

    cols = [
        col for col in balanced_pc_matrix.columns
        if col.startswith(f"{code}_")
    ]

    variance_sum = (
        balanced_pc_matrix[cols]
        .var(ddof=0)
        .sum()
    )

    check_rows.append({
        "도메인": domain_name,
        "조정후_분산합": variance_sum
    })

balance_check = pd.DataFrame(check_rows)

balance_check["전체분산_기여율"] = (
    balance_check["조정후_분산합"]
    / balance_check["조정후_분산합"].sum()
)

print("\n[조정 후 실제 기여율]")

display(
    balance_check.round(4)
)
```

```python
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
import pandas as pd

# ==========================================
# K-Means k 후보 진단
# ==========================================

X_cluster = balanced_domain_feature_matrix.drop(
    columns=["행정동코드", "행정동명"]
)

k_results = []

for k in range(2, 7):

    model = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=50
    )

    labels = model.fit_predict(X_cluster)

    silhouette = silhouette_score(
        X_cluster,
        labels
    )

    cluster_sizes = (
        pd.Series(labels)
        .value_counts()
        .sort_index()
        .to_dict()
    )

    min_cluster_size = min(
        cluster_sizes.values()
    )

    k_results.append({
        "k": k,
        "silhouette_score": silhouette,
        "inertia": model.inertia_,
        "min_cluster_size": min_cluster_size,
        "cluster_sizes": cluster_sizes
    })


k_diagnostic = pd.DataFrame(k_results)

print("[K-Means k 후보 진단]")

display(
    k_diagnostic.round({
        "silhouette_score": 4,
        "inertia": 4
    })
)
```

```python
from itertools import combinations
import numpy as np
import pandas as pd

# ==========================================
# 군집별 도메인 분리 정도 계산
# ==========================================

analysis_df = balanced_domain_feature_matrix.copy()

analysis_df["cluster_k2"] = cluster_compare["cluster_k2"].values
analysis_df["cluster_k3"] = cluster_compare["cluster_k3"].values


def domain_centroid_separation(df, cluster_col):

    results = []

    cluster_ids = sorted(
        df[cluster_col].unique()
    )

    for code, domain_name in domain_prefix_map.items():

        cols = [
            col for col in balanced_pc_matrix.columns
            if col.startswith(f"{code}_")
        ]

        centroids = (
            df.groupby(cluster_col)[cols]
            .mean()
        )

        distances = []

        for c1, c2 in combinations(cluster_ids, 2):

            v1 = centroids.loc[c1].values
            v2 = centroids.loc[c2].values

            distance = np.linalg.norm(v1 - v2)

            distances.append(distance)

        results.append({
            "도메인": domain_name,
            "평균_centroid거리": np.mean(distances),
            "최대_centroid거리": np.max(distances)
        })

    return (
        pd.DataFrame(results)
        .sort_values(
            "평균_centroid거리",
            ascending=False
        )
        .reset_index(drop=True)
    )


# ==========================================
# k=2
# ==========================================

k2_domain_sep = domain_centroid_separation(
    analysis_df,
    "cluster_k2"
)

print("=" * 70)
print("k=2 도메인별 군집 분리 정도")
print("=" * 70)

display(
    k2_domain_sep.round(4)
)


# ==========================================
# k=3
# ==========================================

k3_domain_sep = domain_centroid_separation(
    analysis_df,
    "cluster_k3"
)

print("=" * 70)
print("k=3 도메인별 군집 분리 정도")
print("=" * 70)

display(
    k3_domain_sep.round(4)
)
```

```python
# ==========================================
# k=3 Cluster 0이 다른 군집과
# 어떤 도메인에서 구별되는지 확인
# ==========================================

rows = []

for code, domain_name in domain_prefix_map.items():

    cols = [
        col for col in balanced_pc_matrix.columns
        if col.startswith(f"{code}_")
    ]

    centroids = (
        analysis_df
        .groupby("cluster_k3")[cols]
        .mean()
    )

    # Cluster 0 ↔ Cluster 1
    dist_01 = np.linalg.norm(
        centroids.loc[0].values
        - centroids.loc[1].values
    )

    # Cluster 0 ↔ Cluster 2
    dist_02 = np.linalg.norm(
        centroids.loc[0].values
        - centroids.loc[2].values
    )

    # Cluster 1 ↔ Cluster 2
    dist_12 = np.linalg.norm(
        centroids.loc[1].values
        - centroids.loc[2].values
    )

    rows.append({
        "도메인": domain_name,
        "C0_vs_C1": dist_01,
        "C0_vs_C2": dist_02,
        "C1_vs_C2": dist_12,
        "C0_평균거리": (dist_01 + dist_02) / 2
    })


k3_pairwise_domain = (
    pd.DataFrame(rows)
    .sort_values(
        "C0_평균거리",
        ascending=False
    )
    .reset_index(drop=True)
)

print(
    "[k=3 Cluster 0의 도메인별 분리 특성]"
)

display(
    k3_pairwise_domain.round(4)
)
```

```python
# ==========================================
# k=3 주요 구조 도메인의 군집별 PC 중심값
# ==========================================

important_domains = [
    "J",
    "G",
    "B",
    "I",
    "H"
]

important_cols = []

for code in important_domains:

    cols = [
        col for col in balanced_pc_matrix.columns
        if col.startswith(f"{code}_")
    ]

    important_cols.extend(cols)


k3_centroid_profile = (
    analysis_df
    .groupby("cluster_k3")[important_cols]
    .mean()
    .T
)

k3_centroid_profile.columns = [
    "Cluster_0",
    "Cluster_1",
    "Cluster_2"
]

# Cluster 0가 나머지 두 군집 평균과 얼마나 다른지도 계산
k3_centroid_profile[
    "C0_minus_C1C2_mean"
] = (
    k3_centroid_profile["Cluster_0"]
    -
    (
        k3_centroid_profile["Cluster_1"]
        + k3_centroid_profile["Cluster_2"]
    ) / 2
)

k3_centroid_profile[
    "abs_difference"
] = (
    k3_centroid_profile[
        "C0_minus_C1C2_mean"
    ].abs()
)

k3_centroid_profile = (
    k3_centroid_profile
    .sort_values(
        "abs_difference",
        ascending=False
    )
)

print(
    "[k=3 Cluster 0 주요 구조적 차이 방향]"
)

display(
    k3_centroid_profile.round(4)
)
```

```python
from sklearn.cluster import KMeans
from sklearn.metrics import adjusted_rand_score
import pandas as pd
import numpy as np

# ==========================================
# K-Means 초기값 안정성 검증
# ==========================================

stability_results = []

for k in [2, 3]:

    # 기준 군집
    reference_model = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=50
    )

    reference_labels = reference_model.fit_predict(
        X_cluster
    )

    ari_scores = []

    # 여러 random seed에서 반복
    for seed in range(100):

        model = KMeans(
            n_clusters=k,
            random_state=seed,
            n_init=50
        )

        labels = model.fit_predict(
            X_cluster
        )

        ari = adjusted_rand_score(
            reference_labels,
            labels
        )

        ari_scores.append(ari)

    stability_results.append({
        "k": k,
        "ARI_mean": np.mean(ari_scores),
        "ARI_min": np.min(ari_scores),
        "ARI_max": np.max(ari_scores),
        "ARI_std": np.std(ari_scores),
        "ARI_1.0_count": sum(
            np.isclose(ari_scores, 1.0)
        )
    })


stability_df = pd.DataFrame(
    stability_results
)

print(
    "[K-Means 초기화 안정성 검증]"
)

display(
    stability_df.round(4)
)
```

```python
# ==========================================
# 최종 후보 k=2
# 도메인별 군집 중심 방향 확인
# ==========================================

k2_profile_rows = []

for code, domain_name in domain_prefix_map.items():

    cols = [
        col for col in balanced_pc_matrix.columns
        if col.startswith(f"{code}_")
    ]

    centroids = (
        analysis_df
        .groupby("cluster_k2")[cols]
        .mean()
    )

    for col in cols:

        c0 = centroids.loc[0, col]
        c1 = centroids.loc[1, col]

        k2_profile_rows.append({
            "도메인": domain_name,
            "PC": col,
            "Cluster_0": c0,
            "Cluster_1": c1,
            "C0_minus_C1": c0 - c1,
            "abs_difference": abs(c0 - c1)
        })


k2_profile = (
    pd.DataFrame(k2_profile_rows)
    .sort_values(
        "abs_difference",
        ascending=False
    )
    .reset_index(drop=True)
)

print("[k=2 군집별 PC 중심 차이]")

display(
    k2_profile.round(4)
)
```

```python
# ==========================================
# k=2 주요 A~E 도메인
# 원변수 기준 군집 평균 비교
# ==========================================

main_domains = [
    "A_유동활동규모",
    "B_유동인구구조",
    "C_시간요일활동",
    "D_시민생활관심집단",
    "E_이동체류"
]

main_raw_cols = []

for group in main_domains:
    main_raw_cols.extend(
        feature_groups_v4[group]
    )


# V4 원변수 + 최종 k=2 cluster
raw_profile_df = ml_feature_v4[
    ["행정동코드", "행정동명"] + main_raw_cols
].copy()

raw_profile_df["cluster_k2"] = (
    cluster_compare["cluster_k2"].values
)


# ==========================================
# 원변수 표준화
# 단위가 서로 다르므로 군집 차이 비교용
# ==========================================

scaler_raw = StandardScaler()

raw_scaled = pd.DataFrame(
    scaler_raw.fit_transform(
        raw_profile_df[main_raw_cols]
    ),
    columns=main_raw_cols,
    index=raw_profile_df.index
)

raw_scaled["cluster_k2"] = (
    raw_profile_df["cluster_k2"].values
)


# ==========================================
# 군집별 표준화 평균
# ==========================================

cluster_raw_mean = (
    raw_scaled
    .groupby("cluster_k2")
    .mean()
    .T
)

cluster_raw_mean.columns = [
    "Cluster_0",
    "Cluster_1"
]

cluster_raw_mean["C0_minus_C1"] = (
    cluster_raw_mean["Cluster_0"]
    - cluster_raw_mean["Cluster_1"]
)

cluster_raw_mean["abs_difference"] = (
    cluster_raw_mean["C0_minus_C1"].abs()
)


# 도메인 정보 추가
feature_to_domain = {}

for group in main_domains:
    for col in feature_groups_v4[group]:
        feature_to_domain[col] = group

cluster_raw_mean["도메인"] = (
    cluster_raw_mean.index.map(
        feature_to_domain
    )
)


cluster_raw_mean = (
    cluster_raw_mean
    .sort_values(
        "abs_difference",
        ascending=False
    )
)


print(
    "[k=2 A~E 원변수 기준 군집 차이]"
)

display(
    cluster_raw_mean[
        [
            "도메인",
            "Cluster_0",
            "Cluster_1",
            "C0_minus_C1",
            "abs_difference"
        ]
    ].round(3)
)
```

```python
# ==========================================
# k=2 F~J 원변수 기준 군집 차이
# ==========================================

fj_domains = [
    "F_통신생활서비스금융",
    "G_주민인구구조",
    "H_1인가구구조",
    "I_장애관련구조"
]

fj_raw_cols = []

for group in fj_domains:
    fj_raw_cols.extend(
        feature_groups_v4[group]
    )

# J는 이미 PCA 축으로 압축되어 있으므로 추가
fj_raw_cols.append(
    "welfare_vulnerability_pc1"
)


# ==========================================
# 데이터 구성
# ==========================================

fj_profile_df = ml_feature_v4[
    ["행정동코드", "행정동명"] + fj_raw_cols
].copy()

fj_profile_df["cluster_k2"] = (
    cluster_compare["cluster_k2"].values
)


# ==========================================
# 비교를 위한 표준화
# ==========================================

scaler_fj = StandardScaler()

fj_scaled = pd.DataFrame(
    scaler_fj.fit_transform(
        fj_profile_df[fj_raw_cols]
    ),
    columns=fj_raw_cols,
    index=fj_profile_df.index
)

fj_scaled["cluster_k2"] = (
    fj_profile_df["cluster_k2"].values
)


# ==========================================
# 군집 평균
# ==========================================

fj_cluster_mean = (
    fj_scaled
    .groupby("cluster_k2")
    .mean()
    .T
)

fj_cluster_mean.columns = [
    "Cluster_0",
    "Cluster_1"
]

fj_cluster_mean["C0_minus_C1"] = (
    fj_cluster_mean["Cluster_0"]
    - fj_cluster_mean["Cluster_1"]
)

fj_cluster_mean["abs_difference"] = (
    fj_cluster_mean[
        "C0_minus_C1"
    ].abs()
)


# ==========================================
# 도메인 표시
# ==========================================

feature_to_domain_fj = {}

for group in fj_domains:
    for col in feature_groups_v4[group]:
        feature_to_domain_fj[col] = group

feature_to_domain_fj[
    "welfare_vulnerability_pc1"
] = "J_복지경제취약"

fj_cluster_mean["도메인"] = (
    fj_cluster_mean.index.map(
        feature_to_domain_fj
    )
)

fj_cluster_mean = (
    fj_cluster_mean
    .sort_values(
        "abs_difference",
        ascending=False
    )
)


print(
    "[k=2 F~J 원변수 기준 군집 차이]"
)

display(
    fj_cluster_mean[
        [
            "도메인",
            "Cluster_0",
            "Cluster_1",
            "C0_minus_C1",
            "abs_difference"
        ]
    ].round(3)
)
```

```python
# ==========================================
# DB1 최종 k=2 군집 결과 테이블 생성
# ==========================================

# 군집 해석용 대표 원변수
representative_features = [

    # A. 유동활동 규모
    "total_flow_mean",
    "flow_per_point_mean",

    # B. 유동인구 연령구조
    "age_20_ratio_mean",
    "age_30_ratio_mean",
    "age_50_ratio_mean",
    "age_60_plus_ratio_mean",

    # C. 시간대 활동
    "daytime_ratio_mean",
    "morning_ratio_mean",

    # D. 시민생활 관심집단
    "very_low_outing_ratio_mean",
    "very_low_outing_comm_ratio_mean",
    "low_comm_ratio_mean",

    # E. 이동·체류
    "평일 총 이동 횟수_mean",
    "집 추정 위치 평일 총 체류시간_mean",

    # F. 통신·생활서비스
    "평균 통화대상자 수_mean",
    "배달 서비스 사용일수_mean",

    # G. 주민 인구구조
    "foreigner_ratio_mean",
    "resident_age_20_ratio",
    "resident_age_60_plus_ratio",

    # H. 1인가구 구조
    "single_age_20_gap",
    "single_age_30_gap",

    # I. 장애 관련 구조
    "disabled_per_1000",

    # J. 복지·경제 구조
    "welfare_vulnerability_pc1"
]


# ==========================================
# 행정동 + 최종 군집 + 대표 변수
# ==========================================

final_cluster_result = ml_feature_v4[
    ["행정동코드", "행정동명"] + representative_features
].copy()

final_cluster_result["cluster_k2"] = (
    cluster_compare["cluster_k2"].values
)


# 보기 편하도록 cluster를 앞쪽으로 이동
column_order = [
    "행정동코드",
    "행정동명",
    "cluster_k2"
] + representative_features

final_cluster_result = (
    final_cluster_result[column_order]
    .sort_values(
        ["cluster_k2", "행정동명"]
    )
    .reset_index(drop=True)
)


# ==========================================
# 기본 검증
# ==========================================

print("[DB1 최종 k=2 군집 결과]")
print("shape :", final_cluster_result.shape)
print("행정동 수 :", final_cluster_result["행정동코드"].nunique())
print("중복 key :", final_cluster_result["행정동코드"].duplicated().sum())
print("결측치 :", final_cluster_result.isna().sum().sum())

print("\n[군집별 행정동 수]")
print(
    final_cluster_result[
        "cluster_k2"
    ].value_counts().sort_index()
)

print("\n[군집별 행정동]")
for cluster in sorted(
    final_cluster_result["cluster_k2"].unique()
):
    dongs = final_cluster_result.loc[
        final_cluster_result["cluster_k2"] == cluster,
        "행정동명"
    ].tolist()

    print(
        f"Cluster {cluster} ({len(dongs)}개): "
        + ", ".join(dongs)
    )


display(
    final_cluster_result
)
```

```python
# ==========================================
# 1. 행정동별 최종 군집 결과 저장
# ==========================================

cluster_output_file = (
    "gangnam_db1_final_k2_cluster_result_2025.csv"
)

final_cluster_result.to_csv(
    cluster_output_file,
    index=False,
    encoding="utf-8-sig"
)


# ==========================================
# 2. 군집별 대표변수 프로파일 생성
# ==========================================

cluster_profile = (
    final_cluster_result
    .groupby("cluster_k2")[representative_features]
    .mean()
    .reset_index()
)

profile_output_file = (
    "gangnam_db1_final_k2_cluster_profile_2025.csv"
)

cluster_profile.to_csv(
    profile_output_file,
    index=False,
    encoding="utf-8-sig"
)


# ==========================================
# 검증
# ==========================================

print("[저장 완료]")
print(cluster_output_file, final_cluster_result.shape)
print(profile_output_file, cluster_profile.shape)

display(cluster_profile)
```
