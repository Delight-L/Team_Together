```python
import pandas as pd

months = ["202507", "202508", "202509", "202510", "202511", "202512"]

def read_skt(path):
    return pd.read_csv(
        path,
        sep="|",
        encoding="utf-8-sig",
        dtype={"BLOCK_CD": str},
        low_memory=False
    )

def load_skt_group(prefix):
    dfs = []

    for month in months:
        path = f"{prefix}_{month}.csv"
        df = read_skt(path)

        before = len(df)

        # 완전히 동일한 중복행만 제거
        df = df.drop_duplicates().copy()

        after = len(df)

        print(
            f"{month}: "
            f"{before:,} → {after:,} "
            f"(제거 {before-after:,})"
        )

        dfs.append(df)

    result = pd.concat(dfs, ignore_index=True)

    return result


print("[AGE]")
age_raw = load_skt_group("flow_age_pop")

print("\n[TIME]")
time_raw = load_skt_group("flow_time_pop")

print("\n[WKDY]")
wkdy_raw = load_skt_group("flow_wkdy_pop")


print("\n" + "=" * 70)
print("통합 결과")

for name, df in [
    ("AGE", age_raw),
    ("TIME", time_raw),
    ("WKDY", wkdy_raw)
]:
    print(
        name,
        "shape:", df.shape,
        "| 월:", sorted(df["STD_YM"].unique()),
        "| 결측:", df.isna().sum().sum(),
        "| 완전중복:", df.duplicated().sum()
    )
```

```python
import geopandas as gpd
import pandas as pd

# --------------------------------------------------
# 1. AGE 고유 좌표 생성
# --------------------------------------------------
age_points = (
    age_raw[["X_COORD", "Y_COORD"]]
    .drop_duplicates()
    .copy()
)

age_points_gdf = gpd.GeoDataFrame(
    age_points,
    geometry=gpd.points_from_xy(
        age_points["X_COORD"],
        age_points["Y_COORD"]
    ),
    crs="EPSG:5179"
)

print("AGE 고유 좌표:", len(age_points_gdf))


# --------------------------------------------------
# 2. 강남구 경계와 공간조인
# --------------------------------------------------
gangnam_boundary = gangnam_geo[
    ["adm_cd", "adm_nm", "geometry"]
].copy()

age_gangnam_map = gpd.sjoin(
    age_points_gdf,
    gangnam_boundary,
    how="inner",
    predicate="within"
)

print("강남구에 포함된 고유 좌표:", len(age_gangnam_map))


# --------------------------------------------------
# 3. 현재 행정동명 표준화
# --------------------------------------------------
age_gangnam_map["adm_cd"] = age_gangnam_map["adm_cd"].astype(str)

age_gangnam_map["dong_name"] = (
    age_gangnam_map["adm_nm"]
    .str.replace("서울특별시 강남구 ", "", regex=False)
)

# 2017 일원2동 → 현재 개포3동
age_gangnam_map.loc[
    age_gangnam_map["adm_cd"] == "1123074",
    "dong_name"
] = "개포3동"


# --------------------------------------------------
# 4. 22개 동 확인
# --------------------------------------------------
dong_check = (
    age_gangnam_map
    .groupby(["adm_cd", "dong_name"])
    .size()
    .reset_index(name="point_count")
    .sort_values("adm_cd")
)

print("\n행정동 수:", dong_check["adm_cd"].nunique())
display(dong_check)


# --------------------------------------------------
# 5. AGE 원본에 행정동 매핑
# --------------------------------------------------
age_coord_map = age_gangnam_map[
    ["X_COORD", "Y_COORD", "adm_cd", "dong_name"]
].copy()

age_gangnam = age_raw.merge(
    age_coord_map,
    on=["X_COORD", "Y_COORD"],
    how="inner",
    validate="many_to_one"
)

print("\n강남구 AGE 데이터 shape:", age_gangnam.shape)

print("\n월별 행 수:")
display(
    age_gangnam.groupby("STD_YM")
    .size()
    .reset_index(name="rows")
)

print("\n월별 행정동 수:")
display(
    age_gangnam.groupby("STD_YM")["adm_cd"]
    .nunique()
    .reset_index(name="dong_count")
)
```

```python
import numpy as np
import pandas as pd

male_cols = [
    "MAN_FLOW_POP_CNT_10G",
    "MAN_FLOW_POP_CNT_20G",
    "MAN_FLOW_POP_CNT_30G",
    "MAN_FLOW_POP_CNT_40G",
    "MAN_FLOW_POP_CNT_50G",
    "MAN_FLOW_POP_CNT_60GU"
]

female_cols = [
    "WMAN_FLOW_POP_CNT_10G",
    "WMAN_FLOW_POP_CNT_20G",
    "WMAN_FLOW_POP_CNT_30G",
    "WMAN_FLOW_POP_CNT_40G",
    "WMAN_FLOW_POP_CNT_50G",
    "WMAN_FLOW_POP_CNT_60GU"
]

age_groups = {
    "10": ["MAN_FLOW_POP_CNT_10G", "WMAN_FLOW_POP_CNT_10G"],
    "20": ["MAN_FLOW_POP_CNT_20G", "WMAN_FLOW_POP_CNT_20G"],
    "30": ["MAN_FLOW_POP_CNT_30G", "WMAN_FLOW_POP_CNT_30G"],
    "40": ["MAN_FLOW_POP_CNT_40G", "WMAN_FLOW_POP_CNT_40G"],
    "50": ["MAN_FLOW_POP_CNT_50G", "WMAN_FLOW_POP_CNT_50G"],
    "60_plus": ["MAN_FLOW_POP_CNT_60GU", "WMAN_FLOW_POP_CNT_60GU"]
}

# 행 단위 합계
tmp = age_gangnam.copy()

tmp["male_flow"] = tmp[male_cols].sum(axis=1)
tmp["female_flow"] = tmp[female_cols].sum(axis=1)
tmp["total_flow"] = tmp["male_flow"] + tmp["female_flow"]

for age, cols in age_groups.items():
    tmp[f"age_{age}_flow"] = tmp[cols].sum(axis=1)


# 월 × 행정동 집계
sum_cols = [
    "male_flow",
    "female_flow",
    "total_flow"
] + [f"age_{age}_flow" for age in age_groups]

age_month = (
    tmp.groupby(
        ["STD_YM", "adm_cd", "dong_name"],
        as_index=False
    )[sum_cols]
    .sum()
)


# 해당 월 실제 관측 격자 수
point_count = (
    tmp.groupby(
        ["STD_YM", "adm_cd", "dong_name"]
    )
    .size()
    .reset_index(name="point_count")
)

age_month = age_month.merge(
    point_count,
    on=["STD_YM", "adm_cd", "dong_name"],
    validate="one_to_one"
)


# 격자당 유동규모
age_month["flow_per_point"] = (
    age_month["total_flow"] /
    age_month["point_count"]
)


# 성별 구성비
age_month["male_ratio"] = (
    age_month["male_flow"] /
    age_month["total_flow"]
)

age_month["female_ratio"] = (
    age_month["female_flow"] /
    age_month["total_flow"]
)


# 연령 구성비
for age in age_groups:
    age_month[f"age_{age}_ratio"] = (
        age_month[f"age_{age}_flow"] /
        age_month["total_flow"]
    )


print("shape:", age_month.shape)
print("행정동 수:", age_month["adm_cd"].nunique())
print("월 수:", age_month["STD_YM"].nunique())
print(
    "월×동 key 중복:",
    age_month.duplicated(
        ["STD_YM", "adm_cd"]
    ).sum()
)
print("결측:", age_month.isna().sum().sum())


# 구성비 검증
age_month["sex_ratio_sum"] = (
    age_month["male_ratio"] +
    age_month["female_ratio"]
)

age_ratio_cols = [
    f"age_{age}_ratio"
    for age in age_groups
]

age_month["age_ratio_sum"] = (
    age_month[age_ratio_cols].sum(axis=1)
)

print(
    "\n성별 구성비 합 범위:",
    age_month["sex_ratio_sum"].min(),
    "~",
    age_month["sex_ratio_sum"].max()
)

print(
    "연령 구성비 합 범위:",
    age_month["age_ratio_sum"].min(),
    "~",
    age_month["age_ratio_sum"].max()
)


display(
    age_month[
        [
            "STD_YM", "adm_cd", "dong_name",
            "total_flow", "point_count",
            "flow_per_point",
            "male_ratio",
            *age_ratio_cols
        ]
    ].head()
)
```

```python
check_cols = [
    "total_flow",
    "point_count",
    "flow_per_point",
    "male_ratio",
    "age_10_ratio",
    "age_20_ratio",
    "age_30_ratio",
    "age_40_ratio",
    "age_50_ratio",
    "age_60_plus_ratio"
]

age_stability = (
    age_month
    .groupby(["adm_cd", "dong_name"])[check_cols]
    .agg(["mean", "std"])
)

# MultiIndex 컬럼 정리
age_stability.columns = [
    f"{col}_{stat}"
    for col, stat in age_stability.columns
]

age_stability = age_stability.reset_index()

# CV = 표준편차 / 평균
for col in check_cols:
    age_stability[f"{col}_cv"] = (
        age_stability[f"{col}_std"] /
        age_stability[f"{col}_mean"]
    )

print("shape:", age_stability.shape)

# 전체 22개 동의 CV 요약
cv_cols = [f"{c}_cv" for c in check_cols]

cv_summary = (
    age_stability[cv_cols]
    .agg(["mean", "median", "max"])
    .T
    .sort_values("mean", ascending=False)
)

print("\n[Feature별 6개월 변동성]")
display(cv_summary)


# point_count가 가장 많이 변한 동
print("\n[point_count 변동성이 큰 행정동]")
display(
    age_stability[
        [
            "adm_cd",
            "dong_name",
            "point_count_mean",
            "point_count_std",
            "point_count_cv"
        ]
    ]
    .sort_values("point_count_cv", ascending=False)
    .head(10)
)


# total_flow와 flow_per_point 비교
print("\n[total_flow vs flow_per_point]")
display(
    age_stability[
        [
            "dong_name",
            "total_flow_mean",
            "total_flow_cv",
            "flow_per_point_mean",
            "flow_per_point_cv"
        ]
    ]
    .sort_values("total_flow_cv", ascending=False)
)
```

```python
import numpy as np
import pandas as pd

hour_cols = [f"TMST_{h:02d}" for h in range(24)]

# 시간구간
night_cols   = [f"TMST_{h:02d}" for h in range(0, 7)]    # 00~06
morning_cols = [f"TMST_{h:02d}" for h in range(7, 12)]   # 07~11
daytime_cols = [f"TMST_{h:02d}" for h in range(12, 19)]  # 12~18
evening_cols = [f"TMST_{h:02d}" for h in range(19, 24)]  # 19~23


# 1. 월 × 행정동별 24시간 합계
time_month = (
    time_gangnam
    .groupby(
        ["STD_YM", "adm_cd", "dong_name"],
        as_index=False
    )[hour_cols]
    .sum()
)


# 2. 전체 시간 활동량
time_month["time_total"] = time_month[hour_cols].sum(axis=1)


# 3. 시간구간별 활동량
time_month["night_flow"] = time_month[night_cols].sum(axis=1)
time_month["morning_flow"] = time_month[morning_cols].sum(axis=1)
time_month["daytime_flow"] = time_month[daytime_cols].sum(axis=1)
time_month["evening_flow"] = time_month[evening_cols].sum(axis=1)


# 4. 구성비
for name in ["night", "morning", "daytime", "evening"]:
    time_month[f"{name}_ratio"] = (
        time_month[f"{name}_flow"] /
        time_month["time_total"]
    )


# 5. 피크 시간
time_month["peak_hour"] = (
    time_month[hour_cols]
    .idxmax(axis=1)
    .str.replace("TMST_", "", regex=False)
    .astype(int)
)


# 6. 구성비 검증
time_month["time_ratio_sum"] = (
    time_month[
        [
            "night_ratio",
            "morning_ratio",
            "daytime_ratio",
            "evening_ratio"
        ]
    ].sum(axis=1)
)


print("shape:", time_month.shape)
print("행정동 수:", time_month["adm_cd"].nunique())
print("월 수:", time_month["STD_YM"].nunique())

print(
    "월×동 key 중복:",
    time_month.duplicated(
        ["STD_YM", "adm_cd"]
    ).sum()
)

print("결측:", time_month.isna().sum().sum())

print(
    "시간대 구성비 합 범위:",
    time_month["time_ratio_sum"].min(),
    "~",
    time_month["time_ratio_sum"].max()
)


print("\n[피크 시간 분포]")
display(
    time_month["peak_hour"]
    .value_counts()
    .sort_index()
    .rename_axis("peak_hour")
    .reset_index(name="count")
)


display(
    time_month[
        [
            "STD_YM",
            "adm_cd",
            "dong_name",
            "time_total",
            "night_ratio",
            "morning_ratio",
            "daytime_ratio",
            "evening_ratio",
            "peak_hour"
        ]
    ].head()
)
```

```python
import pandas as pd
import numpy as np

# -----------------------------------------
# 1. 실제 연령구간명
# -----------------------------------------

age_labels = resident_raw.iloc[0, 3:25].tolist()

print("[연령구간]")
print(age_labels)
print("개수:", len(age_labels))


# -----------------------------------------
# 2. 강남구 22개 동만 선택
# -----------------------------------------

dong_rows = resident_raw[
    (resident_raw["동별(1)"] == "강남구") &
    (resident_raw["동별(2)"] != "소계")
].copy()

print("\n강남구 동 데이터 행 수:", len(dong_rows))
print("행정동 수:", dong_rows["동별(2)"].nunique())

print("\n항목별 행 수:")
print(dong_rows["항목"].value_counts())


# -----------------------------------------
# 3. 분기별 변환 함수
# -----------------------------------------

def make_resident_quarter(df, quarter, start_idx, end_idx):

    cols = df.columns[start_idx:end_idx]

    out = df[
        ["동별(2)", "항목"] + list(cols)
    ].copy()

    new_names = ["dong_name", "category"]

    for label in age_labels:
        if label == "합계":
            new_names.append("resident_total")
        else:
            clean = (
                str(label)
                .replace("~", "_")
                .replace("세 이상", "_plus")
                .replace("세", "")
            )

            new_names.append(f"age_{clean}")

    out.columns = new_names

    out["quarter"] = quarter

    # '-' → 0
    value_cols = [
        c for c in out.columns
        if c not in ["dong_name", "category", "quarter"]
    ]

    for col in value_cols:
        out[col] = pd.to_numeric(
            out[col].replace("-", 0),
            errors="coerce"
        )

    return out


# Q3 : columns 3~24
resident_q3 = make_resident_quarter(
    dong_rows,
    "2025Q3",
    3,
    25
)

# Q4 : columns 25~46
resident_q4 = make_resident_quarter(
    dong_rows,
    "2025Q4",
    25,
    47
)


# -----------------------------------------
# 4. 통합
# -----------------------------------------

resident_quarter = pd.concat(
    [resident_q3, resident_q4],
    ignore_index=True
)

print("\n[통합 결과]")
print("shape:", resident_quarter.shape)

print(
    "행정동 수:",
    resident_quarter["dong_name"].nunique()
)

print(
    "분기:",
    resident_quarter["quarter"].unique()
)

print(
    "category:",
    resident_quarter["category"].unique()
)

print(
    "key 중복:",
    resident_quarter.duplicated(
        ["dong_name", "quarter", "category"]
    ).sum()
)

print(
    "결측:",
    resident_quarter.isna().sum().sum()
)

display(resident_quarter.head(10))
```

```python
import pandas as pd
import numpy as np

# 강남구 소계를 제외한 22개 행정동
dong_source = household_raw.iloc[2:].copy()

hh_names = [
    "household_total",
    "hh_1",
    "hh_2",
    "hh_3",
    "hh_4",
    "hh_5",
    "hh_6",
    "hh_7",
    "hh_8",
    "hh_9",
    "hh_10_plus"
]

# -----------------------------------------
# Q3
# -----------------------------------------
q3 = dong_source.iloc[:, [1] + list(range(2, 13))].copy()

q3.columns = [
    "dong_name"
] + hh_names

q3["quarter"] = "2025Q3"


# -----------------------------------------
# Q4
# -----------------------------------------
q4 = dong_source.iloc[:, [1] + list(range(13, 24))].copy()

q4.columns = [
    "dong_name"
] + hh_names

q4["quarter"] = "2025Q4"


# -----------------------------------------
# 통합
# -----------------------------------------
household_quarter = pd.concat(
    [q3, q4],
    ignore_index=True
)

numeric_cols = hh_names

household_quarter[numeric_cols] = (
    household_quarter[numeric_cols]
    .apply(pd.to_numeric, errors="coerce")
)

# 컬럼 순서
household_quarter = household_quarter[
    ["dong_name", "quarter"] + hh_names
]


# -----------------------------------------
# 기본 검증
# -----------------------------------------
print("[가구구조 분기 테이블]")
print("shape:", household_quarter.shape)
print(
    "행정동 수:",
    household_quarter["dong_name"].nunique()
)
print(
    "분기:",
    household_quarter["quarter"].unique()
)
print(
    "key 중복:",
    household_quarter.duplicated(
        ["dong_name", "quarter"]
    ).sum()
)
print(
    "결측:",
    household_quarter.isna().sum().sum()
)


# -----------------------------------------
# 전체세대 = 세대원수별 세대 합계 검증
# -----------------------------------------
hh_size_cols = [
    "hh_1", "hh_2", "hh_3", "hh_4", "hh_5",
    "hh_6", "hh_7", "hh_8", "hh_9", "hh_10_plus"
]

household_quarter["size_sum"] = (
    household_quarter[hh_size_cols]
    .sum(axis=1)
)

household_quarter["difference"] = (
    household_quarter["household_total"]
    - household_quarter["size_sum"]
)

print("\n[합계 정합성]")
print(
    "불일치 행:",
    (household_quarter["difference"] != 0).sum()
)

print(
    "difference 범위:",
    household_quarter["difference"].min(),
    "~",
    household_quarter["difference"].max()
)

# 분기별 전체 세대 / 1인세대 합계도 확인
print("\n[분기별 강남구 합계]")
display(
    household_quarter
    .groupby("quarter")[
        ["household_total", "hh_1"]
    ]
    .sum()
)

print("\n[샘플]")
display(
    household_quarter.head(10)
)
```

```python
# --------------------------------------------------
# 1. 가구구조 4개 범주 생성
# --------------------------------------------------

household_feature_q = household_quarter[
    ["dong_name", "quarter", "household_total"]
].copy()

household_feature_q["hh_1_ratio"] = (
    household_quarter["hh_1"]
    / household_quarter["household_total"]
)

household_feature_q["hh_2_ratio"] = (
    household_quarter["hh_2"]
    / household_quarter["household_total"]
)

household_feature_q["hh_3_ratio"] = (
    household_quarter["hh_3"]
    / household_quarter["household_total"]
)

household_feature_q["hh_4plus_ratio"] = (
    (
        household_quarter[
            [
                "hh_4", "hh_5", "hh_6", "hh_7",
                "hh_8", "hh_9", "hh_10_plus"
            ]
        ].sum(axis=1)
    )
    / household_quarter["household_total"]
)


# --------------------------------------------------
# 2. 구성비 합계 검증
# --------------------------------------------------

ratio_cols = [
    "hh_1_ratio",
    "hh_2_ratio",
    "hh_3_ratio",
    "hh_4plus_ratio"
]

household_feature_q["ratio_sum"] = (
    household_feature_q[ratio_cols].sum(axis=1)
)

print("[가구구조 Feature]")
print("shape:", household_feature_q.shape)
print(
    "구성비 합계 최대 오차:",
    (household_feature_q["ratio_sum"] - 1).abs().max()
)

display(household_feature_q.head())


# --------------------------------------------------
# 3. Q3 → Q4 구성비 변화 (%p)
# --------------------------------------------------

q3 = (
    household_feature_q[
        household_feature_q["quarter"] == "2025Q3"
    ]
    .set_index("dong_name")[ratio_cols]
)

q4 = (
    household_feature_q[
        household_feature_q["quarter"] == "2025Q4"
    ]
    .set_index("dong_name")[ratio_cols]
)

hh_diff_pp = (q4 - q3) * 100

print("\n[Q4 - Q3 구성비 변화 (%p)]")
display(hh_diff_pp.round(3))


# --------------------------------------------------
# 4. 변수별 안정성
# --------------------------------------------------

hh_stability = pd.DataFrame({
    "mean_abs_change_pp":
        hh_diff_pp.abs().mean(),

    "median_abs_change_pp":
        hh_diff_pp.abs().median(),

    "max_abs_change_pp":
        hh_diff_pp.abs().max()
}).sort_values(
    "mean_abs_change_pp",
    ascending=False
)

print("\n[변수별 절대 변화 요약]")
display(hh_stability.round(3))


# --------------------------------------------------
# 5. 행정동별 변화 크기
# --------------------------------------------------

hh_dong_stability = pd.DataFrame({
    "mean_abs_change_pp":
        hh_diff_pp.abs().mean(axis=1),

    "max_abs_change_pp":
        hh_diff_pp.abs().max(axis=1)
}).sort_values(
    "max_abs_change_pp",
    ascending=False
)

print("\n[행정동별 변화 크기]")
display(hh_dong_stability.round(3))
```

```python
# --------------------------------------------------
# 1. 장애유형별 동별 인원
#    성별='계', 장애유형='합계' 제외
# --------------------------------------------------

disability_type = (
    disability_check[
        (disability_check["성별"] == "계")
        & (disability_check["장애유형별"] != "합계")
    ][["동별", "장애유형별", "value"]]
    .copy()
)

# wide 형태
disability_type_wide = (
    disability_type
    .pivot(
        index="동별",
        columns="장애유형별",
        values="value"
    )
    .reset_index()
    .rename(columns={"동별": "dong_name"})
)

# 전체 장애인 수 결합
disability_type_wide = disability_type_wide.merge(
    disability_base[
        ["dong_name", "disability_total"]
    ],
    on="dong_name",
    how="left",
    validate="one_to_one"
)

type_cols = [
    c for c in disability_type_wide.columns
    if c not in ["dong_name", "disability_total"]
]


# --------------------------------------------------
# 2. 장애유형 구성비
# --------------------------------------------------

for col in type_cols:
    disability_type_wide[f"{col}_ratio"] = (
        disability_type_wide[col]
        / disability_type_wide["disability_total"]
    )

type_ratio_cols = [
    f"{col}_ratio"
    for col in type_cols
]

print("[장애유형 구성]")
print("shape:", disability_type_wide.shape)
print("장애유형 수:", len(type_cols))
print("결측:", disability_type_wide.isna().sum().sum())

print(
    "구성비 합계 최대 오차:",
    (
        disability_type_wide[type_ratio_cols]
        .sum(axis=1) - 1
    ).abs().max()
)


# --------------------------------------------------
# 3. 유형별 희소성 및 분포
# --------------------------------------------------

type_summary = pd.DataFrame({
    "total_count":
        disability_type_wide[type_cols].sum(),

    "zero_dong_count":
        (disability_type_wide[type_cols] == 0).sum(),

    "mean_ratio":
        disability_type_wide[type_ratio_cols]
        .mean()
        .set_axis(type_cols),

    "std_ratio":
        disability_type_wide[type_ratio_cols]
        .std()
        .set_axis(type_cols),

    "min_ratio":
        disability_type_wide[type_ratio_cols]
        .min()
        .set_axis(type_cols),

    "max_ratio":
        disability_type_wide[type_ratio_cols]
        .max()
        .set_axis(type_cols)
})

type_summary["mean_ratio_pct"] = (
    type_summary["mean_ratio"] * 100
)

type_summary["std_ratio_pct"] = (
    type_summary["std_ratio"] * 100
)

type_summary["max_ratio_pct"] = (
    type_summary["max_ratio"] * 100
)

print("\n[장애유형별 분포]")
display(
    type_summary[
        [
            "total_count",
            "zero_dong_count",
            "mean_ratio_pct",
            "std_ratio_pct",
            "max_ratio_pct"
        ]
    ]
    .sort_values(
        "mean_ratio_pct",
        ascending=False
    )
    .round(3)
)
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
import numpy as np
import pandas as pd

# --------------------------------------------------
# 1. 장애유형 재구성
# --------------------------------------------------

main_types = [
    "지체",
    "청각",
    "뇌병변",
    "시각",
    "신장장애",
    "지적장애",
    "자폐성",
    "정신장애",
    "간"
]

rare_types = [
    "언어",
    "호흡기",
    "심장장애",
    "안면",
    "장루요루",
    "뇌전증"
]

disability_grouped = disability_type_wide[
    ["dong_name", "disability_total"] + main_types
].copy()

disability_grouped["기타희소"] = (
    disability_type_wide[rare_types].sum(axis=1)
)

group_cols = main_types + ["기타희소"]


# --------------------------------------------------
# 2. 구성비 변환
# --------------------------------------------------

ratio_group_cols = []

for col in group_cols:
    ratio_col = f"{col}_ratio"

    disability_grouped[ratio_col] = (
        disability_grouped[col]
        / disability_grouped["disability_total"]
    )

    ratio_group_cols.append(ratio_col)


print("[재구성된 장애유형]")
print("범주 수:", len(group_cols))
print(
    "구성비 합계 최대 오차:",
    (
        disability_grouped[ratio_group_cols]
        .sum(axis=1) - 1
    ).abs().max()
)


# --------------------------------------------------
# 3. PCA
# --------------------------------------------------

X_disability_type = StandardScaler().fit_transform(
    disability_grouped[ratio_group_cols]
)

pca_disability = PCA().fit(X_disability_type)

disability_pca_result = pd.DataFrame({
    "PC": [
        f"PC{i+1}"
        for i in range(len(group_cols))
    ],
    "explained_ratio":
        pca_disability.explained_variance_ratio_,
    "cumulative_ratio":
        np.cumsum(
            pca_disability.explained_variance_ratio_
        )
})

print("\n[장애유형 PCA 설명력]")
display(disability_pca_result)


# --------------------------------------------------
# 4. PC1~PC3 loading
# --------------------------------------------------

disability_loading = pd.DataFrame(
    pca_disability.components_.T,
    index=group_cols,
    columns=[
        f"PC{i+1}"
        for i in range(len(group_cols))
    ]
)

print("\n[PC1~PC3 Loading]")
display(
    disability_loading[
        ["PC1", "PC2", "PC3"]
    ].round(4)
)


# --------------------------------------------------
# 5. 상관관계
# --------------------------------------------------

print("\n[구성비 상관관계]")
display(
    disability_grouped[
        ratio_group_cols
    ].corr().round(3)
)
```

```python
# 강남구 시작점
gangnam_start = min(gangnam_idx)

# 강남구 이후 자치구 행 탐색
after_gangnam = welfare.loc[gangnam_start:].copy()

district_rows = (
    after_gangnam[
        after_gangnam["area"].astype(str).str.endswith("구")
    ][["area"]]
    .copy()
)

district_rows = district_rows[
    district_rows["area"].ne(district_rows["area"].shift())
]

print("[강남구 이후 자치구 순서]")
display(district_rows.head(5))

# 다음 자치구 시작점
next_district_start = (
    district_rows[
        district_rows["area"] != "강남구"
    ].index.min()
)

print("다음 자치구 시작 index:", next_district_start)
print("다음 자치구:", welfare.loc[next_district_start, "area"])

# 강남구 블록
welfare_gangnam_block = welfare.loc[
    gangnam_start:next_district_start - 1
].copy()

print("\n[강남구 전체 블록]")
print("shape:", welfare_gangnam_block.shape)
print(welfare_gangnam_block["area"].unique())

# 강남구 22개 동만
welfare_gangnam_dong = welfare_gangnam_block[
    welfare_gangnam_block["area"].isin(gangnam_dongs)
].copy()

print("\n[강남구 행정동]")
print("shape:", welfare_gangnam_dong.shape)
print("행정동 수:", welfare_gangnam_dong["area"].nunique())

display(
    welfare_gangnam_dong.groupby("area")
    .size()
    .sort_values()
    .to_frame("row_count")
)
```

```python
# --------------------------------------------------
# 1. 강남구 전체 블록의 자격 × 연령 조합
# --------------------------------------------------

qualification_age_structure = (
    welfare_gangnam_block[
        ["qualification", "age_group"]
    ]
    .drop_duplicates()
    .sort_values(
        ["qualification", "age_group"]
    )
    .reset_index(drop=True)
)

print("[강남구 블록에 존재하는 자격 × 연령 조합]")
display(qualification_age_structure)

print(
    "조합 수:",
    len(qualification_age_structure)
)


# --------------------------------------------------
# 2. 자격별 존재하는 연령구간
# --------------------------------------------------

print("\n[자격별 연령구간]")
display(
    welfare_gangnam_block
    .groupby("qualification")["age_group"]
    .unique()
    .to_frame("age_groups")
)


# --------------------------------------------------
# 3. 각 동에 어떤 조합이 빠져 있는지 확인
# --------------------------------------------------

combination_count = (
    welfare_gangnam_dong
    .groupby("area")
    .size()
    .rename("observed_combinations")
    .reset_index()
)

display(
    combination_count
    .sort_values("observed_combinations")
)


# --------------------------------------------------
# 4. 가장 행 수가 적은 대치1동 원자료 확인
# --------------------------------------------------

print("\n[대치1동]")
display(
    welfare_gangnam_dong[
        welfare_gangnam_dong["area"] == "대치1동"
    ].sort_values(
        ["qualification", "age_group"]
    )
)


# --------------------------------------------------
# 5. 행 수가 많은 역삼1동과 비교
# --------------------------------------------------

print("\n[역삼1동]")
display(
    welfare_gangnam_dong[
        welfare_gangnam_dong["area"] == "역삼1동"
    ].sort_values(
        ["qualification", "age_group"]
    )
)
```

```python
from itertools import product
import pandas as pd

# --------------------------------------------------
# 1. 원자료에서 가능한 11개 조합 정의
# --------------------------------------------------

valid_combinations = (
    welfare_gangnam_block[
        ["qualification", "age_group"]
    ]
    .drop_duplicates()
    .reset_index(drop=True)
)

print("[가능한 급여 × 연령 조합]")
display(valid_combinations)
print("조합 수:", len(valid_combinations))


# --------------------------------------------------
# 2. 22개 동 × 11개 조합 완전 격자 생성
# --------------------------------------------------

full_grid = pd.MultiIndex.from_product(
    [
        gangnam_dongs,
        valid_combinations["qualification"].unique(),
        valid_combinations["age_group"].unique()
    ],
    names=["area", "qualification", "age_group"]
).to_frame(index=False)

# 실제 존재하는 11개 조합만 남김
full_grid = full_grid.merge(
    valid_combinations,
    on=["qualification", "age_group"],
    how="inner"
)

print("\n[완전 격자]")
print("shape:", full_grid.shape)   # 22 × 11 = 242


# --------------------------------------------------
# 3. 실제 자료와 결합 → 누락 조합 0 처리
# --------------------------------------------------

welfare_complete = full_grid.merge(
    welfare_gangnam_dong,
    on=["area", "qualification", "age_group"],
    how="left",
    validate="one_to_one"
)

missing_before = welfare_complete[
    "recipient_count"
].isna().sum()

print("0 처리 전 누락 조합:", missing_before)

welfare_complete["recipient_count"] = (
    welfare_complete["recipient_count"]
    .fillna(0)
)

print(
    "0 처리 후 결측:",
    welfare_complete["recipient_count"].isna().sum()
)


# --------------------------------------------------
# 4. 22개 동 합계 계산
# --------------------------------------------------

dong_sum = (
    welfare_complete
    .groupby(
        ["qualification", "age_group"],
        as_index=False
    )["recipient_count"]
    .sum()
    .rename(
        columns={
            "recipient_count": "dong_sum"
        }
    )
)


# --------------------------------------------------
# 5. 원자료 강남구 합계
# --------------------------------------------------

gangnam_total = (
    welfare_gangnam_block[
        welfare_gangnam_block["area"] == "강남구"
    ][
        ["qualification", "age_group", "recipient_count"]
    ]
    .rename(
        columns={
            "recipient_count": "gangnam_total"
        }
    )
)


# --------------------------------------------------
# 6. 비교
# --------------------------------------------------

validation = valid_combinations.merge(
    gangnam_total,
    on=["qualification", "age_group"],
    how="left"
).merge(
    dong_sum,
    on=["qualification", "age_group"],
    how="left"
)

validation["difference"] = (
    validation["dong_sum"]
    - validation["gangnam_total"]
)

print("\n[강남구 합계 검증]")
display(
    validation.sort_values(
        ["qualification", "age_group"]
    )
)

print(
    "\n모든 조합 정확히 일치:",
    (validation["difference"] == 0).all()
)

print(
    "최대 절대 차이:",
    validation["difference"].abs().max()
)
```

```python
# --------------------------------------------------
# 1. Analysis 1 Feature 영역 정의
# --------------------------------------------------

domain_features = {
    "A_유동활동규모": [
        "activity_intensity"
    ],

    "B_유동인구구조": [
        "active_pop_pc1",
        "active_pop_pc2"
    ],

    "C_시간요일활동": [
        "time_structure_pc1",
        "time_structure_pc2",
        "weekend_weekday_index"
    ],

    "D_주민인구구조": [
        "resident_age_pc1",
        "resident_age_pc2",
        "foreigner_ratio"
    ],

    "E_가구구조": [
        "household_structure_pc1",
        "household_structure_pc2"
    ],

    "F_장애구조": [
        "disability_ratio"
    ],

    "G_복지경제구조": [
        "livelihood_recipient_ratio"
    ]
}

feature_cols = [
    col
    for cols in domain_features.values()
    for col in cols
]

print("[군집 후보 Feature]")
print("개수:", len(feature_cols))
print(feature_cols)


# --------------------------------------------------
# 2. 전체 상관계수
# --------------------------------------------------

corr = (
    analysis1_feature[feature_cols]
    .corr()
)

print("\n[상관계수 행렬]")
display(corr.round(3))


# --------------------------------------------------
# 3. |r| >= 0.8 변수쌍 추출
# --------------------------------------------------

high_corr_pairs = []

for i in range(len(feature_cols)):
    for j in range(i + 1, len(feature_cols)):

        r = corr.iloc[i, j]

        if abs(r) >= 0.8:
            high_corr_pairs.append({
                "feature_1": feature_cols[i],
                "feature_2": feature_cols[j],
                "correlation": r
            })

high_corr_df = pd.DataFrame(high_corr_pairs)

if len(high_corr_df) > 0:
    high_corr_df["abs_correlation"] = (
        high_corr_df["correlation"].abs()
    )

    high_corr_df = (
        high_corr_df
        .sort_values(
            "abs_correlation",
            ascending=False
        )
        .reset_index(drop=True)
    )

print("\n[|r| >= 0.8 변수쌍]")
display(high_corr_df)
```

```python
from sklearn.preprocessing import StandardScaler
import pandas as pd
import numpy as np

# --------------------------------------------------
# 1. 13개 Feature Z-score
# --------------------------------------------------

scaler_diag = StandardScaler()

Z = pd.DataFrame(
    scaler_diag.fit_transform(
        analysis1_feature[feature_cols]
    ),
    columns=feature_cols,
    index=analysis1_feature["dong_name"]
)

print("[|Z| >= 2인 값]")

extreme_rows = []

for dong in Z.index:
    for feature in Z.columns:
        z = Z.loc[dong, feature]

        if abs(z) >= 2:
            extreme_rows.append({
                "dong_name": dong,
                "feature": feature,
                "z_score": z
            })

extreme_df = pd.DataFrame(extreme_rows)

if len(extreme_df) > 0:
    extreme_df["abs_z"] = extreme_df["z_score"].abs()
    extreme_df = extreme_df.sort_values(
        "abs_z",
        ascending=False
    )

display(extreme_df)


# --------------------------------------------------
# 2. 동별 극단값 개수
# --------------------------------------------------

print("\n[동별 |Z| >= 2 개수]")

extreme_count = (
    (Z.abs() >= 2)
    .sum(axis=1)
    .sort_values(ascending=False)
    .to_frame("extreme_feature_count")
)

display(extreme_count)


# --------------------------------------------------
# 3. F-G 상관관계 Leave-One-Out
# --------------------------------------------------

loo_corr = []

for dong in analysis1_feature["dong_name"]:

    temp = analysis1_feature[
        analysis1_feature["dong_name"] != dong
    ]

    r = temp[
        [
            "disability_ratio",
            "livelihood_recipient_ratio"
        ]
    ].corr().iloc[0, 1]

    loo_corr.append({
        "excluded_dong": dong,
        "correlation": r
    })

loo_corr_df = (
    pd.DataFrame(loo_corr)
    .sort_values("correlation")
)

print("\n[F-G Leave-One-Out 상관계수]")
display(loo_corr_df)

print(
    "\n전체 상관계수:",
    analysis1_feature[
        [
            "disability_ratio",
            "livelihood_recipient_ratio"
        ]
    ].corr().iloc[0, 1]
)


# --------------------------------------------------
# 4. 주요 고상관 변수쌍도 수서동 제외 시 확인
# --------------------------------------------------

without_suseo = analysis1_feature[
    analysis1_feature["dong_name"] != "수서동"
]

pairs_to_check = [
    ("disability_ratio",
     "livelihood_recipient_ratio"),

    ("resident_age_pc1",
     "household_structure_pc1"),

    ("resident_age_pc2",
     "household_structure_pc2"),

    ("resident_age_pc1",
     "foreigner_ratio"),

    ("active_pop_pc1",
     "foreigner_ratio"),

    ("resident_age_pc2",
     "disability_ratio")
]

comparison = []

for f1, f2 in pairs_to_check:

    full_r = analysis1_feature[
        [f1, f2]
    ].corr().iloc[0, 1]

    no_suseo_r = without_suseo[
        [f1, f2]
    ].corr().iloc[0, 1]

    comparison.append({
        "feature_1": f1,
        "feature_2": f2,
        "full_r": full_r,
        "without_suseo_r": no_suseo_r,
        "change": no_suseo_r - full_r
    })

print("\n[수서동 영향 비교]")
display(
    pd.DataFrame(comparison).round(4)
)
```

```python
from scipy.spatial.distance import pdist, squareform
from scipy.stats import spearmanr
import numpy as np
import pandas as pd

# --------------------------------------------------
# 1. 5개 상위 영역 정의
# --------------------------------------------------

super_domains = {
    "A_활동규모": [
        "activity_intensity"
    ],

    "B_생활활동구조": [
        "active_pop_pc1",
        "active_pop_pc2",
        "time_structure_pc1",
        "time_structure_pc2",
        "weekend_weekday_index"
    ],

    "C_인구가구구조": [
        "resident_age_pc1",
        "resident_age_pc2",
        "foreigner_ratio",
        "household_structure_pc1",
        "household_structure_pc2"
    ],

    "D_장애구조": [
        "disability_ratio"
    ],

    "E_경제복지구조": [
        "livelihood_recipient_ratio"
    ]
}


# --------------------------------------------------
# 2. 5개 상위 영역 균형화
# --------------------------------------------------

Z_super_balanced = Z_all.copy()

for domain, cols in super_domains.items():

    weight = 1 / np.sqrt(len(cols))

    Z_super_balanced[cols] = (
        Z_super_balanced[cols] * weight
    )


# --------------------------------------------------
# 3. 기여도 검증
# --------------------------------------------------

super_contribution = []

for domain, cols in super_domains.items():

    dist2 = (
        Z_super_balanced[cols] ** 2
    ).sum(axis=1)

    super_contribution.append({
        "domain": domain,
        "n_features": len(cols),
        "total_squared_distance": dist2.sum()
    })

super_contribution = pd.DataFrame(
    super_contribution
)

super_contribution["contribution_ratio"] = (
    super_contribution["total_squared_distance"]
    / super_contribution["total_squared_distance"].sum()
)

print("[5개 상위 영역 기여도]")
display(super_contribution)


# --------------------------------------------------
# 4. 7-domain vs 5-domain 거리행렬
# --------------------------------------------------

dist_7 = pdist(
    Z_balanced.values,
    metric="euclidean"
)

dist_5 = pdist(
    Z_super_balanced.values,
    metric="euclidean"
)


# Pearson
pearson_distance_corr = np.corrcoef(
    dist_7,
    dist_5
)[0, 1]


# Spearman
spearman_distance_corr = spearmanr(
    dist_7,
    dist_5
).statistic


print(
    "\n7-domain vs 5-domain 거리 Pearson:",
    round(pearson_distance_corr, 4)
)

print(
    "7-domain vs 5-domain 거리 Spearman:",
    round(spearman_distance_corr, 4)
)


# --------------------------------------------------
# 5. 각 동의 평균 거리 변화
# --------------------------------------------------

D7 = squareform(dist_7)
D5 = squareform(dist_5)

distance_change = pd.DataFrame({
    "dong_name": analysis1_feature["dong_name"],
    "mean_distance_7":
        D7.mean(axis=1),
    "mean_distance_5":
        D5.mean(axis=1)
})

distance_change["change_ratio"] = (
    distance_change["mean_distance_5"]
    / distance_change["mean_distance_7"]
)

print("\n[동별 평균거리 변화]")

display(
    distance_change
    .sort_values(
        "change_ratio",
        ascending=False
    )
)
```

```python
from sklearn.cluster import KMeans
from sklearn.metrics import adjusted_rand_score
import numpy as np
import pandas as pd

X = Z_super_balanced.values

# --------------------------------------------------
# 1. 기준 군집
# --------------------------------------------------

base_labels = {}

for k in [2, 3]:

    model = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=100
    )

    base_labels[k] = model.fit_predict(X)


# --------------------------------------------------
# 2. 작은 잡음 perturbation
# --------------------------------------------------

rng = np.random.default_rng(42)

noise_levels = [
    0.05,
    0.10,
    0.20
]

stability_rows = []

for k in [2, 3]:

    for noise_sd in noise_levels:

        ari_scores = []

        for i in range(500):

            X_noise = (
                X
                + rng.normal(
                    0,
                    noise_sd,
                    size=X.shape
                )
            )

            labels_noise = KMeans(
                n_clusters=k,
                random_state=i,
                n_init=50
            ).fit_predict(X_noise)

            ari = adjusted_rand_score(
                base_labels[k],
                labels_noise
            )

            ari_scores.append(ari)

        stability_rows.append({
            "k": k,
            "noise_sd": noise_sd,
            "mean_ARI": np.mean(ari_scores),
            "median_ARI": np.median(ari_scores),
            "min_ARI": np.min(ari_scores),
            "ARI_ge_0.8":
                np.mean(
                    np.array(ari_scores) >= 0.8
                ),
            "exact_ARI_1":
                np.mean(
                    np.isclose(
                        ari_scores,
                        1.0
                    )
                )
        })


stability_result = pd.DataFrame(
    stability_rows
)

print("[Perturbation 안정성]")
display(stability_result)
```
