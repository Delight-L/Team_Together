```python
import pandas as pd
from google.colab import files

# 파일 업로드
uploaded = files.upload()

# 실제 업로드된 파일명 자동 확인
file_name = list(uploaded.keys())[0]

print("업로드된 파일:", file_name)
```

```python
card_df = pd.read_csv(
    file_name,
    encoding="utf-8"
)

print("전체 데이터 크기:", card_df.shape)
display(card_df.head())
```

```python
# 불필요한 인덱스 컬럼 제거
card_df = card_df.drop(columns=["Unnamed: 0"], errors="ignore")

# 결제 발생지가 강남구인 데이터만 추출
gangnam_card = card_df[
    card_df["MCT_SGG_CD"] == "서울 강남구"
].copy()

gangnam_card = gangnam_card.reset_index(drop=True)

print("전체 데이터 수:", len(card_df))
print("강남구 데이터 수:", len(gangnam_card))
print("강남구 데이터 비율:",
      round(len(gangnam_card) / len(card_df) * 100, 2), "%")

display(gangnam_card.head())
```

```python
print("MCT_SGG_CD 고유값:")
print(gangnam_card["MCT_SGG_CD"].unique())

print("\n강남구 데이터 크기:")
print(gangnam_card.shape)
```

```python
# 1. 날짜 범위
print("[날짜]")
print("최초:", gangnam_card["TA_YMD"].min())
print("최종:", gangnam_card["TA_YMD"].max())
print("날짜 수:", gangnam_card["TA_YMD"].nunique())

# 2. 업종
print("\n[업종]")
print("업종 수:", gangnam_card["MCT_RY_CD"].nunique())
print(gangnam_card["MCT_RY_CD"].value_counts())

# 3. 결제자의 본거주지
print("\n[본거주지]")
print("거주지역 수:", gangnam_card["CLN_SGG_CD"].nunique())
print(gangnam_card["CLN_SGG_CD"].value_counts())

# 4. 성별
print("\n[성별]")
print(gangnam_card["SEX_CCD"].value_counts())

# 5. 연령
print("\n[연령]")
print(gangnam_card["AGE_CCD"].value_counts())

# 6. 결제금액/건수 기본 통계
print("\n[TS_AT]")
print(gangnam_card["TS_AT"].describe())

print("\n[USE_CNT]")
print(gangnam_card["USE_CNT"].describe())
```

```python
# 날짜 구조
print("[날짜]")
print("최초:", gangnam_card["TA_YMD"].min())
print("최종:", gangnam_card["TA_YMD"].max())
print("날짜 수:", gangnam_card["TA_YMD"].nunique())

# 월별 날짜 수
temp_date = pd.to_datetime(
    gangnam_card["TA_YMD"].astype(str),
    format="%Y%m%d"
)

print("\n[월별 날짜 수]")
print(
    temp_date
    .dt.to_period("M")
    .value_counts()
    .sort_index()
)

# 업종 구조
print("\n[업종]")
print("업종 수:", gangnam_card["MCT_RY_CD"].nunique())
print(gangnam_card["MCT_RY_CD"].value_counts())
```

```python
# 강남구 필터 후 한 행을 구성하는 차원
key_cols = [
    "TA_YMD",
    "MCT_RY_CD",
    "CLN_SGG_CD",
    "SEX_CCD",
    "AGE_CCD"
]

# 동일한 조합이 여러 행 존재하는지 확인
duplicate_keys = gangnam_card.duplicated(
    subset=key_cols,
    keep=False
)

print("동일 조합 중복 행 수:", duplicate_keys.sum())

# 고유 조합 수
print("전체 행 수:", len(gangnam_card))
print(
    "고유 조합 수:",
    gangnam_card[key_cols].drop_duplicates().shape[0]
)
```

```python
# 날짜형 변환
gangnam_card["date"] = pd.to_datetime(
    gangnam_card["TA_YMD"].astype(str),
    format="%Y%m%d"
)

# YYYYMM 생성
gangnam_card["STD_YM"] = (
    gangnam_card["date"]
    .dt.strftime("%Y%m")
)

# 월별 집계
monthly_card = (
    gangnam_card
    .groupby("STD_YM")
    .agg(
        total_amount=("TS_AT", "sum"),
        total_count=("USE_CNT", "sum")
    )
    .reset_index()
)

# 건당 평균 결제금액
monthly_card["amount_per_use"] = (
    monthly_card["total_amount"]
    / monthly_card["total_count"]
)

display(monthly_card)
```

```python
monthly_card["amount_change_pct"] = (
    monthly_card["total_amount"]
    .pct_change() * 100
)

monthly_card["count_change_pct"] = (
    monthly_card["total_count"]
    .pct_change() * 100
)

monthly_card["amount_per_use_change_pct"] = (
    monthly_card["amount_per_use"]
    .pct_change() * 100
)

display(monthly_card.round(2))
```

```python
# 성별 소비 분석
gender_summary = (
    gangnam_card
    .groupby("SEX_CCD")
    .agg(
        total_amount=("TS_AT", "sum"),
        total_count=("USE_CNT", "sum")
    )
    .reset_index()
)

gender_summary["amount_per_use"] = (
    gender_summary["total_amount"]
    / gender_summary["total_count"]
)

gender_summary["amount_ratio"] = (
    gender_summary["total_amount"]
    / gender_summary["total_amount"].sum()
)

gender_summary["count_ratio"] = (
    gender_summary["total_count"]
    / gender_summary["total_count"].sum()
)

display(gender_summary.round(4))
```

```python
# 연령별 소비 분석
age_summary = (
    gangnam_card
    .groupby("AGE_CCD")
    .agg(
        total_amount=("TS_AT", "sum"),
        total_count=("USE_CNT", "sum")
    )
    .reset_index()
)

age_summary["amount_per_use"] = (
    age_summary["total_amount"]
    / age_summary["total_count"]
)

age_summary["amount_ratio"] = (
    age_summary["total_amount"]
    / age_summary["total_amount"].sum()
)

age_summary["count_ratio"] = (
    age_summary["total_count"]
    / age_summary["total_count"].sum()
)

# 연령 순서 정렬
age_order = [
    "10 대", "20 대", "30 대", "40 대", "50 대",
    "60 대", "70 대", "80 대", "90 대"
]

age_summary["AGE_CCD"] = pd.Categorical(
    age_summary["AGE_CCD"],
    categories=age_order,
    ordered=True
)

age_summary = (
    age_summary
    .sort_values("AGE_CCD")
    .reset_index(drop=True)
)

display(age_summary.round(4))
```

```python
industry_summary = (
    gangnam_card
    .groupby("MCT_RY_CD")
    .agg(
        total_amount=("TS_AT", "sum"),
        total_count=("USE_CNT", "sum")
    )
    .reset_index()
)

industry_summary["amount_per_use"] = (
    industry_summary["total_amount"]
    / industry_summary["total_count"]
)

industry_summary["amount_ratio"] = (
    industry_summary["total_amount"]
    / industry_summary["total_amount"].sum()
)

industry_summary["count_ratio"] = (
    industry_summary["total_count"]
    / industry_summary["total_count"].sum()
)

# 결제건수 기준 상위 20개 업종
top20_count = (
    industry_summary
    .sort_values("total_count", ascending=False)
    .head(20)
)

display(
    top20_count[
        [
            "MCT_RY_CD",
            "total_amount",
            "total_count",
            "amount_per_use",
            "amount_ratio",
            "count_ratio"
        ]
    ].round(4)
)
```

```python
top20_amount = (
    industry_summary
    .sort_values("total_amount", ascending=False)
    .head(20)
)

display(
    top20_amount[
        [
            "MCT_RY_CD",
            "total_amount",
            "total_count",
            "amount_per_use",
            "amount_ratio",
            "count_ratio"
        ]
    ].round(4)
)
```

```python
# 월 × 업종 대분류별 결제건수/금액
monthly_group = (
    gangnam_card
    .groupby(["STD_YM", "industry_group"])
    .agg(
        total_amount=("TS_AT", "sum"),
        total_count=("USE_CNT", "sum")
    )
    .reset_index()
)

# 월 전체 결제건수
monthly_total = (
    gangnam_card
    .groupby("STD_YM")
    .agg(
        month_amount=("TS_AT", "sum"),
        month_count=("USE_CNT", "sum")
    )
    .reset_index()
)

monthly_group = monthly_group.merge(
    monthly_total,
    on="STD_YM",
    how="left"
)

# 월별 구성비
monthly_group["amount_ratio"] = (
    monthly_group["total_amount"]
    / monthly_group["month_amount"]
)

monthly_group["count_ratio"] = (
    monthly_group["total_count"]
    / monthly_group["month_count"]
)

display(
    monthly_group[
        [
            "STD_YM",
            "industry_group",
            "total_amount",
            "total_count",
            "amount_ratio",
            "count_ratio"
        ]
    ].round(4)
)
```

```python
count_ratio_pivot = (
    monthly_group
    .pivot(
        index="STD_YM",
        columns="industry_group",
        values="count_ratio"
    )
)

display(count_ratio_pivot.round(4))
```

```python
# 연령 × 업종 대분류
age_group = (
    gangnam_card
    .groupby(["AGE_CCD", "industry_group"])
    .agg(
        total_amount=("TS_AT", "sum"),
        total_count=("USE_CNT", "sum")
    )
    .reset_index()
)

# 연령별 전체 소비
age_total = (
    gangnam_card
    .groupby("AGE_CCD")
    .agg(
        age_amount=("TS_AT", "sum"),
        age_count=("USE_CNT", "sum")
    )
    .reset_index()
)

age_group = age_group.merge(
    age_total,
    on="AGE_CCD",
    how="left"
)

# 각 연령 내부에서의 업종 구성비
age_group["amount_ratio"] = (
    age_group["total_amount"]
    / age_group["age_amount"]
)

age_group["count_ratio"] = (
    age_group["total_count"]
    / age_group["age_count"]
)

display(
    age_group[
        [
            "AGE_CCD",
            "industry_group",
            "total_amount",
            "total_count",
            "amount_ratio",
            "count_ratio"
        ]
    ].round(4)
)
```

```python
age_count_pivot = (
    age_group
    .pivot(
        index="AGE_CCD",
        columns="industry_group",
        values="count_ratio"
    )
)

# 연령 순서 정렬
age_order = [
    "10 대", "20 대", "30 대", "40 대", "50 대",
    "60 대", "70 대", "80 대", "90 대"
]

age_count_pivot = age_count_pivot.reindex(age_order)

display(age_count_pivot.round(4))
```

```python
# 연령별 전체 소비량 + 주요 소비구성 결합

age_activity = age_summary[
    [
        "AGE_CCD",
        "total_amount",
        "total_count",
        "amount_per_use",
        "amount_ratio",
        "count_ratio"
    ]
].copy()

# 주요 소비유형 구성비 추가
selected_groups = [
    "생활·식품",
    "식생활·외식",
    "문화·여가·운동",
    "의료·건강",
    "쇼핑·패션"
]

for group in selected_groups:
    age_activity[group + "_ratio"] = (
        age_count_pivot[group].reindex(age_activity["AGE_CCD"]).values
    )

display(age_activity.round(4))
```

```python
# ============================================================
# 소비자 거주지역별 소비 분석
# ============================================================

residence_summary = (
    gangnam_card
    .groupby("CLN_SGG_CD")
    .agg(
        total_amount=("TS_AT", "sum"),
        total_count=("USE_CNT", "sum")
    )
    .reset_index()
)

# 건당 결제금액
residence_summary["amount_per_use"] = (
    residence_summary["total_amount"]
    / residence_summary["total_count"]
)

# 전체 강남 소비에서 차지하는 비율
residence_summary["amount_ratio"] = (
    residence_summary["total_amount"]
    / residence_summary["total_amount"].sum()
)

residence_summary["count_ratio"] = (
    residence_summary["total_count"]
    / residence_summary["total_count"].sum()
)

# 결제건수 기준 정렬
residence_summary = (
    residence_summary
    .sort_values("total_count", ascending=False)
    .reset_index(drop=True)
)

display(residence_summary.round(4))
```

```python
# 거주지역 × 업종 대분류
residence_group = (
    gangnam_card
    .groupby(["CLN_SGG_CD", "industry_group"])
    .agg(
        total_amount=("TS_AT", "sum"),
        total_count=("USE_CNT", "sum")
    )
    .reset_index()
)

# 각 거주지역의 전체 결제건수/금액
residence_total = (
    gangnam_card
    .groupby("CLN_SGG_CD")
    .agg(
        residence_amount=("TS_AT", "sum"),
        residence_count=("USE_CNT", "sum")
    )
    .reset_index()
)

residence_group = residence_group.merge(
    residence_total,
    on="CLN_SGG_CD",
    how="left"
)

# 각 거주지역 내부의 소비유형 구성비
residence_group["count_ratio"] = (
    residence_group["total_count"]
    / residence_group["residence_count"]
)

residence_group["amount_ratio"] = (
    residence_group["total_amount"]
    / residence_group["residence_amount"]
)

# 결제건수 구성비 피벗
residence_count_pivot = (
    residence_group
    .pivot(
        index="CLN_SGG_CD",
        columns="industry_group",
        values="count_ratio"
    )
)

display(residence_count_pivot.round(4))
```

```python
# ============================================================
# 1. 월별 기본 Feature
# ============================================================

card_month_features = (
    gangnam_card
    .groupby("STD_YM")
    .agg(
        total_amount=("TS_AT", "sum"),
        total_count=("USE_CNT", "sum")
    )
    .reset_index()
)

card_month_features["amount_per_use"] = (
    card_month_features["total_amount"]
    / card_month_features["total_count"]
)

# ------------------------------------------------------------
# 성별 결제건수 구성비
# ------------------------------------------------------------

sex_month = (
    gangnam_card
    .groupby(["STD_YM", "SEX_CCD"])["USE_CNT"]
    .sum()
    .unstack(fill_value=0)
)

sex_month_ratio = sex_month.div(
    sex_month.sum(axis=1),
    axis=0
)

sex_month_ratio = sex_month_ratio.rename(
    columns={
        "남성": "male_count_ratio",
        "여성": "female_count_ratio"
    }
).reset_index()

card_month_features = card_month_features.merge(
    sex_month_ratio,
    on="STD_YM",
    how="left"
)

display(card_month_features.round(4))
```

```python
# ============================================================
# 2. 월 × 연령 결제건수 구성비
# ============================================================

age_month = (
    gangnam_card
    .groupby(["STD_YM", "AGE_CCD"])["USE_CNT"]
    .sum()
    .unstack(fill_value=0)
)

age_month_ratio = age_month.div(
    age_month.sum(axis=1),
    axis=0
)

age_column_map = {
    "10 대": "age_10_count_ratio",
    "20 대": "age_20_count_ratio",
    "30 대": "age_30_count_ratio",
    "40 대": "age_40_count_ratio",
    "50 대": "age_50_count_ratio",
    "60 대": "age_60_count_ratio",
    "70 대": "age_70_count_ratio",
    "80 대": "age_80_count_ratio",
    "90 대": "age_90_count_ratio"
}

age_month_ratio = (
    age_month_ratio
    .rename(columns=age_column_map)
    .reset_index()
)

card_month_features = card_month_features.merge(
    age_month_ratio,
    on="STD_YM",
    how="left"
)
```

```python
# ============================================================
# 3. 월 × 소비유형 결제건수 구성비
# ============================================================

industry_month = (
    gangnam_card
    .groupby(["STD_YM", "industry_group"])["USE_CNT"]
    .sum()
    .unstack(fill_value=0)
)

industry_month_ratio = industry_month.div(
    industry_month.sum(axis=1),
    axis=0
)

industry_column_map = {
    "개인·생활서비스": "ind_personal_ratio",
    "교육·학습": "ind_education_ratio",
    "교통·여행·숙박": "ind_transport_travel_ratio",
    "기타 특정 소비": "ind_other_specific_ratio",
    "문화·여가·운동": "ind_culture_leisure_ratio",
    "비활동성·해석제한": "ind_limited_ratio",
    "생활·식품": "ind_daily_food_ratio",
    "쇼핑·패션": "ind_shopping_ratio",
    "식생활·외식": "ind_dining_ratio",
    "의료·건강": "ind_health_ratio",
    "주거·자동차·전문서비스": "ind_housing_auto_service_ratio"
}

industry_month_ratio = (
    industry_month_ratio
    .rename(columns=industry_column_map)
    .reset_index()
)

card_month_features = card_month_features.merge(
    industry_month_ratio,
    on="STD_YM",
    how="left"
)
```

```python
# ============================================================
# 4. 월 × 소비자 거주지역 결제건수 구성비
# ============================================================

residence_month = (
    gangnam_card
    .groupby(["STD_YM", "CLN_SGG_CD"])["USE_CNT"]
    .sum()
    .unstack(fill_value=0)
)

residence_month_ratio = residence_month.div(
    residence_month.sum(axis=1),
    axis=0
)

residence_column_map = {
    "서울": "res_seoul_ratio",
    "경기": "res_gyeonggi_ratio",
    "인천": "res_incheon_ratio",
    "강원": "res_gangwon_ratio",
    "충북": "res_chungbuk_ratio",
    "충남": "res_chungnam_ratio",
    "대전": "res_daejeon_ratio",
    "세종": "res_sejong_ratio",
    "전북": "res_jeonbuk_ratio",
    "전남광주": "res_jeonnam_gwangju_ratio",
    "경북": "res_gyeongbuk_ratio",
    "경남": "res_gyeongnam_ratio",
    "대구": "res_daegu_ratio",
    "울산": "res_ulsan_ratio",
    "부산": "res_busan_ratio",
    "제주": "res_jeju_ratio",
    "정보없음": "res_unknown_ratio"
}

residence_month_ratio = (
    residence_month_ratio
    .rename(columns=residence_column_map)
    .reset_index()
)

card_month_features = card_month_features.merge(
    residence_month_ratio,
    on="STD_YM",
    how="left"
)
```

```python
# 성별 구성비
print(
    "성별 비율 합:",
    (
        card_month_features["male_count_ratio"]
        + card_month_features["female_count_ratio"]
    ).round(6).tolist()
)

# 연령 구성비
age_cols = [
    c for c in card_month_features.columns
    if c.startswith("age_")
]

print(
    "연령 비율 합:",
    card_month_features[age_cols]
    .sum(axis=1)
    .round(6)
    .tolist()
)

# 업종 구성비
industry_cols = [
    c for c in card_month_features.columns
    if c.startswith("ind_")
]

print(
    "업종 비율 합:",
    card_month_features[industry_cols]
    .sum(axis=1)
    .round(6)
    .tolist()
)

# 거주지역 구성비
residence_cols = [
    c for c in card_month_features.columns
    if c.startswith("res_")
]

print(
    "거주지역 비율 합:",
    card_month_features[residence_cols]
    .sum(axis=1)
    .round(6)
    .tolist()
)
```

```python
# ============================================================
# 강남구 × 월 × 연령 Feature Table
# ============================================================

# 1. 기본 소비 규모
card_month_age_features = (
    gangnam_card
    .groupby(["STD_YM", "AGE_CCD"])
    .agg(
        total_amount=("TS_AT", "sum"),
        total_count=("USE_CNT", "sum")
    )
    .reset_index()
)

card_month_age_features["amount_per_use"] = (
    card_month_age_features["total_amount"]
    / card_month_age_features["total_count"]
)


# ============================================================
# 2. 성별 구성비
# ============================================================

sex_age = (
    gangnam_card
    .groupby(["STD_YM", "AGE_CCD", "SEX_CCD"])["USE_CNT"]
    .sum()
    .unstack(fill_value=0)
)

sex_age_ratio = sex_age.div(
    sex_age.sum(axis=1),
    axis=0
)

sex_age_ratio = (
    sex_age_ratio
    .rename(columns={
        "남성": "male_count_ratio",
        "여성": "female_count_ratio"
    })
    .reset_index()
)

card_month_age_features = card_month_age_features.merge(
    sex_age_ratio,
    on=["STD_YM", "AGE_CCD"],
    how="left"
)


# ============================================================
# 3. 업종 구성비
# ============================================================

industry_age = (
    gangnam_card
    .groupby(
        ["STD_YM", "AGE_CCD", "industry_group"]
    )["USE_CNT"]
    .sum()
    .unstack(fill_value=0)
)

industry_age_ratio = industry_age.div(
    industry_age.sum(axis=1),
    axis=0
)

industry_age_ratio = (
    industry_age_ratio
    .rename(columns=industry_column_map)
    .reset_index()
)

card_month_age_features = card_month_age_features.merge(
    industry_age_ratio,
    on=["STD_YM", "AGE_CCD"],
    how="left"
)


# ============================================================
# 4. 거주지역 구성비
# ============================================================

residence_age = (
    gangnam_card
    .groupby(
        ["STD_YM", "AGE_CCD", "CLN_SGG_CD"]
    )["USE_CNT"]
    .sum()
    .unstack(fill_value=0)
)

residence_age_ratio = residence_age.div(
    residence_age.sum(axis=1),
    axis=0
)

residence_age_ratio = (
    residence_age_ratio
    .rename(columns=residence_column_map)
    .reset_index()
)

card_month_age_features = card_month_age_features.merge(
    residence_age_ratio,
    on=["STD_YM", "AGE_CCD"],
    how="left"
)

display(card_month_age_features.round(4))
```
