```python
import pandas as pd
import re
from google.colab import files

# 1. 7~12월 관심집단 파일 직접 선택
uploaded = files.upload()

data_list = []

for filename in uploaded.keys():

    # Excel 읽기
    df = pd.read_excel(filename)

    # 파일명에서 월 추출
    match = re.search(r'(\d{1,2})월', filename)

    if match:
        month = int(match.group(1))
    else:
        raise ValueError(f"파일명에서 월을 찾을 수 없습니다: {filename}")

    # 분석 기준 월 생성
    df["기준연월"] = f"2025-{month:02d}"

    data_list.append(df)

    print(f"{month}월: {len(df):,}행 / {len(df.columns)}열")

# 2. 세로 결합
interest_all = pd.concat(
    data_list,
    ignore_index=True
)

# 3. 기준연월 순서 정렬
interest_all = interest_all.sort_values(
    ["기준연월", "행정동코드", "성별", "연령대"]
).reset_index(drop=True)

print("\n통합 완료")
print("전체 크기:", interest_all.shape)

display(interest_all.head())
```

```python
# 1. 월별 행 수 확인
print("월별 행 수")
print(interest_all.groupby("기준연월").size())

# 2. 월별 행정동 수 확인
print("\n월별 행정동 수")
print(
    interest_all.groupby("기준연월")["행정동코드"]
    .nunique()
)

# 3. 분석 단위의 중복 확인
key_cols = [
    "기준연월",
    "행정동코드",
    "성별",
    "연령대"
]

duplicate_count = interest_all.duplicated(
    subset=key_cols
).sum()

print("\n중복 행 수:", duplicate_count)

# 4. 결측치 확인
print("\n컬럼별 결측치")
print(interest_all.isnull().sum())

# 5. 성별·연령대 범위 확인
print("\n성별:", sorted(interest_all["성별"].unique()))
print("연령대:", sorted(interest_all["연령대"].unique()))
```

```python
# 강남구 데이터 추출
gangnam_interest = (
    interest_all[interest_all["자치구"] == "강남구"]
    .copy()
    .reset_index(drop=True)
)

print("전체 크기:", gangnam_interest.shape)

print("\n월별 행 수")
print(gangnam_interest.groupby("기준연월").size())

print("\n월별 행정동 수")
print(
    gangnam_interest.groupby("기준연월")["행정동코드"]
    .nunique()
)

print("\n강남구 행정동")
print(
    gangnam_interest[["행정동코드", "행정동명"]]
    .drop_duplicates()
    .sort_values("행정동코드")
    .to_string(index=False)
)

display(gangnam_interest.head())
```

```python
# 일원2동 → 개포3동 명칭 통일
gangnam_interest["행정동명"] = gangnam_interest["행정동명"].replace({
    "일원2동": "개포3동"
})

# 확인
print(
    gangnam_interest[
        gangnam_interest["행정동명"] == "개포3동"
    ][["행정동코드", "행정동명"]]
    .drop_duplicates()
)
```

```python
# 행정동 × 성별 × 연령대별 1인가구수의 월별 변화 확인
single_change = (
    gangnam_interest
    .groupby(
        ["행정동코드", "행정동명", "성별", "연령대"]
    )["1인가구수"]
    .agg(
        월별값개수="nunique",
        최소값="min",
        최대값="max",
        평균="mean",
        표준편차="std"
    )
    .reset_index()
)

print("전체 집단 수:", len(single_change))

print("\n6개월 내 1인가구수가 변한 집단:")
print((single_change["월별값개수"] > 1).sum())

print("\n6개월 내 1인가구수가 동일한 집단:")
print((single_change["월별값개수"] == 1).sum())

# 월별 강남구 전체 1인가구수
monthly_single = (
    gangnam_interest
    .groupby("기준연월")["1인가구수"]
    .sum()
)

print("\n월별 강남구 1인가구수 합계")
print(monthly_single)
```

```python
import pandas as pd
from google.colab import files

# 1. 파일 업로드
uploaded = files.upload()
filename = list(uploaded.keys())[0]

# 2. 원본 데이터 불러오기
household_raw = pd.read_csv(filename)

print("전체 크기:", household_raw.shape)
print("\n컬럼 목록:")
print(household_raw.columns.tolist())

display(household_raw.head())
```

```python
# 강남구 행정동별 1인세대 추출
gangnam_household = household_raw[
    (household_raw["동별(1)"] == "강남구") &
    (household_raw["동별(2)"] != "소계")
][
    ["동별(2)", "2025 3/4.1", "2025 4/4.1"]
].copy()

# 컬럼명 정리
gangnam_household.columns = [
    "행정동명",
    "2025_Q3_1인세대",
    "2025_Q4_1인세대"
]

# 숫자형 변환
gangnam_household["2025_Q3_1인세대"] = pd.to_numeric(
    gangnam_household["2025_Q3_1인세대"]
)

gangnam_household["2025_Q4_1인세대"] = pd.to_numeric(
    gangnam_household["2025_Q4_1인세대"]
)

gangnam_household = gangnam_household.reset_index(drop=True)

print("행정동 수:", len(gangnam_household))

print("\n3분기 강남구 1인세대 합계:")
print(gangnam_household["2025_Q3_1인세대"].sum())

print("\n4분기 강남구 1인세대 합계:")
print(gangnam_household["2025_Q4_1인세대"].sum())

display(gangnam_household)
```

```python
# 관심집단 데이터의 행정동 목록
interest_dongs = set(gangnam_interest["행정동명"].unique())

# 공식 1인세대 데이터의 행정동 목록
household_dongs = set(gangnam_household["행정동명"].unique())

print("관심집단 행정동 수:", len(interest_dongs))
print("공식 세대수 행정동 수:", len(household_dongs))

print("\n관심집단에만 존재:")
print(sorted(interest_dongs - household_dongs))

print("\n공식 세대수에만 존재:")
print(sorted(household_dongs - interest_dongs))

print("\n양쪽 모두 존재:")
print(len(interest_dongs & household_dongs))
```

```python
# 1. 관심집단 데이터에 공식 1인세대 데이터 결합
gangnam_merged = gangnam_interest.merge(
    gangnam_household,
    on="행정동명",
    how="left",
    validate="many_to_one"
)

# 2. 월에 따라 사용할 공식 1인세대 수 선택
gangnam_merged["공식_1인세대수"] = gangnam_merged.apply(
    lambda row:
        row["2025_Q3_1인세대"]
        if row["기준연월"] in ["2025-07", "2025-08", "2025-09"]
        else row["2025_Q4_1인세대"],
    axis=1
)

# 3. 결합 검증
print("결합 후 크기:", gangnam_merged.shape)

print("\n공식 1인세대수 결측치:")
print(gangnam_merged["공식_1인세대수"].isna().sum())

print("\n월별 적용된 공식 기준값 종류:")
print(
    gangnam_merged
    .groupby("기준연월")["공식_1인세대수"]
    .sum()
)

display(
    gangnam_merged[
        ["기준연월", "행정동코드", "행정동명", "성별", "연령대",
         "1인가구수", "공식_1인세대수"]
    ].head()
)
```

```python
# 행정동 × 월 단위로 통신 1인가구수 집계
dong_month_compare = (
    gangnam_merged
    .groupby(
        ["기준연월", "행정동코드", "행정동명"],
        as_index=False
    )
    .agg(
        통신_1인가구수=("1인가구수", "sum"),
        공식_1인세대수=("공식_1인세대수", "first")
    )
)

# 공식 기준 대비 통신 데이터 비율
dong_month_compare["대비율_pct"] = (
    dong_month_compare["통신_1인가구수"]
    / dong_month_compare["공식_1인세대수"]
    * 100
)

# 절대 차이
dong_month_compare["차이"] = (
    dong_month_compare["통신_1인가구수"]
    - dong_month_compare["공식_1인세대수"]
)

print("데이터 크기:", dong_month_compare.shape)

# 월별 강남구 전체 비교
monthly_compare = (
    dong_month_compare
    .groupby("기준연월", as_index=False)
    .agg(
        통신_1인가구수=("통신_1인가구수", "sum"),
        공식_1인세대수=("공식_1인세대수", "sum")
    )
)

monthly_compare["대비율_pct"] = (
    monthly_compare["통신_1인가구수"]
    / monthly_compare["공식_1인세대수"]
    * 100
)

display(monthly_compare.round(2))

print("\n행정동별 비교 예시")
display(dong_month_compare.head(22).round(2))
```

```python
# 행정동별 6개월 대비율의 안정성 확인
dong_ratio_summary = (
    dong_month_compare
    .groupby(["행정동코드", "행정동명"], as_index=False)
    .agg(
        평균_대비율=("대비율_pct", "mean"),
        최소_대비율=("대비율_pct", "min"),
        최대_대비율=("대비율_pct", "max"),
        표준편차=("대비율_pct", "std")
    )
)

dong_ratio_summary["변동폭"] = (
    dong_ratio_summary["최대_대비율"]
    - dong_ratio_summary["최소_대비율"]
)

# 평균 대비율이 높은 순서로 확인
dong_ratio_summary = dong_ratio_summary.sort_values(
    "평균_대비율",
    ascending=False
).reset_index(drop=True)

display(dong_ratio_summary.round(2))

print("\n평균 표준편차:",
      round(dong_ratio_summary["표준편차"].mean(), 2))

print("가장 큰 표준편차:",
      round(dong_ratio_summary["표준편차"].max(), 2))
```

```python
# 관심집단 10개 지표
group_cols = [
    "커뮤니케이션이 적은 집단",
    "평일 외출이 적은 집단",
    "휴일 외출이 적은 집단",
    "출근소요시간 및 근무시간이 많은 집단",
    "외출이 매우 적은 집단(전체)",
    "외출이 매우 많은 집단",
    "동영상서비스 이용이 많은 집단",
    "생활서비스 이용이 많은 집단",
    "재정상태에 대한 관심집단",
    "외출-커뮤니케이션이 모두 적은 집단(전체)"
]

result = []

for col in group_cols:
    exceed = gangnam_merged[col] > gangnam_merged["1인가구수"]

    result.append({
        "지표": col,
        "1인가구수_초과건수": exceed.sum(),
        "초과비율_pct": exceed.mean() * 100,
        "최대값_대비_1인가구수": (
            gangnam_merged[col] / gangnam_merged["1인가구수"]
        ).max() * 100
    })

check_groups = pd.DataFrame(result)

display(check_groups.round(2))
```

```python
# 1. 1인가구 분석에 필요한 컬럼만 분리
single_household = gangnam_merged[
    [
        "기준연월",
        "행정동코드",
        "행정동명",
        "성별",
        "연령대",
        "1인가구수"
    ]
].copy()

# 2. 행정동 × 월별 전체 1인가구수
single_household["행정동_월_1인가구수"] = (
    single_household
    .groupby(["기준연월", "행정동코드"])["1인가구수"]
    .transform("sum")
)

# 3. 각 성별 × 연령대 집단의 구성비
single_household["구성비_pct"] = (
    single_household["1인가구수"]
    / single_household["행정동_월_1인가구수"]
    * 100
)

# 4. 기본 검증
print("데이터 크기:", single_household.shape)

print("\n월별 행 수:")
print(single_household.groupby("기준연월").size())

# 각 행정동 × 월에서 구성비 합계가 100%인지 확인
ratio_sum = (
    single_household
    .groupby(["기준연월", "행정동명"])["구성비_pct"]
    .sum()
)

print("\n구성비 합계 최소값:", ratio_sum.min())
print("구성비 합계 최대값:", ratio_sum.max())

display(single_household.head(12).round(2))
```

```python
# =========================
# 1. 성별 1인가구 데이터
# =========================

gender_single = (
    single_household
    .groupby(
        ["기준연월", "행정동코드", "행정동명", "성별"],
        as_index=False
    )
    .agg(
        single_count=("1인가구수", "sum"),
        dong_total=("행정동_월_1인가구수", "first")
    )
)

# 컬럼명 변경
gender_single = gender_single.rename(columns={
    "single_count": "1인가구수",
    "dong_total": "행정동_월_1인가구수"
})

gender_single["성별_구성비_pct"] = (
    gender_single["1인가구수"]
    / gender_single["행정동_월_1인가구수"]
    * 100
)


# =========================
# 2. 연령대별 1인가구 데이터
# =========================

age_single = (
    single_household
    .groupby(
        ["기준연월", "행정동코드", "행정동명", "연령대"],
        as_index=False
    )
    .agg(
        single_count=("1인가구수", "sum"),
        dong_total=("행정동_월_1인가구수", "first")
    )
)

# 컬럼명 변경
age_single = age_single.rename(columns={
    "single_count": "1인가구수",
    "dong_total": "행정동_월_1인가구수"
})

age_single["연령대_구성비_pct"] = (
    age_single["1인가구수"]
    / age_single["행정동_월_1인가구수"]
    * 100
)


# =========================
# 3. 검증
# =========================

print("성별 데이터 크기:", gender_single.shape)
print("연령대 데이터 크기:", age_single.shape)

gender_check = (
    gender_single
    .groupby(["기준연월", "행정동명"])["성별_구성비_pct"]
    .sum()
)

age_check = (
    age_single
    .groupby(["기준연월", "행정동명"])["연령대_구성비_pct"]
    .sum()
)

print("\n성별 구성비 합계:")
print(gender_check.min(), "~", gender_check.max())

print("\n연령대 구성비 합계:")
print(age_check.min(), "~", age_check.max())

print("\n성별 데이터 예시")
display(gender_single.head(6).round(2))

print("\n연령대 데이터 예시")
display(age_single.head(12).round(2))
```

```python
# ========================================
# single_household ↔ resident 결합 key 검증
# ========================================

# 1. 1인가구 쪽 행정동 × 분기 key
single_keys = (
    single_household[
        ["행정동명", "분기"]
    ]
    .drop_duplicates()
    .rename(columns={
        "행정동명": "dong_name",
        "분기": "quarter"
    })
)

# 2. 등록인구 쪽 행정동 × 분기 key
resident_keys = (
    resident[
        ["dong_name", "quarter"]
    ]
    .drop_duplicates()
)

print("1인가구 key 수:", len(single_keys))
print("등록인구 key 수:", len(resident_keys))


# 3. 양쪽 key 비교
key_check = single_keys.merge(
    resident_keys,
    on=["dong_name", "quarter"],
    how="outer",
    indicator=True
)

print("\nkey 대응 결과")
print(key_check["_merge"].value_counts())


# 4. 한쪽에만 존재하는 key 확인
print("\n1인가구에만 존재:")
display(
    key_check[
        key_check["_merge"] == "left_only"
    ]
)

print("\n등록인구에만 존재:")
display(
    key_check[
        key_check["_merge"] == "right_only"
    ]
)


# 5. 행정동 목록 확인
print("\n1인가구 행정동 수:",
      single_keys["dong_name"].nunique())

print("등록인구 행정동 수:",
      resident_keys["dong_name"].nunique())
```

```python
# 1인가구 연령대를 등록인구 Feature와 동일한 구간으로 재구성

def make_age_group(age):
    if age in [20, 25]:
        return "20"
    elif age in [30, 35]:
        return "30"
    elif age in [40, 45]:
        return "40"
    elif age in [50, 55]:
        return "50"
    elif age in [60, 65, 70, 75]:
        return "60_plus"
    else:
        return None

single_household["age_group"] = (
    single_household["연령대"]
    .apply(make_age_group)
)

print("연령그룹별 행 수")
print(single_household["age_group"].value_counts(dropna=False))

print("\n결측값:")
print(single_household["age_group"].isna().sum())
```

```python
# ========================================
# 등록인구: 20세 이상 기준 연령구성비 생성
# ========================================

resident["resident_20plus"] = (
    resident["resident_age_20"]
    + resident["resident_age_30"]
    + resident["resident_age_40"]
    + resident["resident_age_50"]
    + resident["resident_age_60_plus"]
)

age_groups = ["20", "30", "40", "50", "60_plus"]

for age in age_groups:
    resident[f"resident_age_{age}_ratio_20plus"] = (
        resident[f"resident_age_{age}"]
        / resident["resident_20plus"]
    )

# 검증
ratio_cols = [
    f"resident_age_{age}_ratio_20plus"
    for age in age_groups
]

resident["ratio_20plus_sum"] = resident[ratio_cols].sum(axis=1)

print("20세 이상 인구 최소값:", resident["resident_20plus"].min())

print(
    "20세 이상 연령 구성비 합계:",
    resident["ratio_20plus_sum"].min(),
    "~",
    resident["ratio_20plus_sum"].max()
)

display(
    resident[
        ["dong_name", "quarter", "resident_20plus"] + ratio_cols
    ].head()
)
```

```python
# ========================================
# 통신 1인가구: 20세 이상 연령그룹별 집계
# ========================================

single_age_group = (
    single_household
    .groupby(
        ["기준연월", "분기", "행정동코드", "행정동명", "age_group"],
        as_index=False
    )
    .agg(
        single_count=("1인가구수", "sum")
    )
)

# 행정동 × 월 전체 20세 이상 1인가구수
single_age_group["single_20plus"] = (
    single_age_group
    .groupby(["기준연월", "행정동코드"])["single_count"]
    .transform("sum")
)

# 1인가구 내부 연령 구성비
single_age_group["single_age_ratio_20plus"] = (
    single_age_group["single_count"]
    / single_age_group["single_20plus"]
)

# ========================================
# 검증
# ========================================

print("데이터 크기:", single_age_group.shape)

print("\n연령그룹별 행 수")
print(single_age_group["age_group"].value_counts())

ratio_check = (
    single_age_group
    .groupby(["기준연월", "행정동명"])["single_age_ratio_20plus"]
    .sum()
)

print(
    "\n1인가구 연령 구성비 합계:",
    ratio_check.min(),
    "~",
    ratio_check.max()
)

print("\n예시")
display(single_age_group.head(10).round(4))
```

```python
# ========================================
# 1. 등록인구 20+ 연령구조를 long 형태로 변환
# ========================================

resident_age_long = resident[
    [
        "dong_name",
        "quarter",
        "resident_20plus",
        "resident_age_20_ratio_20plus",
        "resident_age_30_ratio_20plus",
        "resident_age_40_ratio_20plus",
        "resident_age_50_ratio_20plus",
        "resident_age_60_plus_ratio_20plus"
    ]
].melt(
    id_vars=["dong_name", "quarter", "resident_20plus"],
    var_name="resident_age_variable",
    value_name="resident_age_ratio_20plus"
)

# 연령그룹 이름 추출
resident_age_long["age_group"] = (
    resident_age_long["resident_age_variable"]
    .str.replace("resident_age_", "", regex=False)
    .str.replace("_ratio_20plus", "", regex=False)
)

resident_age_long = resident_age_long.drop(
    columns="resident_age_variable"
)


# ========================================
# 2. 통신 1인가구와 등록인구 결합
# ========================================

single_age_compare = single_age_group.merge(
    resident_age_long,
    left_on=["행정동명", "분기", "age_group"],
    right_on=["dong_name", "quarter", "age_group"],
    how="left",
    validate="many_to_one"
)


# ========================================
# 3. 검증
# ========================================

print("결합 후 크기:", single_age_compare.shape)

print(
    "등록인구 구성비 결측값:",
    single_age_compare["resident_age_ratio_20plus"].isna().sum()
)

print(
    "등록인구 20+ 인구 결측값:",
    single_age_compare["resident_20plus"].isna().sum()
)

print("\n예시")
display(
    single_age_compare[
        [
            "기준연월",
            "분기",
            "행정동명",
            "age_group",
            "single_count",
            "single_age_ratio_20plus",
            "resident_20plus",
            "resident_age_ratio_20plus"
        ]
    ].head(10).round(4)
)
```

```python
# ========================================
# 1인가구 ↔ 등록인구 연령구조 차이 Feature
# ========================================

# 1. 구성비 차이
single_age_compare["single_resident_age_gap"] = (
    single_age_compare["single_age_ratio_20plus"]
    - single_age_compare["resident_age_ratio_20plus"]
)

# 2. 상대비율
single_age_compare["single_resident_age_ratio"] = (
    single_age_compare["single_age_ratio_20plus"]
    / single_age_compare["resident_age_ratio_20plus"]
)

# ========================================
# 검증 및 확인
# ========================================

print(
    "gap 결측값:",
    single_age_compare["single_resident_age_gap"].isna().sum()
)

print(
    "ratio 결측값:",
    single_age_compare["single_resident_age_ratio"].isna().sum()
)

display(
    single_age_compare[
        [
            "기준연월",
            "행정동명",
            "age_group",
            "single_age_ratio_20plus",
            "resident_age_ratio_20plus",
            "single_resident_age_gap",
            "single_resident_age_ratio"
        ]
    ].head(15).round(4)
)
```

```python
# ========================================
# 연령별 gap의 월별 안정성 확인
# ========================================

gap_stability = (
    single_age_compare
    .groupby(["행정동명", "age_group"], as_index=False)
    .agg(
        gap_mean=("single_resident_age_gap", "mean"),
        gap_min=("single_resident_age_gap", "min"),
        gap_max=("single_resident_age_gap", "max"),
        gap_std=("single_resident_age_gap", "std")
    )
)

gap_stability["gap_range"] = (
    gap_stability["gap_max"]
    - gap_stability["gap_min"]
)

# %p 단위로 보기 쉽게 변환
for col in ["gap_mean", "gap_min", "gap_max", "gap_std", "gap_range"]:
    gap_stability[col] = gap_stability[col] * 100

print("전체 행 수:", len(gap_stability))

print("\n연령그룹별 평균 변동성(%p)")
print(
    gap_stability
    .groupby("age_group")["gap_std"]
    .mean()
    .round(2)
)

print("\n변동성이 큰 조합 TOP 15")
display(
    gap_stability
    .sort_values("gap_std", ascending=False)
    .head(15)
    .round(2)
)
```

```python
# ========================================
# 행정동별 1인가구 연령구조 Feature Table
# ========================================

# gap 6개월 평균
single_age_feature = (
    single_age_compare
    .groupby(["행정동코드", "행정동명", "age_group"], as_index=False)
    .agg(
        age_gap_mean=("single_resident_age_gap", "mean")
    )
)

# wide 형태로 변환
single_age_feature = (
    single_age_feature
    .pivot(
        index=["행정동코드", "행정동명"],
        columns="age_group",
        values="age_gap_mean"
    )
    .reset_index()
)

# 컬럼명 정리
single_age_feature.columns.name = None

single_age_feature = single_age_feature.rename(columns={
    "20": "single_age_20_gap",
    "30": "single_age_30_gap",
    "40": "single_age_40_gap",
    "50": "single_age_50_gap",
    "60_plus": "single_age_60_plus_gap"
})

# %p 단위로 변환하지 않고 0~1 단위 유지
# 예: 0.12 = +12%p

print("Feature Table 크기:", single_age_feature.shape)

print("\n결측값:")
print(single_age_feature.isna().sum())

display(single_age_feature.head().round(4))
```

```python
# ========================================
# 행정동별 통신 기반 1인가구 규모 Feature
# ========================================

single_scale_feature = (
    single_household[
        ["기준연월", "행정동코드", "행정동명", "행정동_월_1인가구수"]
    ]
    .drop_duplicates()
    .groupby(["행정동코드", "행정동명"], as_index=False)
    .agg(
        single_household_mean=("행정동_월_1인가구수", "mean"),
        single_household_min=("행정동_월_1인가구수", "min"),
        single_household_max=("행정동_월_1인가구수", "max"),
        single_household_std=("행정동_월_1인가구수", "std")
    )
)

single_scale_feature["single_household_range"] = (
    single_scale_feature["single_household_max"]
    - single_scale_feature["single_household_min"]
)

print("Feature Table 크기:", single_scale_feature.shape)

print("\n결측값:")
print(single_scale_feature.isna().sum())

display(
    single_scale_feature
    .sort_values("single_household_mean", ascending=False)
    .round(2)
)
```
