```python
# ============================================================
# 1. CSV 읽기
# ============================================================

dfs = {}

for filename in uploaded.keys():

    # 파일명에서 연도 추출
    year = None
    for y in range(2022, 2026):
        if str(y) in filename:
            year = y
            break

    if year is None:
        print(f"[주의] 연도를 찾을 수 없음: {filename}")
        continue

    # 인코딩 자동 시도
    try:
        temp = pd.read_csv(filename, encoding="utf-8")
    except UnicodeDecodeError:
        temp = pd.read_csv(filename, encoding="cp949")

    dfs[year] = temp

    print(f"{year}: {temp.shape}")

print("\n로드된 연도:", sorted(dfs.keys()))
```

```python
# ============================================================
# 9. 서울 상권분석 기준 강남구 22개 행정동 추출
# ============================================================

gangnam_code_map = {
    11680510: "신사동",
    11680521: "논현1동",
    11680531: "논현2동",
    11680545: "압구정동",
    11680565: "청담동",
    11680580: "삼성1동",
    11680590: "삼성2동",
    11680600: "대치1동",
    11680610: "대치2동",
    11680630: "대치4동",
    11680640: "역삼1동",
    11680650: "역삼2동",
    11680655: "도곡1동",
    11680656: "도곡2동",
    11680660: "개포1동",
    11680670: "개포2동",
    11680690: "개포4동",
    11680700: "세곡동",
    11680720: "일원본동",
    11680730: "일원1동",

    # 서울 상권분석에서는 과거 명칭 유지
    11680740: "개포3동",

    11680750: "수서동"
}

gangnam_codes = list(gangnam_code_map.keys())

gangnam_dfs = {}

for year in sorted(dfs):

    temp = dfs[year][
        dfs[year]["행정동_코드"].isin(gangnam_codes)
    ].copy()

    # 이름 표준화
    temp["행정동명_표준"] = (
        temp["행정동_코드"]
        .map(gangnam_code_map)
    )

    gangnam_dfs[year] = temp

    print(
        year,
        "행 수:", len(temp),
        "| 동:", temp["행정동_코드"].nunique(),
        "| 분기:", temp["기준_년분기_코드"].nunique(),
        "| 업종:", temp["서비스_업종_코드"].nunique()
    )
```

```python
# ============================================================
# 10. 동 × 분기 × 업종 관측 완전성 검사
# ============================================================

for year in sorted(gangnam_dfs):

    temp = gangnam_dfs[year]

    print("\n" + "=" * 65)
    print(year)
    print("=" * 65)

    # 분석 key 중복
    dup = temp.duplicated(
        subset=[
            "기준_년분기_코드",
            "행정동_코드",
            "서비스_업종_코드"
        ]
    ).sum()

    print("key 중복:", dup)

    # 동×분기별 관측 업종 수
    coverage = (
        temp
        .groupby([
            "기준_년분기_코드",
            "행정동명_표준"
        ])["서비스_업종_코드"]
        .nunique()
        .reset_index(name="업종수")
    )

    print("동×분기 관측 수:", len(coverage))
    print("예상 동×분기 수:", 22 * 4)

    print(
        "업종수 min / median / max:",
        coverage["업종수"].min(),
        coverage["업종수"].median(),
        coverage["업종수"].max()
    )

    print("\n업종수가 63개 미만인 동×분기:")
    display(
        coverage[
            coverage["업종수"] < 63
        ].sort_values(
            ["업종수", "기준_년분기_코드"]
        )
    )
```

```python
# ============================================================
# 11. 연도별 업종 목록 및 사라진/추가된 업종 확인
# ============================================================

industry_sets = {}

for year in sorted(gangnam_dfs):

    temp = gangnam_dfs[year]

    industry_table = (
        temp[
            ["서비스_업종_코드", "서비스_업종_코드_명"]
        ]
        .drop_duplicates()
        .sort_values("서비스_업종_코드")
    )

    industry_sets[year] = set(
        industry_table["서비스_업종_코드"]
    )

    print(f"\n[{year}] 업종 수: {len(industry_table)}")
    display(industry_table)


# 연도 간 차이
for year in [2023, 2024, 2025]:

    removed = industry_sets[2022] - industry_sets[year]
    added = industry_sets[year] - industry_sets[2022]

    print(f"\n=== 2022 vs {year} ===")

    print("2022에는 있지만 해당 연도에는 없는 업종:")

    display(
        gangnam_dfs[2022][
            gangnam_dfs[2022]["서비스_업종_코드"].isin(removed)
        ][
            ["서비스_업종_코드", "서비스_업종_코드_명"]
        ].drop_duplicates()
    )

    print("해당 연도에 새로 존재하는 업종:")

    display(
        gangnam_dfs[year][
            gangnam_dfs[year]["서비스_업종_코드"].isin(added)
        ][
            ["서비스_업종_코드", "서비스_업종_코드_명"]
        ].drop_duplicates()
    )
```

```python
# ============================================================
# 12. 주요 행동신호 지역의 업종 관측 안정성
# ============================================================

target_dongs = [
    "개포3동",
    "일원본동",
    "일원1동",
    "신사동"
]

all_gangnam = pd.concat(
    gangnam_dfs.values(),
    ignore_index=True
)

coverage_target = (
    all_gangnam[
        all_gangnam["행정동명_표준"].isin(target_dongs)
    ]
    .groupby([
        "행정동명_표준",
        "기준_년분기_코드"
    ])["서비스_업종_코드"]
    .nunique()
    .reset_index(name="관측업종수")
)

pivot_target = coverage_target.pivot(
    index="기준_년분기_코드",
    columns="행정동명_표준",
    values="관측업종수"
)

display(pivot_target)
```

```python
# ============================================================
# 13. 주요 4개 동 업종의 16분기 관측 지속성 확인
# ============================================================

target_dongs = [
    "개포3동",
    "일원본동",
    "일원1동",
    "신사동"
]

# 업종별로 16개 분기 중 몇 개 분기에 관측됐는지 계산
industry_persistence = (
    all_gangnam[
        all_gangnam["행정동명_표준"].isin(target_dongs)
    ]
    .groupby([
        "행정동명_표준",
        "서비스_업종_코드",
        "서비스_업종_코드_명"
    ])["기준_년분기_코드"]
    .nunique()
    .reset_index(name="관측분기수")
)

# 각 동별 분포
for dong in target_dongs:

    temp = industry_persistence[
        industry_persistence["행정동명_표준"] == dong
    ].sort_values(
        ["관측분기수", "서비스_업종_코드"],
        ascending=[False, True]
    )

    print("\n" + "=" * 65)
    print(dong)
    print("=" * 65)

    print("한 번이라도 등장한 업종:", len(temp))
    print("16분기 모두 관측:", (temp["관측분기수"] == 16).sum())
    print("12분기 이상 관측:", (temp["관측분기수"] >= 12).sum())
    print("8분기 이상 관측:", (temp["관측분기수"] >= 8).sum())

    display(temp)
```

```python
# ============================================================
# 14. 주요 지역 업종 지속성 요약
# ============================================================

summary = []

for dong in target_dongs:

    temp = industry_persistence[
        industry_persistence["행정동명_표준"] == dong
    ]

    summary.append({
        "행정동": dong,
        "등장업종수": len(temp),
        "16분기_모두관측": (temp["관측분기수"] == 16).sum(),
        "12분기이상": (temp["관측분기수"] >= 12).sum(),
        "8분기이상": (temp["관측분기수"] >= 8).sum(),
        "4분기미만": (temp["관측분기수"] < 4).sum()
    })

persistence_summary = pd.DataFrame(summary)

display(persistence_summary)
```

```python
# ============================================================
# 16. 전체 매출건수 vs 연령별 매출건수 합 검증
# ============================================================

age_count_cols = [
    "연령대_10_매출_건수",
    "연령대_20_매출_건수",
    "연령대_30_매출_건수",
    "연령대_40_매출_건수",
    "연령대_50_매출_건수",
    "연령대_60_이상_매출_건수"
]

check_df = all_gangnam.copy()

check_df["연령별_매출건수_합"] = (
    check_df[age_count_cols].sum(axis=1)
)

check_df["연령합_대비_전체_비율"] = (
    check_df["연령별_매출건수_합"]
    / check_df["당월_매출_건수"].replace(0, pd.NA)
)

print("전체 행:", len(check_df))

print("\n[연령별 합 / 전체 매출건수]")
print(
    check_df["연령합_대비_전체_비율"]
    .describe(
        percentiles=[0.01, 0.05, 0.25, 0.5, 0.75, 0.95, 0.99]
    )
)

print("\n전체 매출건수 0인 행:",
      (check_df["당월_매출_건수"] == 0).sum())

print("연령별 합이 전체보다 큰 행:",
      (
          check_df["연령별_매출건수_합"]
          > check_df["당월_매출_건수"]
      ).sum())
```

```python
# ============================================================
# 19. Analysis 3-A
# 행정동 × 분기 × 연령 × 소비영역 Feature Table 생성
# ============================================================

age_map = {
    "연령대_10_매출_건수": "10대",
    "연령대_20_매출_건수": "20대",
    "연령대_30_매출_건수": "30대",
    "연령대_40_매출_건수": "40대",
    "연령대_50_매출_건수": "50대",
    "연령대_60_이상_매출_건수": "60대이상"
}

# Core 소비영역만
core = analysis3[
    analysis3["소비영역"].notna()
].copy()

# wide → long
age_long = core.melt(
    id_vars=[
        "기준_년분기_코드",
        "행정동_코드",
        "행정동명_표준",
        "서비스_업종_코드",
        "서비스_업종_코드_명",
        "소비영역"
    ],
    value_vars=list(age_map.keys()),
    var_name="연령컬럼",
    value_name="매출건수"
)

age_long["연령대"] = (
    age_long["연령컬럼"].map(age_map)
)

# 영역 단위 집계
age_domain = (
    age_long
    .groupby([
        "기준_년분기_코드",
        "행정동_코드",
        "행정동명_표준",
        "연령대",
        "소비영역"
    ], as_index=False)
    .agg(
        매출건수=("매출건수", "sum"),
        관측업종수=("서비스_업종_코드", "nunique")
    )
)

print("shape:", age_domain.shape)
print("행정동:", age_domain["행정동명_표준"].nunique())
print("분기:", age_domain["기준_년분기_코드"].nunique())
print("연령대:", age_domain["연령대"].nunique())
print("소비영역:", age_domain["소비영역"].nunique())

print(
    "key 중복:",
    age_domain.duplicated(
        subset=[
            "기준_년분기_코드",
            "행정동_코드",
            "연령대",
            "소비영역"
        ]
    ).sum()
)

print("결측:", age_domain.isna().sum().sum())
```

```python
# ============================================================
# 21. 2025 Q3/Q4 vs 2022~2024 동일분기 baseline
# ============================================================

compare = age_domain.copy()

# 연도 / 분기 분리
compare["연도"] = (
    compare["기준_년분기_코드"]
    .astype(str)
    .str[:4]
    .astype(int)
)

compare["분기"] = (
    compare["기준_년분기_코드"]
    .astype(str)
    .str[-1]
    .astype(int)
)

keys = [
    "행정동_코드",
    "행정동명_표준",
    "연령대",
    "소비영역",
    "분기"
]

# ------------------------------------------------------------
# 2022~2024 historical baseline
# ------------------------------------------------------------

baseline = (
    compare[
        (compare["연도"].between(2022, 2024))
        &
        (compare["분기"].isin([3, 4]))
    ]
    .groupby(keys, as_index=False)
    .agg(
        baseline_mean=("매출건수", "mean"),
        baseline_median=("매출건수", "median"),
        baseline_std=("매출건수", "std"),
        baseline_n=("매출건수", "count"),
        baseline_min=("매출건수", "min"),
        baseline_max=("매출건수", "max")
    )
)

# ------------------------------------------------------------
# 2025 Q3/Q4
# ------------------------------------------------------------

current = (
    compare[
        (compare["연도"] == 2025)
        &
        (compare["분기"].isin([3, 4]))
    ][
        keys
        + [
            "기준_년분기_코드",
            "매출건수",
            "관측업종수"
        ]
    ]
    .rename(
        columns={
            "매출건수": "current_count",
            "관측업종수": "current_industry_count"
        }
    )
)

# baseline 결합
card_detection = current.merge(
    baseline,
    on=keys,
    how="left",
    validate="one_to_one"
)

# ------------------------------------------------------------
# 변화량
# ------------------------------------------------------------

card_detection["change_from_mean"] = (
    card_detection["current_count"]
    - card_detection["baseline_mean"]
)

card_detection["change_rate_pct"] = (
    card_detection["change_from_mean"]
    / card_detection["baseline_mean"].replace(0, pd.NA)
    * 100
)

# 과거 범위보다 낮거나 높은지도 확인
card_detection["below_historical_min"] = (
    card_detection["current_count"]
    < card_detection["baseline_min"]
)

card_detection["above_historical_max"] = (
    card_detection["current_count"]
    > card_detection["baseline_max"]
)

print("shape:", card_detection.shape)

print(
    "baseline 3개 연도 모두 존재:",
    (card_detection["baseline_n"] == 3).sum(),
    "/",
    len(card_detection)
)

print(
    "baseline 결측:",
    card_detection["baseline_mean"].isna().sum()
)

print(
    "과거 3개년 최소보다 낮음:",
    card_detection["below_historical_min"].sum()
)

print(
    "과거 3개년 최대보다 높음:",
    card_detection["above_historical_max"].sum()
)
```

```python
# ============================================================
# 22. 주요 4개 지역 baseline 비교 상태 확인
# ============================================================

target_detection = card_detection[
    card_detection["행정동명_표준"].isin(
        ["개포3동", "일원본동", "일원1동", "신사동"]
    )
].copy()

print("주요 4개 지역 행 수:", len(target_detection))

print(
    "baseline_n 분포:"
)
print(
    target_detection["baseline_n"]
    .value_counts(dropna=False)
    .sort_index()
)

print(
    "\n변화율 계산 불가:",
    target_detection["change_rate_pct"].isna().sum()
)

display(
    target_detection[
        [
            "기준_년분기_코드",
            "행정동명_표준",
            "연령대",
            "소비영역",
            "current_count",
            "baseline_mean",
            "baseline_min",
            "baseline_max",
            "baseline_n",
            "change_rate_pct",
            "below_historical_min",
            "above_historical_max"
        ]
    ]
    .sort_values([
        "행정동명_표준",
        "기준_년분기_코드",
        "연령대",
        "소비영역"
    ])
)
```

```python
# ============================================================
# 23. 강남구 내 상대적 변화 위치 계산
# 동 × 연령 × 소비영역의 변화가
# 강남구 다른 동과 비교해 어느 위치인지 확인
# ============================================================

valid = card_detection[
    card_detection["change_rate_pct"].notna()
].copy()

# 동일 분기 × 연령 × 소비영역 내 percentile rank
valid["gangnam_change_percentile"] = (
    valid
    .groupby([
        "기준_년분기_코드",
        "연령대",
        "소비영역"
    ])["change_rate_pct"]
    .rank(
        method="average",
        pct=True
    )
)

# 비교집단 크기도 보존
group_n = (
    valid
    .groupby([
        "기준_년분기_코드",
        "연령대",
        "소비영역"
    ])
    .size()
    .rename("gangnam_compare_n")
    .reset_index()
)

valid = valid.merge(
    group_n,
    on=[
        "기준_년분기_코드",
        "연령대",
        "소비영역"
    ],
    how="left",
    validate="many_to_one"
)

print("분석 가능 행:", len(valid))

print("\n비교집단 크기 분포")
print(
    valid["gangnam_compare_n"]
    .value_counts()
    .sort_index()
)

print("\npercentile 요약")
print(
    valid["gangnam_change_percentile"]
    .describe()
)
```

```python
# ============================================================
# 24. 주요 4개 동의 강남구 내 상대 위치
# ============================================================

target_relative = valid[
    valid["행정동명_표준"].isin(
        ["개포3동", "일원본동", "일원1동", "신사동"]
    )
].copy()

display(
    target_relative[
        [
            "기준_년분기_코드",
            "행정동명_표준",
            "연령대",
            "소비영역",
            "current_count",
            "baseline_mean",
            "change_rate_pct",
            "gangnam_change_percentile",
            "gangnam_compare_n",
            "below_historical_min",
            "above_historical_max"
        ]
    ]
    .sort_values([
        "기준_년분기_코드",
        "행정동명_표준",
        "연령대",
        "소비영역"
    ])
)
```

```python
# ============================================================
# 25. Historical × Relative 이중 변화신호 확인
# ============================================================

signal = valid.copy()

# 상대적 하위/상위 20%
signal["relative_low"] = (
    signal["gangnam_change_percentile"] <= 0.20
)

signal["relative_high"] = (
    signal["gangnam_change_percentile"] >= 0.80
)

# 자기 과거 + 강남구 상대비교가 같은 방향인 경우
signal["dual_low_signal"] = (
    signal["below_historical_min"]
    & signal["relative_low"]
)

signal["dual_high_signal"] = (
    signal["above_historical_max"]
    & signal["relative_high"]
)

print("전체 분석 가능 행:", len(signal))

print("\n이중 감소신호:", signal["dual_low_signal"].sum())
print("이중 증가신호:", signal["dual_high_signal"].sum())

print("\n분기별 이중 변화신호")
print(
    signal.groupby("기준_년분기_코드")[
        ["dual_low_signal", "dual_high_signal"]
    ].sum()
)

print("\n소비영역별 이중 변화신호")
print(
    signal.groupby("소비영역")[
        ["dual_low_signal", "dual_high_signal"]
    ].sum()
)
```

```python
print("\n연령대별 이중 변화신호")

print(
    signal.groupby("연령대")[
        ["dual_low_signal", "dual_high_signal"]
    ].sum()
)
```

```python
# ============================================================
# 26. 동 × 분기 × 연령별 소비영역 변화 패턴 요약
# ============================================================

age_signal_summary = (
    signal
    .groupby([
        "기준_년분기_코드",
        "행정동_코드",
        "행정동명_표준",
        "연령대"
    ], as_index=False)
    .agg(
        observed_domains=("소비영역", "nunique"),
        dual_low_domains=("dual_low_signal", "sum"),
        dual_high_domains=("dual_high_signal", "sum")
    )
)

print("shape:", age_signal_summary.shape)

print("\n이중 감소 영역 수 분포")
print(
    age_signal_summary["dual_low_domains"]
    .value_counts()
    .sort_index()
)

print("\n이중 증가 영역 수 분포")
print(
    age_signal_summary["dual_high_domains"]
    .value_counts()
    .sort_index()
)
```

```python
# ============================================================
# 31. Analysis 3-B
# 강남구 × 월 × 연령 × 업종 Feature Table 생성
# ============================================================

# 강남구만 추출
shinhan_gangnam = shinhan_raw[
    shinhan_raw["MCT_SGG_CD"] == "서울 강남구"
].copy()

# 날짜형 변환
shinhan_gangnam["TA_YMD"] = pd.to_datetime(
    shinhan_gangnam["TA_YMD"].astype(str),
    format="%Y%m%d"
)

# 월 생성
shinhan_gangnam["기준연월"] = (
    shinhan_gangnam["TA_YMD"]
    .dt.strftime("%Y%m")
    .astype(int)
)

# 월 × 연령 × 업종 집계
shinhan_month_age_industry = (
    shinhan_gangnam
    .groupby(
        [
            "기준연월",
            "AGE_CCD",
            "MCT_RY_CD"
        ],
        as_index=False
    )
    .agg(
        이용건수=("USE_CNT", "sum"),
        이용금액=("TS_AT", "sum"),
        관측일수=("TA_YMD", "nunique")
    )
)

print("shape:", shinhan_month_age_industry.shape)

print(
    "월:",
    sorted(shinhan_month_age_industry["기준연월"].unique())
)

print(
    "연령대:",
    sorted(shinhan_month_age_industry["AGE_CCD"].unique())
)

print(
    "업종 수:",
    shinhan_month_age_industry["MCT_RY_CD"].nunique()
)

print(
    "key 중복:",
    shinhan_month_age_industry.duplicated(
        ["기준연월", "AGE_CCD", "MCT_RY_CD"]
    ).sum()
)

print(
    "결측:",
    shinhan_month_age_industry.isna().sum().sum()
)

print("\n관측일수 분포:")
print(
    shinhan_month_age_industry["관측일수"]
    .value_counts()
    .sort_index()
)
```

```python
# ============================================================
# 33. Analysis 3-B
# 월 길이를 보정한 일평균 이용건수 생성
# ============================================================

import calendar

shinhan_month_age_industry["연도"] = (
    shinhan_month_age_industry["기준연월"] // 100
)

shinhan_month_age_industry["월"] = (
    shinhan_month_age_industry["기준연월"] % 100
)

# 실제 달력 일수
shinhan_month_age_industry["월일수"] = (
    shinhan_month_age_industry.apply(
        lambda x: calendar.monthrange(
            int(x["연도"]),
            int(x["월"])
        )[1],
        axis=1
    )
)

# 월 전체 일수 기준 일평균
shinhan_month_age_industry["일평균_이용건수"] = (
    shinhan_month_age_industry["이용건수"]
    / shinhan_month_age_industry["월일수"]
)

# 검증
print(
    shinhan_month_age_industry[
        ["기준연월", "월일수"]
    ]
    .drop_duplicates()
    .sort_values("기준연월")
)

print("\n일평균 이용건수 결측:",
      shinhan_month_age_industry["일평균_이용건수"].isna().sum())

print(
    "일평균 이용건수 <= 0:",
    (shinhan_month_age_industry["일평균_이용건수"] <= 0).sum()
)
```

```python
# ============================================================
# 34. Analysis 3-B
# 연령대 × 월 전체 소비활동 기준선 생성
# ============================================================

age_month_baseline = (
    shinhan_month_age_industry
    .groupby(
        ["기준연월", "AGE_CCD"],
        as_index=False
    )
    .agg(
        월이용건수=("이용건수", "sum")
    )
)

# 월 일수
days_map = {
    202507: 31,
    202508: 31,
    202509: 30,
    202510: 31,
    202511: 30,
    202512: 31
}

age_month_baseline["월일수"] = (
    age_month_baseline["기준연월"].map(days_map)
)

age_month_baseline["일평균_전체이용건수"] = (
    age_month_baseline["월이용건수"]
    / age_month_baseline["월일수"]
)

# 전월 대비 변화율
age_month_baseline = age_month_baseline.sort_values(
    ["AGE_CCD", "기준연월"]
)

age_month_baseline["전월대비_pct"] = (
    age_month_baseline
    .groupby("AGE_CCD")["일평균_전체이용건수"]
    .pct_change(fill_method=None)
    * 100
)

print("shape:", age_month_baseline.shape)
print(
    "key 중복:",
    age_month_baseline.duplicated(
        ["기준연월", "AGE_CCD"]
    ).sum()
)

display(
    age_month_baseline.pivot(
        index="AGE_CCD",
        columns="기준연월",
        values="전월대비_pct"
    ).round(2)
)
```

```python
# ============================================================
# 35. Analysis 3-B
# 연령 전체 기준선 대비 개별 업종의 상대적 월 변화 계산
# ============================================================

industry_change = shinhan_month_age_industry.copy()

industry_change = industry_change.sort_values(
    ["AGE_CCD", "MCT_RY_CD", "기준연월"]
)

# 개별 연령 × 업종의 일평균 이용건수 전월 대비 변화율
industry_change["업종_전월대비_pct"] = (
    industry_change
    .groupby(
        ["AGE_CCD", "MCT_RY_CD"]
    )["일평균_이용건수"]
    .pct_change(fill_method=None)
    * 100
)

# 연령 전체 기준선 결합
industry_change = industry_change.merge(
    age_month_baseline[
        [
            "기준연월",
            "AGE_CCD",
            "전월대비_pct"
        ]
    ].rename(
        columns={
            "전월대비_pct": "연령전체_전월대비_pct"
        }
    ),
    on=["기준연월", "AGE_CCD"],
    how="left",
    validate="many_to_one"
)

# 연령 전체 변화와 비교한 업종의 상대 변화
industry_change["업종_상대변화_pp"] = (
    industry_change["업종_전월대비_pct"]
    - industry_change["연령전체_전월대비_pct"]
)

print("shape:", industry_change.shape)

print(
    "key 중복:",
    industry_change.duplicated(
        ["기준연월", "AGE_CCD", "MCT_RY_CD"]
    ).sum()
)

# 7월은 전월 자료가 없으므로 제외
valid_industry_change = industry_change[
    industry_change["기준연월"] >= 202508
].copy()

print("\n상대변화 계산 가능:",
      valid_industry_change["업종_상대변화_pp"].notna().sum())

print(
    "상대변화 계산 불가:",
    valid_industry_change["업종_상대변화_pp"].isna().sum()
)

print("\n상대변화 분포:")
print(
    valid_industry_change["업종_상대변화_pp"]
    .describe()
)
```

```python
# ============================================================
# 38. Analysis 3-B
# 월 × 연령 × 소비영역 이용건수 집계 및 변화율 계산
# ============================================================

import calendar

# 월 × 연령 × 소비영역 집계
age_domain_month = (
    shinhan_domain_core
    .groupby(
        ["기준연월", "AGE_CCD", "소비영역"],
        as_index=False
    )
    .agg(
        이용건수=("USE_CNT", "sum"),
        이용금액=("TS_AT", "sum"),
        포함업종수=("MCT_RY_CD", "nunique")
    )
)

# 월 일수
age_domain_month["연도"] = (
    age_domain_month["기준연월"] // 100
)

age_domain_month["월"] = (
    age_domain_month["기준연월"] % 100
)

age_domain_month["월일수"] = age_domain_month.apply(
    lambda x: calendar.monthrange(
        int(x["연도"]),
        int(x["월"])
    )[1],
    axis=1
)

# 월 길이 보정
age_domain_month["일평균_이용건수"] = (
    age_domain_month["이용건수"]
    / age_domain_month["월일수"]
)

# 정렬
age_domain_month = age_domain_month.sort_values(
    ["AGE_CCD", "소비영역", "기준연월"]
)

# 소비영역 자체의 전월 대비 변화율
age_domain_month["영역_전월대비_pct"] = (
    age_domain_month
    .groupby(
        ["AGE_CCD", "소비영역"]
    )["일평균_이용건수"]
    .pct_change(fill_method=None)
    * 100
)

# 앞에서 만든 연령 전체 소비 기준선 결합
age_domain_month = age_domain_month.merge(
    age_month_baseline[
        [
            "기준연월",
            "AGE_CCD",
            "전월대비_pct"
        ]
    ].rename(
        columns={
            "전월대비_pct":
            "연령전체_전월대비_pct"
        }
    ),
    on=["기준연월", "AGE_CCD"],
    how="left",
    validate="many_to_one"
)

# 연령 전체 흐름 대비 소비영역 상대변화
age_domain_month["영역_상대변화_pp"] = (
    age_domain_month["영역_전월대비_pct"]
    - age_domain_month["연령전체_전월대비_pct"]
)

# ------------------------------------------------------------
# 검증
# ------------------------------------------------------------

print("shape:", age_domain_month.shape)

print(
    "key 중복:",
    age_domain_month.duplicated(
        ["기준연월", "AGE_CCD", "소비영역"]
    ).sum()
)

print(
    "결측:",
    age_domain_month.isna().sum()
)

print("\n=== 월 × 연령별 소비영역 수 ===")
display(
    age_domain_month
    .groupby(
        ["기준연월", "AGE_CCD"]
    )["소비영역"]
    .nunique()
    .unstack()
)

print("\n=== 8~12월 영역 상대변화 분포 ===")

print(
    age_domain_month.loc[
        age_domain_month["기준연월"] >= 202508,
        "영역_상대변화_pp"
    ].describe()
)
```

```python
# ============================================================
# 42. Analysis 3-B
# 신한카드 60·70·80·90대 → 60대이상 재집계
# ============================================================

senior_ages = ["60 대", "70 대", "80 대", "90 대"]

# 5개 핵심 소비영역 중 60대 이상만
senior_raw = shinhan_domain_core[
    shinhan_domain_core["AGE_CCD"].isin(senior_ages)
].copy()

# 월 × 소비영역으로 원자료부터 합산
senior_domain_month = (
    senior_raw
    .groupby(
        ["기준연월", "소비영역"],
        as_index=False
    )
    .agg(
        이용건수=("USE_CNT", "sum"),
        이용금액=("TS_AT", "sum")
    )
)

senior_domain_month["연령대"] = "60대이상"

# 월 일수
senior_domain_month["월일수"] = (
    senior_domain_month["기준연월"]
    .map({
        202507: 31,
        202508: 31,
        202509: 30,
        202510: 31,
        202511: 30,
        202512: 31
    })
)

senior_domain_month["일평균_이용건수"] = (
    senior_domain_month["이용건수"]
    / senior_domain_month["월일수"]
)

# ------------------------------------------------------------
# 60대 이상 전체 소비 기준선
# 핵심 5개 영역이 아니라 원본 88개 업종 전체 USE_CNT 사용
# ------------------------------------------------------------

senior_total_month = (
    shinhan_gangnam[
        shinhan_gangnam["AGE_CCD"].isin(senior_ages)
    ]
    .groupby(
        "기준연월",
        as_index=False
    )
    .agg(
        연령전체_이용건수=("USE_CNT", "sum")
    )
)

senior_total_month["월일수"] = (
    senior_total_month["기준연월"]
    .map({
        202507: 31,
        202508: 31,
        202509: 30,
        202510: 31,
        202511: 30,
        202512: 31
    })
)

senior_total_month["연령전체_일평균"] = (
    senior_total_month["연령전체_이용건수"]
    / senior_total_month["월일수"]
)

# 전월 대비
senior_domain_month = senior_domain_month.sort_values(
    ["소비영역", "기준연월"]
)

senior_domain_month["영역_전월대비_pct"] = (
    senior_domain_month
    .groupby("소비영역")["일평균_이용건수"]
    .pct_change(fill_method=None)
    * 100
)

senior_total_month = senior_total_month.sort_values("기준연월")

senior_total_month["연령전체_전월대비_pct"] = (
    senior_total_month["연령전체_일평균"]
    .pct_change(fill_method=None)
    * 100
)

# 결합
senior_domain_month = senior_domain_month.merge(
    senior_total_month[
        [
            "기준연월",
            "연령전체_전월대비_pct"
        ]
    ],
    on="기준연월",
    how="left",
    validate="many_to_one"
)

senior_domain_month["영역_상대변화_pp"] = (
    senior_domain_month["영역_전월대비_pct"]
    - senior_domain_month["연령전체_전월대비_pct"]
)

# 검증
print("shape:", senior_domain_month.shape)

print(
    "월별 소비영역 수:"
)
print(
    senior_domain_month
    .groupby("기준연월")["소비영역"]
    .nunique()
)

print("\n결측:")
print(senior_domain_month.isna().sum())

print("\n=== 60대이상 월별 영역 변화 ===")

display(
    senior_domain_month[
        [
            "기준연월",
            "연령대",
            "소비영역",
            "영역_전월대비_pct",
            "연령전체_전월대비_pct",
            "영역_상대변화_pp"
        ]
    ].sort_values(
        ["기준연월", "소비영역"]
    )
)
```

```python
# ============================================================
# 43. Analysis 2 + 3-A + 3-B 최종 연결
# 10~50대 + 60대이상 통합
# ============================================================

import pandas as pd

# ------------------------------------------------------------
# 1) 3-A의 60대이상 dual-low 사례
# ------------------------------------------------------------

behavior_3a_senior = behavior_3a_low[
    behavior_3a_low["연령대"] == "60대이상"
].copy()

print("3-A 60대이상 dual-low:", len(behavior_3a_senior))


# ------------------------------------------------------------
# 2) 행동신호 발생 월로 펼치기
# ------------------------------------------------------------

behavior_3a_senior["기준연월"] = (
    behavior_3a_senior["behavior_signal_month_list"]
    .str.split(", ")
)

behavior_3a_senior = behavior_3a_senior.explode("기준연월")

behavior_3a_senior["기준연월"] = (
    behavior_3a_senior["기준연월"].astype(int)
)


# ------------------------------------------------------------
# 3) 60대이상 신한카드 월별 Context 연결
# ------------------------------------------------------------

senior_context_cols = [
    "기준연월",
    "연령대",
    "소비영역",
    "영역_전월대비_pct",
    "연령전체_전월대비_pct",
    "영역_상대변화_pp"
]

analysis_senior = behavior_3a_senior.merge(
    senior_domain_month[senior_context_cols],
    on=[
        "기준연월",
        "연령대",
        "소비영역"
    ],
    how="left",
    validate="many_to_one"
)

print(
    "60대이상 월 단위 연결:",
    len(analysis_senior)
)

print(
    "60대이상 3-B 결측:",
    analysis_senior["영역_상대변화_pp"].isna().sum()
)


# ------------------------------------------------------------
# 4) 10~50대 + 60대이상 통합
# ------------------------------------------------------------

common_cols = [
    "행정동명_표준",
    "기준_년분기_코드",
    "기준연월",
    "behavior_signal_months",
    "behavior_signal_month_list",
    "연령대",
    "소비영역",
    "change_rate_pct",
    "gangnam_change_percentile",
    "dual_low_signal",
    "영역_전월대비_pct",
    "연령전체_전월대비_pct",
    "영역_상대변화_pp"
]

analysis_2_3a_3b_final = pd.concat(
    [
        analysis_2_3a_3b[common_cols],
        analysis_senior[common_cols]
    ],
    ignore_index=True
)

analysis_2_3a_3b_final = (
    analysis_2_3a_3b_final
    .sort_values(
        [
            "기준연월",
            "행정동명_표준",
            "연령대",
            "소비영역"
        ]
    )
    .reset_index(drop=True)
)


# ------------------------------------------------------------
# 5) 검증
# ------------------------------------------------------------

print("\n=== 최종 연결 검증 ===")

print(
    "최종 행:",
    len(analysis_2_3a_3b_final)
)

print(
    "포함 동:",
    analysis_2_3a_3b_final["행정동명_표준"].nunique()
)

print(
    "동×분기:",
    analysis_2_3a_3b_final[
        ["행정동명_표준", "기준_년분기_코드"]
    ].drop_duplicates().shape[0]
)

print(
    "동×행동신호월:",
    analysis_2_3a_3b_final[
        ["행정동명_표준", "기준연월"]
    ].drop_duplicates().shape[0]
)

print(
    "3-B 상대변화 결측:",
    analysis_2_3a_3b_final[
        "영역_상대변화_pp"
    ].isna().sum()
)

print("\n=== 동별 연결 행 수 ===")

print(
    analysis_2_3a_3b_final[
        "행정동명_표준"
    ].value_counts()
)
```
