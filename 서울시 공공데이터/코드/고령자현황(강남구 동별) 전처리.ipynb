```python
import pandas as pd

file_path = "고령자현황(강남구 동별).csv"

raw_elderly = pd.read_csv(
    file_path,
    encoding="utf-8-sig"
)

print("원본 shape:", raw_elderly.shape)
print()
print("컬럼:")
print(raw_elderly.columns.tolist())
print()
print("앞부분:")
print(raw_elderly.head(15))
print()
print("결측값:")
print(raw_elderly.isna().sum())
```

```python
elderly = raw_elderly.copy()

elderly.columns = [
    "dong_name",
    "category",
    "description",
    "empty_col",
    "2025Q3",
    "2025Q4"
]

# 완전히 비어 있는 컬럼 제거
elderly = elderly.drop(columns=["empty_col"])

print(elderly.head(15))
print()
print("shape:", elderly.shape)
print()
print("지역 목록:")
print(elderly["dong_name"].unique())
print()
print("지역별 행 수:")
print(elderly.groupby("dong_name").size())
```

```python
# 강남구 전체 행 제외 → 22개 행정동만
elderly_dong = elderly[
    elderly["dong_name"] != "강남구"
].copy()

print("shape:", elderly_dong.shape)
print("행정동 수:", elderly_dong["dong_name"].nunique())
print("결측값:", elderly_dong.isna().sum().sum())

# 각 동의 category 순서가 동일한지 확인
for dong in elderly_dong["dong_name"].unique()[:3]:
    print(f"\n[{dong}]")
    print(
        elderly_dong[
            elderly_dong["dong_name"] == dong
        ][["category", "2025Q3", "2025Q4"]].to_string(index=False)
    )
```

```python
# 기준이 되는 12행 구조
expected_categories = [
    "전체인구",
    "남자",
    "여자",
    "65세이상 인구",
    "남자",
    "여자",
    "내국인",
    "남자",
    "여자",
    "등록외국인",
    "남자",
    "여자"
]

# 모든 행정동의 구조가 동일한지 검증
structure_check = True

for dong in elderly_dong["dong_name"].unique():
    categories = (
        elderly_dong[elderly_dong["dong_name"] == dong]["category"]
        .tolist()
    )

    if categories != expected_categories:
        print("구조 불일치:", dong)
        structure_check = False

print("모든 행정동 구조 동일:", structure_check)
```

```python
elderly_records = []

for dong in elderly_dong["dong_name"].unique():

    dong_df = (
        elderly_dong[elderly_dong["dong_name"] == dong]
        .reset_index(drop=True)
    )

    for quarter in ["2025Q3", "2025Q4"]:

        elderly_records.append({
            "dong_name": dong,
            "quarter": quarter,

            # 전체 주민
            "resident_total_check": dong_df.loc[0, quarter],
            "resident_male": dong_df.loc[1, quarter],
            "resident_female": dong_df.loc[2, quarter],

            # 65세 이상
            "elderly_65_total": dong_df.loc[3, quarter],
            "elderly_65_male": dong_df.loc[4, quarter],
            "elderly_65_female": dong_df.loc[5, quarter],

            # 65세 이상 내국인
            "elderly_65_korean": dong_df.loc[6, quarter],
            "elderly_65_korean_male": dong_df.loc[7, quarter],
            "elderly_65_korean_female": dong_df.loc[8, quarter],

            # 65세 이상 등록외국인
            "elderly_65_foreigner": dong_df.loc[9, quarter],
            "elderly_65_foreigner_male": dong_df.loc[10, quarter],
            "elderly_65_foreigner_female": dong_df.loc[11, quarter]
        })

elderly_features = pd.DataFrame(elderly_records)

print("shape:", elderly_features.shape)
print("행정동 수:", elderly_features["dong_name"].nunique())
print("결측값:", elderly_features.isna().sum().sum())
print()
print(elderly_features.head())
```

```python
numeric_cols = [
    "resident_total_check",
    "resident_male",
    "resident_female",
    "elderly_65_total",
    "elderly_65_male",
    "elderly_65_female",
    "elderly_65_korean",
    "elderly_65_korean_male",
    "elderly_65_korean_female",
    "elderly_65_foreigner",
    "elderly_65_foreigner_male",
    "elderly_65_foreigner_female"
]

elderly_features[numeric_cols] = (
    elderly_features[numeric_cols]
    .apply(pd.to_numeric, errors="coerce")
)

print(elderly_features[numeric_cols].dtypes)
print("숫자 변환 후 결측값:", elderly_features[numeric_cols].isna().sum().sum())
```

```python
import numpy as np

check_1 = np.allclose(
    elderly_features["resident_total_check"],
    elderly_features["resident_male"]
    + elderly_features["resident_female"]
)

check_2 = np.allclose(
    elderly_features["elderly_65_total"],
    elderly_features["elderly_65_male"]
    + elderly_features["elderly_65_female"]
)

check_3 = np.allclose(
    elderly_features["elderly_65_total"],
    elderly_features["elderly_65_korean"]
    + elderly_features["elderly_65_foreigner"]
)

check_4 = np.allclose(
    elderly_features["elderly_65_korean"],
    elderly_features["elderly_65_korean_male"]
    + elderly_features["elderly_65_korean_female"]
)

check_5 = np.allclose(
    elderly_features["elderly_65_foreigner"],
    elderly_features["elderly_65_foreigner_male"]
    + elderly_features["elderly_65_foreigner_female"]
)

print("전체인구 성별 합계:", check_1)
print("65세 이상 성별 합계:", check_2)
print("65세 이상 내·외국인 합계:", check_3)
print("고령 내국인 성별 합계:", check_4)
print("고령 외국인 성별 합계:", check_5)
```

```python
resident = pd.read_csv(
    "gangnam_resident_population_features_2025Q3_Q4.csv",
    encoding="utf-8-sig"
)

compare = elderly_features.merge(
    resident[
        [
            "dong_name",
            "quarter",
            "resident_total",
            "resident_age_65_plus"
        ]
    ],
    on=["dong_name", "quarter"],
    how="left"
)

compare["total_match"] = (
    compare["resident_total_check"]
    == compare["resident_total"]
)

compare["elderly_match"] = (
    compare["elderly_65_total"]
    == compare["resident_age_65_plus"]
)

print("비교 행 수:", len(compare))
print("전체인구 모두 일치:", compare["total_match"].all())
print("65세 이상 모두 일치:", compare["elderly_match"].all())

print("\n불일치 행:")
print(
    compare[
        ~(compare["total_match"] & compare["elderly_match"])
    ][
        [
            "dong_name",
            "quarter",
            "resident_total_check",
            "resident_total",
            "elderly_65_total",
            "resident_age_65_plus"
        ]
    ]
)
```

```python
elderly_final = elderly_features[
    [
        "dong_name",
        "quarter",
        "elderly_65_male",
        "elderly_65_female",
        "elderly_65_korean",
        "elderly_65_foreigner",
        "elderly_65_korean_male",
        "elderly_65_korean_female",
        "elderly_65_foreigner_male",
        "elderly_65_foreigner_female"
    ]
].copy()

# 65세 이상 전체 인구는 비율 계산을 위한 분모로만 사용
elderly_total = elderly_features["elderly_65_total"]

elderly_final["elderly_65_male_ratio"] = (
    elderly_final["elderly_65_male"] / elderly_total
)

elderly_final["elderly_65_female_ratio"] = (
    elderly_final["elderly_65_female"] / elderly_total
)

elderly_final["elderly_65_korean_ratio"] = (
    elderly_final["elderly_65_korean"] / elderly_total
)

elderly_final["elderly_65_foreigner_ratio"] = (
    elderly_final["elderly_65_foreigner"] / elderly_total
)

print("shape:", elderly_final.shape)
print("결측값:", elderly_final.isna().sum().sum())
print()
print(elderly_final.head())
```

```python
import numpy as np

print(
    "고령자 성별 비율 합:",
    np.allclose(
        elderly_final["elderly_65_male_ratio"]
        + elderly_final["elderly_65_female_ratio"],
        1
    )
)

print(
    "고령자 내·외국인 비율 합:",
    np.allclose(
        elderly_final["elderly_65_korean_ratio"]
        + elderly_final["elderly_65_foreigner_ratio"],
        1
    )
)

print(
    "중복 키:",
    elderly_final.duplicated(
        subset=["dong_name", "quarter"]
    ).sum()
)
```

```python
output_file = "gangnam_elderly_features_2025Q3_Q4.csv"

elderly_final.to_csv(
    output_file,
    index=False,
    encoding="utf-8-sig"
)

print("저장 완료:", output_file)
print("shape:", elderly_final.shape)
```
