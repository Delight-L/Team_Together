```python
import pandas as pd
import numpy as np

from google.colab import files

# ==========================================
# STEP 1. 관심집단 장기 데이터 업로드
# ==========================================

uploaded = files.upload()

print("\n업로드 파일 수:", len(uploaded))

for filename in uploaded.keys():
    print(filename)
```

```python
# ==========================================
# STEP 2-1. 파일별 인코딩 진단
# ==========================================

from pathlib import Path

encoding_candidates = [
    "cp949",
    "utf-8-sig",
    "utf-8",
    "euc-kr",
    "utf-16",
    "utf-16-le",
    "utf-16-be"
]

encoding_result = []


for filename in uploaded.keys():

    success_encoding = None
    error_messages = []

    for enc in encoding_candidates:

        try:
            # 실제 CSV 파싱까지 해봄
            test_df = pd.read_csv(
                filename,
                encoding=enc,
                nrows=5
            )

            success_encoding = enc
            break

        except Exception as e:

            error_messages.append(
                f"{enc}: {type(e).__name__}"
            )


    encoding_result.append(
        {
            "filename": filename,
            "encoding": success_encoding,
            "readable": success_encoding is not None,
            "errors": " / ".join(error_messages)
        }
    )


encoding_check = pd.DataFrame(
    encoding_result
)


print("=== 파일별 인코딩 진단 ===")

print(
    encoding_check[
        [
            "filename",
            "encoding",
            "readable"
        ]
    ]
    .sort_values("filename")
    .to_string(index=False)
)


print(
    "\n=== 인코딩별 파일 수 ==="
)

print(
    encoding_check[
        "encoding"
    ]
    .fillna("판독 실패")
    .value_counts(
        dropna=False
    )
)


print(
    "\n판독 실패 파일 수:",
    (~encoding_check["readable"]).sum()
)


if (~encoding_check["readable"]).any():

    print(
        "\n=== 판독 실패 파일 ==="
    )

    print(
        encoding_check.loc[
            ~encoding_check["readable"],
            "filename"
        ]
        .to_string(index=False)
    )
```

```python
# ==========================================
# STEP 2 수정
# 관심집단 Excel 파일 구조 검사
# ==========================================

import pandas as pd

interest_raw = {}
read_errors = []


for filename in uploaded.keys():

    try:
        df = pd.read_excel(
            filename,
            engine="openpyxl"
        )

        interest_raw[filename] = df

    except Exception as e:

        read_errors.append(
            {
                "filename": filename,
                "error": str(e)
            }
        )


print("=== Excel 파일 읽기 결과 ===")
print("전체 파일 수:", len(uploaded))
print("성공:", len(interest_raw))
print("실패:", len(read_errors))


if read_errors:

    print("\n=== 읽기 실패 파일 ===")

    for item in read_errors:
        print(
            item["filename"],
            "→",
            item["error"]
        )


# ==========================================
# 파일별 Shape
# ==========================================

shape_summary = pd.DataFrame(
    [
        {
            "filename": filename,
            "rows": len(df),
            "columns": len(df.columns)
        }

        for filename, df
        in interest_raw.items()
    ]
)


print("\n=== 행 수 빈도 ===")

print(
    shape_summary["rows"]
    .value_counts()
    .sort_index()
)


print("\n=== 컬럼 수 빈도 ===")

print(
    shape_summary["columns"]
    .value_counts()
    .sort_index()
)


# ==========================================
# 컬럼 구조 비교
# ==========================================

column_structures = {}

for filename, df in interest_raw.items():

    structure = tuple(
        str(col).strip()
        for col in df.columns
    )

    column_structures.setdefault(
        structure,
        []
    ).append(filename)


print(
    "\n서로 다른 컬럼 구조 수:",
    len(column_structures)
)


for i, (columns, filenames) in enumerate(
    column_structures.items(),
    start=1
):

    print(
        f"\n=== 컬럼 구조 {i} ==="
    )

    print(
        "파일 수:",
        len(filenames)
    )

    print(
        "기간 예시:",
        filenames[:5]
    )

    print(
        "컬럼 수:",
        len(columns)
    )

    print(
        list(columns)
    )


# ==========================================
# 전체 파일 크기 확인
# ==========================================

print(
    "\n=== 파일별 크기 ==="
)

print(
    shape_summary
    .sort_values("filename")
    .to_string(index=False)
)
```

```python
import pandas as pd
import re

# =====================================================
# STEP 3-1. 파일명에서 기준연월 추출 후 48개월 통합
# =====================================================

interest_list = []

for filename, df in interest_raw.items():

    # 예:
    # 2022.1월_10개 관심집단수.xlsx
    # 2025.10월_10개 관심집단수.xlsx

    match = re.search(
        r"(\d{4})\.(\d{1,2})월",
        filename
    )

    if match is None:
        raise ValueError(
            f"기준연월 추출 실패: {filename}"
        )

    year = int(match.group(1))
    month = int(match.group(2))

    temp = df.copy()

    temp["기준연월"] = pd.Period(
        f"{year}-{month:02d}",
        freq="M"
    )

    interest_list.append(temp)


interest_all = pd.concat(
    interest_list,
    ignore_index=True
)


print("=== 전체 관심집단 통합 ===")
print("shape:", interest_all.shape)

print(
    "기간:",
    interest_all["기준연월"].min(),
    "~",
    interest_all["기준연월"].max()
)

print(
    "월 수:",
    interest_all["기준연월"].nunique()
)


# =====================================================
# STEP 3-2. 48개월 연속성 검사
# =====================================================

expected_months = pd.period_range(
    "2022-01",
    "2025-12",
    freq="M"
)

actual_months = (
    interest_all["기준연월"]
    .drop_duplicates()
    .sort_values()
)


missing_months = expected_months.difference(
    actual_months
)

extra_months = actual_months[
    ~actual_months.isin(expected_months)
]


print("\n=== 기간 연속성 ===")

print(
    "기대 월 수:",
    len(expected_months)
)

print(
    "실제 월 수:",
    len(actual_months)
)

print(
    "누락 월:",
    list(missing_months)
)

print(
    "예상 외 월:",
    list(extra_months)
)


# =====================================================
# STEP 3-3. 강남구 필터
# =====================================================

gangnam_interest = (
    interest_all[
        interest_all["자치구"] == "강남구"
    ]
    .copy()
)


print("\n=== 강남구 데이터 ===")

print(
    "shape:",
    gangnam_interest.shape
)

print(
    "행정동 수:",
    gangnam_interest["행정동명"].nunique()
)

print(
    "행정동코드 수:",
    gangnam_interest["행정동코드"].nunique()
)


# =====================================================
# STEP 3-4. 월별 행 수 / 행정동 수
# =====================================================

monthly_check = (
    gangnam_interest
    .groupby("기준연월")
    .agg(
        rows=("행정동명", "size"),
        dong_count=("행정동명", "nunique"),
        dong_code_count=("행정동코드", "nunique")
    )
    .reset_index()
)


print("\n=== 월별 구조 빈도 ===")

print(
    monthly_check[
        ["rows", "dong_count", "dong_code_count"]
    ]
    .value_counts()
)


print("\n=== 구조가 다른 월 ===")

abnormal_month = monthly_check[
    (monthly_check["rows"] != 528)
    |
    (monthly_check["dong_count"] != 22)
    |
    (monthly_check["dong_code_count"] != 22)
]

if len(abnormal_month) == 0:
    print("없음")
else:
    print(
        abnormal_month.to_string(
            index=False
        )
    )


# =====================================================
# STEP 3-5. Key 중복 검사
# 기준연월 × 행정동 × 성별 × 연령대
# =====================================================

key_cols = [
    "기준연월",
    "행정동코드",
    "성별",
    "연령대"
]

duplicate_count = (
    gangnam_interest
    .duplicated(
        subset=key_cols
    )
    .sum()
)


print("\n=== Key 검증 ===")

print(
    "Key 중복:",
    duplicate_count
)


# =====================================================
# STEP 3-6. 결측 검사
# =====================================================

missing_summary = (
    gangnam_interest
    .isna()
    .sum()
)

missing_summary = missing_summary[
    missing_summary > 0
]


print("\n=== 결측 컬럼 ===")

if len(missing_summary) == 0:
    print("없음")
else:
    print(missing_summary)


# =====================================================
# STEP 3-7. 행정동명 × 코드 관계 확인
# =====================================================

dong_map = (
    gangnam_interest[
        [
            "행정동코드",
            "행정동명"
        ]
    ]
    .drop_duplicates()
    .sort_values(
        [
            "행정동코드",
            "행정동명"
        ]
    )
)


print(
    "\n=== 전체 기간 행정동 코드-명칭 조합 ==="
)

print(
    dong_map.to_string(
        index=False
    )
)
```

```python
# =====================================================
# STEP 4. 관심집단 장기 데이터 값 일관성 검사
# =====================================================

# 1. 행정동 명칭 통일
gangnam_interest["행정동명"] = (
    gangnam_interest["행정동명"]
    .replace({
        "일원2동": "개포3동"
    })
)

# ---------------------------------------------
# 2. 분석 대상 수치 컬럼
# ---------------------------------------------

value_cols = [
    "총인구",
    "1인가구수",
    "커뮤니케이션이 적은 집단",
    "평일 외출이 적은 집단",
    "휴일 외출이 적은 집단",
    "출근소요시간 및 근무시간이 많은 집단",
    "외출이 매우 적은 집단(전체)",
    "외출이 매우 많은 집단",
    "동영상서비스 이용이 많은 집단",
    "생활서비스 이용이 많은 집단",
    "재정상태에 대한 관심집단",
    "외출-커뮤니케이션이 모두 적은 집단(전체)"
]

# 숫자형 확인
print("=== 데이터 타입 ===")
print(
    gangnam_interest[value_cols]
    .dtypes
)


# ---------------------------------------------
# 3. 월별 합계
# ---------------------------------------------

monthly_sum = (
    gangnam_interest
    .groupby("기준연월")[value_cols]
    .sum()
    .sort_index()
)

print("\n=== 월별 합계 앞 6개월 ===")
print(
    monthly_sum
    .head(6)
    .round(2)
    .to_string()
)

print("\n=== 월별 합계 마지막 6개월 ===")
print(
    monthly_sum
    .tail(6)
    .round(2)
    .to_string()
)


# ---------------------------------------------
# 4. 전월 대비 변화율
# 구조 변경 의심 시점 탐색용
# ---------------------------------------------

monthly_pct_change = (
    monthly_sum
    .pct_change(fill_method=None)
    * 100
)

# 각 월에서 변수들 가운데
# 가장 큰 절대 전월변화율
monthly_jump = pd.DataFrame({
    "max_abs_change_pct":
        monthly_pct_change.abs().max(axis=1),

    "median_abs_change_pct":
        monthly_pct_change.abs().median(axis=1)
})


print("\n=== 월별 변화가 큰 시점 TOP 15 ===")

print(
    monthly_jump
    .sort_values(
        "max_abs_change_pct",
        ascending=False
    )
    .head(15)
    .round(2)
    .to_string()
)


# ---------------------------------------------
# 5. 연도별 분포
# ---------------------------------------------

temp = gangnam_interest.copy()

temp["연도"] = (
    temp["기준연월"]
    .astype(str)
    .str[:4]
    .astype(int)
)

year_summary = (
    temp
    .groupby("연도")[value_cols]
    .agg(["mean", "std"])
)

print("\n=== 연도별 평균/표준편차 ===")

print(
    year_summary
    .round(2)
    .to_string()
)


# ---------------------------------------------
# 6. 0값 비율 검사
# ---------------------------------------------

zero_ratio = (
    temp[value_cols]
    .eq(0)
    .groupby(temp["연도"])
    .mean()
    * 100
)

print("\n=== 연도별 0값 비율(%) ===")

print(
    zero_ratio
    .round(2)
    .to_string()
)
```

```python
# =====================================================
# STEP 5. 2024년 전후 관심집단 값 구조 변화 점검
# =====================================================

interest_cols = [
    "커뮤니케이션이 적은 집단",
    "평일 외출이 적은 집단",
    "휴일 외출이 적은 집단",
    "출근소요시간 및 근무시간이 많은 집단",
    "외출이 매우 적은 집단(전체)",
    "외출이 매우 많은 집단",
    "동영상서비스 이용이 많은 집단",
    "생활서비스 이용이 많은 집단",
    "재정상태에 대한 관심집단",
    "외출-커뮤니케이션이 모두 적은 집단(전체)"
]

# -----------------------------------------------------
# 1. 월별 0값 비율
# -----------------------------------------------------

monthly_zero = (
    gangnam_interest
    .groupby("기준연월")[interest_cols]
    .agg(lambda x: (x == 0).mean() * 100)
)

print("=== 월별 0값 비율 (%) ===")

print(
    monthly_zero
    .round(2)
    .to_string()
)


# -----------------------------------------------------
# 2. 2024년 이후 0값이 어느 성별/연령대에 집중되는지
# -----------------------------------------------------

after_2024 = gangnam_interest[
    gangnam_interest["기준연월"] >= pd.Period("2024-01", freq="M")
].copy()

zero_records = []

for col in interest_cols:

    temp_zero = after_2024[
        after_2024[col] == 0
    ]

    if len(temp_zero) > 0:

        summary = (
            temp_zero
            .groupby(
                ["성별", "연령대"]
            )
            .size()
            .reset_index(name="zero_count")
        )

        summary["변수"] = col

        zero_records.append(summary)


if zero_records:

    zero_by_demo = pd.concat(
        zero_records,
        ignore_index=True
    )

    print(
        "\n=== 2024년 이후 0값이 많은 성별×연령대 TOP 30 ==="
    )

    print(
        zero_by_demo
        .sort_values(
            "zero_count",
            ascending=False
        )
        .head(30)
        [
            ["변수", "성별", "연령대", "zero_count"]
        ]
        .to_string(index=False)
    )


# -----------------------------------------------------
# 3. 2023-12 vs 2024-01 직접 비교
# -----------------------------------------------------

compare_months = gangnam_interest[
    gangnam_interest["기준연월"].isin(
        [
            pd.Period("2023-12", freq="M"),
            pd.Period("2024-01", freq="M")
        ]
    )
]

transition_summary = (
    compare_months
    .groupby("기준연월")[
        ["총인구", "1인가구수"] + interest_cols
    ]
    .agg(["mean", "median", "std"])
)

print(
    "\n=== 2023-12 vs 2024-01 분포 비교 ==="
)

print(
    transition_summary
    .round(3)
    .to_string()
)


# -----------------------------------------------------
# 4. 총인구의 월별 합계
# 2023~2024 경계 확인
# -----------------------------------------------------

population_transition = (
    gangnam_interest[
        (gangnam_interest["기준연월"] >= pd.Period("2023-07", freq="M"))
        &
        (gangnam_interest["기준연월"] <= pd.Period("2024-06", freq="M"))
    ]
    .groupby("기준연월")[
        ["총인구", "1인가구수"]
    ]
    .sum()
)

print(
    "\n=== 2023-07 ~ 2024-06 총인구/1인가구수 월별 합계 ==="
)

print(
    population_transition
    .round(2)
    .to_string()
)
```

```python
# =====================================================
# STEP 6. 2024년 이후 구조적 0 셀 확인
# =====================================================

check_cols = [
    "커뮤니케이션이 적은 집단",
    "평일 외출이 적은 집단",
    "동영상서비스 이용이 많은 집단",
    "재정상태에 대한 관심집단"
]

after_2024 = gangnam_interest[
    gangnam_interest["기준연월"] >= pd.Period("2024-01", freq="M")
].copy()


# -----------------------------------------------------
# 1. 각 셀(동×성×연령)이 24개월 중 몇 번 0인지
# -----------------------------------------------------

cell_keys = [
    "행정동코드",
    "행정동명",
    "성별",
    "연령대"
]

zero_frequency = (
    after_2024
    .groupby(cell_keys)[check_cols]
    .agg(lambda x: (x == 0).sum())
    .reset_index()
)


for col in check_cols:

    print(f"\n=== {col} : 24개월 중 0 발생 횟수 ===")

    print(
        zero_frequency[col]
        .value_counts()
        .sort_index()
        .to_string()
    )


# -----------------------------------------------------
# 2. 24개월 내내 0인 셀 확인
# -----------------------------------------------------

for col in check_cols:

    always_zero = zero_frequency[
        zero_frequency[col] == 24
    ][
        cell_keys
    ]

    print(
        f"\n=== {col} : 2024~2025 24개월 내내 0인 셀 ==="
    )

    print(
        "셀 수:",
        len(always_zero)
    )

    if len(always_zero) > 0:

        print(
            always_zero
            .to_string(index=False)
        )


# -----------------------------------------------------
# 3. 동일 셀의 2023년 값 확인
# -----------------------------------------------------

for col in check_cols:

    always_zero_keys = zero_frequency[
        zero_frequency[col] == 24
    ][cell_keys]

    if len(always_zero_keys) == 0:
        continue

    before = (
        gangnam_interest[
            (gangnam_interest["기준연월"] >= pd.Period("2023-01", freq="M"))
            &
            (gangnam_interest["기준연월"] <= pd.Period("2023-12", freq="M"))
        ]
        .merge(
            always_zero_keys,
            on=cell_keys,
            how="inner"
        )
    )

    print(
        f"\n=== {col} : 해당 셀의 2023년 값 요약 ==="
    )

    print(
        before[col]
        .describe()
        .round(3)
        .to_string()
    )
```

```python
import pandas as pd
import re

# =====================================================
# STEP T3. 48개월 통신정보 컬럼 표준화 + 통합
# =====================================================

telecom_list = []

for filename, df in telecom_raw.items():

    temp = df.copy()

    # -----------------------------------------
    # 1. 컬럼명 표준화
    # -----------------------------------------

    temp = temp.rename(columns={
        "행정동": "행정동명",
        "총인구수": "총인구"
    })

    # -----------------------------------------
    # 2. 파일명에서 기준연월 추출
    # -----------------------------------------

    match = re.search(
        r"(\d{4})\.(\d{1,2})월",
        filename
    )

    if match is None:
        raise ValueError(
            f"기준연월 추출 실패: {filename}"
        )

    year = int(match.group(1))
    month = int(match.group(2))

    temp["기준연월"] = pd.Period(
        f"{year}-{month:02d}",
        freq="M"
    )

    telecom_list.append(temp)


# ---------------------------------------------
# 3. 전체 통합
# ---------------------------------------------

telecom_all = pd.concat(
    telecom_list,
    ignore_index=True
)

print("=== 전체 통신정보 통합 ===")
print("shape:", telecom_all.shape)
print(
    "기간:",
    telecom_all["기준연월"].min(),
    "~",
    telecom_all["기준연월"].max()
)
print(
    "월 수:",
    telecom_all["기준연월"].nunique()
)


# =====================================================
# STEP T4. 기간 연속성
# =====================================================

expected_months = pd.period_range(
    "2022-01",
    "2025-12",
    freq="M"
)

actual_months = (
    telecom_all["기준연월"]
    .drop_duplicates()
    .sort_values()
)

print("\n=== 기간 연속성 ===")

print(
    "누락 월:",
    list(
        expected_months.difference(
            actual_months
        )
    )
)


# =====================================================
# STEP T5. 강남구 필터
# =====================================================

gangnam_telecom = (
    telecom_all[
        telecom_all["자치구"] == "강남구"
    ]
    .copy()
)

print("\n=== 강남구 데이터 ===")

print(
    "shape:",
    gangnam_telecom.shape
)

print(
    "행정동명 수:",
    gangnam_telecom["행정동명"].nunique()
)

print(
    "행정동코드 수:",
    gangnam_telecom["행정동코드"].nunique()
)


# =====================================================
# STEP T6. 월별 구조
# =====================================================

monthly_structure = (
    gangnam_telecom
    .groupby("기준연월")
    .agg(
        rows=("행정동코드", "size"),
        dong_count=("행정동명", "nunique"),
        dong_code_count=("행정동코드", "nunique")
    )
)

print("\n=== 월별 구조 빈도 ===")

print(
    monthly_structure
    .value_counts()
)


abnormal = monthly_structure[
    (monthly_structure["rows"] != 528)
    |
    (monthly_structure["dong_count"] != 22)
    |
    (monthly_structure["dong_code_count"] != 22)
]

print("\n=== 구조가 다른 월 ===")

if len(abnormal) == 0:
    print("없음")
else:
    print(abnormal)


# =====================================================
# STEP T7. Key 중복
# =====================================================

key_cols = [
    "기준연월",
    "행정동코드",
    "성별",
    "연령대"
]

duplicate_count = (
    gangnam_telecom
    .duplicated(
        subset=key_cols
    )
    .sum()
)

print("\n=== Key 검증 ===")
print(
    "Key 중복:",
    duplicate_count
)


# =====================================================
# STEP T8. 결측
# =====================================================

missing = (
    gangnam_telecom
    .isna()
    .sum()
)

missing = missing[
    missing > 0
]

print("\n=== 결측 컬럼 ===")

if len(missing) == 0:
    print("없음")
else:
    print(missing.to_string())


# =====================================================
# STEP T9. 행정동 코드-명칭 관계
# =====================================================

dong_map = (
    gangnam_telecom[
        [
            "행정동코드",
            "행정동명"
        ]
    ]
    .drop_duplicates()
    .sort_values(
        [
            "행정동코드",
            "행정동명"
        ]
    )
)

print(
    "\n=== 전체 기간 행정동 코드-명칭 조합 ==="
)

print(
    dong_map.to_string(
        index=False
    )
)


# =====================================================
# STEP T10. 1123074 월별 명칭 확인
# =====================================================

target_name = (
    gangnam_telecom[
        gangnam_telecom["행정동코드"]
        == 1123074
    ]
    .groupby("기준연월")[
        "행정동명"
    ]
    .unique()
)

print(
    "\n=== 1123074 월별 행정동명 ==="
)

print(
    target_name.to_string()
)
```

```python
# =====================================================
# STEP B1. A2 5개 변수 동×월 평균 생성
# =====================================================

a2_cols = [
    "평일 총 이동 횟수",
    "휴일 총 이동 횟수 평균",
    "집 추정 위치 평일 총 체류시간",
    "집 추정 위치 휴일 총 체류시간",
    "평균 통화대상자 수"
]

# 동 × 월 평균
a2_month = (
    gangnam_telecom
    .groupby(
        ["행정동코드", "행정동명", "기준연월"],
        as_index=False
    )[a2_cols]
    .mean()
)

print("=== A2 동×월 데이터 ===")
print("shape:", a2_month.shape)
print("동 수:", a2_month["행정동코드"].nunique())
print("월 수:", a2_month["기준연월"].nunique())

print(
    "Key 중복:",
    a2_month.duplicated(
        ["행정동코드", "기준연월"]
    ).sum()
)

print(
    "결측:",
    a2_month[a2_cols].isna().sum().sum()
)


# =====================================================
# STEP B2. 전월 대비 로그변화 계산
# =====================================================

a2_month = a2_month.sort_values(
    ["행정동코드", "기준연월"]
).copy()

for col in a2_cols:

    a2_month[f"{col}_log_change"] = (
        a2_month
        .groupby("행정동코드")[col]
        .transform(
            lambda x: np.log(x / x.shift(1))
        )
    )


log_cols = [
    f"{col}_log_change"
    for col in a2_cols
]


# =====================================================
# STEP B3. 월별 강남구 공통효과 계산
# =====================================================

monthly_common = (
    a2_month
    .groupby("기준연월")[log_cols]
    .median()
)

monthly_common.columns = [
    f"{col}_common"
    for col in a2_cols
]

monthly_common = (
    monthly_common
    .reset_index()
)

a2_change_long = a2_month.merge(
    monthly_common,
    on="기준연월",
    how="left",
    validate="many_to_one"
)


# =====================================================
# STEP B4. 공통효과 제거 residual
# =====================================================

residual_cols = []

for col in a2_cols:

    residual_col = f"{col}_residual"

    a2_change_long[residual_col] = (
        a2_change_long[f"{col}_log_change"]
        -
        a2_change_long[f"{col}_common"]
    )

    residual_cols.append(
        residual_col
    )


# =====================================================
# STEP B5. Historical / Evaluation 분리
# =====================================================

historical = a2_change_long[
    (
        a2_change_long["기준연월"]
        >= pd.Period("2022-02", freq="M")
    )
    &
    (
        a2_change_long["기준연월"]
        <= pd.Period("2025-06", freq="M")
    )
].copy()


evaluation = a2_change_long[
    (
        a2_change_long["기준연월"]
        >= pd.Period("2025-07", freq="M")
    )
    &
    (
        a2_change_long["기준연월"]
        <= pd.Period("2025-12", freq="M")
    )
].copy()


# =====================================================
# STEP B6. 검증
# =====================================================

print("\n=== Historical Baseline ===")
print("shape:", historical.shape)
print(
    "기간:",
    historical["기준연월"].min(),
    "~",
    historical["기준연월"].max()
)
print(
    "월 수:",
    historical["기준연월"].nunique()
)
print(
    "동 수:",
    historical["행정동코드"].nunique()
)
print(
    "residual 결측:",
    historical[residual_cols].isna().sum().sum()
)


print("\n=== Evaluation ===")
print("shape:", evaluation.shape)
print(
    "기간:",
    evaluation["기준연월"].min(),
    "~",
    evaluation["기준연월"].max()
)
print(
    "월 수:",
    evaluation["기준연월"].nunique()
)
print(
    "동 수:",
    evaluation["행정동코드"].nunique()
)
print(
    "residual 결측:",
    evaluation[residual_cols].isna().sum().sum()
)


# =====================================================
# STEP B7. 공통효과 제거 확인
# residual의 월별 중앙값은 거의 0이어야 함
# =====================================================

residual_monthly_median = (
    historical
    .groupby("기준연월")[residual_cols]
    .median()
)

print(
    "\n=== Historical residual 월별 중앙값 최대 절대값 ==="
)

print(
    residual_monthly_median
    .abs()
    .max()
    .to_string()
)


# =====================================================
# STEP B8. Historical residual 분포
# =====================================================

print(
    "\n=== Historical residual 분포 ==="
)

print(
    historical[residual_cols]
    .describe()
    .round(4)
    .T
    .to_string()
)
```

```python
# =====================================================
# STEP B9. 동별 Historical Baseline 생성
# =====================================================

historical_baseline_rows = []

for dong_code, group in historical.groupby("행정동코드"):

    dong_name = group["행정동명"].iloc[0]

    for col in a2_cols:

        residual_col = f"{col}_residual"

        values = group[residual_col].dropna()

        hist_median = values.median()

        mad = np.median(
            np.abs(values - hist_median)
        )

        robust_scale = 1.4826 * mad

        historical_baseline_rows.append({
            "행정동코드": dong_code,
            "행정동명": dong_name,
            "변수": col,
            "hist_n": len(values),
            "hist_median": hist_median,
            "hist_mad": mad,
            "hist_robust_scale": robust_scale
        })


historical_baseline = pd.DataFrame(
    historical_baseline_rows
)


print("=== Historical Baseline ===")
print("shape:", historical_baseline.shape)

print(
    "동 수:",
    historical_baseline["행정동코드"].nunique()
)

print(
    "변수 수:",
    historical_baseline["변수"].nunique()
)

print(
    "\nhist_n 빈도:"
)

print(
    historical_baseline["hist_n"]
    .value_counts()
    .sort_index()
)


print(
    "\n=== Robust scale 요약 ==="
)

print(
    historical_baseline
    .groupby("변수")["hist_robust_scale"]
    .describe()
    .round(6)
    .to_string()
)


print(
    "\n=== Robust scale = 0 검사 ==="
)

print(
    historical_baseline[
        historical_baseline["hist_robust_scale"] == 0
    ][
        ["행정동코드", "행정동명", "변수"]
    ]
    .to_string(index=False)
)


# =====================================================
# STEP B10. Evaluation을 Long 형태로 변환
# =====================================================

eval_long_list = []

for col in a2_cols:

    temp = evaluation[
        [
            "행정동코드",
            "행정동명",
            "기준연월",
            f"{col}_residual"
        ]
    ].copy()

    temp["변수"] = col

    temp = temp.rename(
        columns={
            f"{col}_residual":
            "eval_residual"
        }
    )

    eval_long_list.append(temp)


eval_long = pd.concat(
    eval_long_list,
    ignore_index=True
)


# =====================================================
# STEP B11. Historical Baseline 결합
# =====================================================

eval_hist = eval_long.merge(
    historical_baseline[
        [
            "행정동코드",
            "변수",
            "hist_n",
            "hist_median",
            "hist_mad",
            "hist_robust_scale"
        ]
    ],
    on=[
        "행정동코드",
        "변수"
    ],
    how="left",
    validate="many_to_one"
)


# =====================================================
# STEP B12. Historical Robust Z
# =====================================================

eval_hist["historical_robust_z"] = (
    (
        eval_hist["eval_residual"]
        -
        eval_hist["hist_median"]
    )
    /
    eval_hist["hist_robust_scale"]
)


# =====================================================
# STEP B13. 검증
# =====================================================

print("\n=== Evaluation Historical-Z ===")

print(
    "shape:",
    eval_hist.shape
)

print(
    "결측:",
    eval_hist[
        "historical_robust_z"
    ].isna().sum()
)

print(
    "inf:",
    np.isinf(
        eval_hist["historical_robust_z"]
    ).sum()
)


print(
    "\n=== 변수별 Historical-Z 분포 ==="
)

print(
    eval_hist
    .groupby("변수")[
        "historical_robust_z"
    ]
    .describe()
    .round(3)
    .to_string()
)


# =====================================================
# STEP B14. 절대 Z가 큰 관측치 개수
# 아직 threshold 판정용이 아니라 진단용
# =====================================================

diagnostic = []

for col in a2_cols:

    z = eval_hist.loc[
        eval_hist["변수"] == col,
        "historical_robust_z"
    ]

    diagnostic.append({
        "변수": col,
        "|Z|>=1": (z.abs() >= 1).sum(),
        "|Z|>=2": (z.abs() >= 2).sum(),
        "|Z|>=3": (z.abs() >= 3).sum(),
        "max_abs_Z": z.abs().max()
    })


diagnostic = pd.DataFrame(
    diagnostic
)

print(
    "\n=== Historical-Z 진단 ==="
)

print(
    diagnostic
    .round(3)
    .to_string(index=False)
)
```

```python
# =====================================================
# STEP B15. Historical Z를 기존 A2 방향으로 정렬
# =====================================================

direction_map = {
    "평일 총 이동 횟수": -1,
    "휴일 총 이동 횟수 평균": -1,
    "집 추정 위치 평일 총 체류시간": 1,
    "집 추정 위치 휴일 총 체류시간": 1,
    "평균 통화대상자 수": -1
}

eval_hist["direction_aligned_hist_z"] = (
    eval_hist["historical_robust_z"]
    *
    eval_hist["변수"].map(direction_map)
)


# =====================================================
# STEP B16. Wide 형태로 변환
# =====================================================

hist_wide = (
    eval_hist
    .pivot(
        index=[
            "행정동코드",
            "행정동명",
            "기준연월"
        ],
        columns="변수",
        values="direction_aligned_hist_z"
    )
    .reset_index()
)

hist_wide.columns.name = None


# =====================================================
# STEP B17. 기존 A2와 동일한 3개 축 생성
# =====================================================

hist_wide["hist_outside_activity_axis"] = (
    hist_wide["평일 총 이동 횟수"]
    +
    hist_wide["휴일 총 이동 횟수 평균"]
) / 2

hist_wide["hist_home_stay_axis"] = (
    hist_wide["집 추정 위치 평일 총 체류시간"]
    +
    hist_wide["집 추정 위치 휴일 총 체류시간"]
) / 2

hist_wide["hist_communication_axis"] = (
    hist_wide["평균 통화대상자 수"]
)


hist_axes = [
    "hist_outside_activity_axis",
    "hist_home_stay_axis",
    "hist_communication_axis"
]


# =====================================================
# STEP B18. 기본 검증
# =====================================================

print("=== Historical 검증축 ===")
print("shape:", hist_wide.shape)
print(
    "동 수:",
    hist_wide["행정동코드"].nunique()
)
print(
    "월 수:",
    hist_wide["기준연월"].nunique()
)
print(
    "Key 중복:",
    hist_wide.duplicated(
        ["행정동코드", "기준연월"]
    ).sum()
)
print(
    "축 결측:",
    hist_wide[hist_axes].isna().sum().sum()
)


# =====================================================
# STEP B19. 축 분포
# =====================================================

print("\n=== Historical 방향정렬 3축 분포 ===")

print(
    hist_wide[hist_axes]
    .describe()
    .round(3)
    .T
    .to_string()
)


# =====================================================
# STEP B20. 축 간 상관
# =====================================================

print("\n=== Historical 방향정렬 3축 상관 ===")

print(
    hist_wide[hist_axes]
    .corr()
    .round(3)
    .to_string()
)


# =====================================================
# STEP B21. 진단용 양의 Historical 신호 빈도
# 아직 최종 threshold 아님
# =====================================================

for threshold in [1, 2]:

    print(
        f"\n=== Historical axis >= {threshold} 진단 ==="
    )

    for col in hist_axes:

        print(
            col,
            ":",
            (hist_wide[col] >= threshold).sum(),
            "/",
            len(hist_wide)
        )


# =====================================================
# STEP B22. 가장 큰 양의 Historical 변화 사례
# =====================================================

for col in hist_axes:

    print(
        f"\n=== {col} TOP 10 ==="
    )

    print(
        hist_wide[
            [
                "행정동명",
                "기준연월",
                col
            ]
        ]
        .sort_values(
            col,
            ascending=False
        )
        .head(10)
        .round(3)
        .to_string(index=False)
    )
```

```python
# =====================================================
# STEP C2. 장기 관심집단 원자료에서 A1 Core 재계산
# =====================================================

interest_fix = gangnam_interest.copy()

# 행정동명 통일
interest_fix["행정동명"] = (
    interest_fix["행정동명"]
    .replace({"일원2동": "개포3동"})
)

# -----------------------------------------------------
# 1. 개포3동 구조적 오류 처리
# -----------------------------------------------------

invalid_mask = (
    (interest_fix["행정동코드"] == 1123074)
    &
    (interest_fix["기준연월"] >= pd.Period("2024-01", freq="M"))
)

# 문제가 확인된 변수만 결측 처리
interest_fix.loc[
    invalid_mask,
    "평일 외출이 적은 집단"
] = np.nan


print("=== 개포3동 구조적 오류 처리 ===")

print(
    "평일 외출 결측 수:",
    interest_fix.loc[
        interest_fix["행정동코드"] == 1123074,
        "평일 외출이 적은 집단"
    ].isna().sum()
)

print(
    "매우 낮은 외출 결측 수:",
    interest_fix.loc[
        interest_fix["행정동코드"] == 1123074,
        "외출이 매우 적은 집단(전체)"
    ].isna().sum()
)


# =====================================================
# STEP C3. 동×월 A1 비율 계산
#
# 비율 = 관심집단 합 / 1인가구수 합
# =====================================================

a1_month = (
    interest_fix
    .groupby(
        [
            "행정동코드",
            "행정동명",
            "기준연월"
        ],
        as_index=False
    )
    .agg(
        single_household_sum=(
            "1인가구수",
            "sum"
        ),
        low_weekday_outing_sum=(
            "평일 외출이 적은 집단",
            lambda x: x.sum(min_count=1)
        ),
        very_low_outing_sum=(
            "외출이 매우 적은 집단(전체)",
            "sum"
        )
    )
)


a1_month["low_weekday_outing_ratio"] = (
    a1_month["low_weekday_outing_sum"]
    /
    a1_month["single_household_sum"]
)

a1_month["very_low_outing_ratio"] = (
    a1_month["very_low_outing_sum"]
    /
    a1_month["single_household_sum"]
)


print("\n=== A1 동×월 ===")
print("shape:", a1_month.shape)

print(
    "동 수:",
    a1_month["행정동코드"].nunique()
)

print(
    "월 수:",
    a1_month["기준연월"].nunique()
)

print(
    "Key 중복:",
    a1_month.duplicated(
        ["행정동코드", "기준연월"]
    ).sum()
)


print("\n=== A1 비율 결측 ===")

print(
    a1_month[
        [
            "low_weekday_outing_ratio",
            "very_low_outing_ratio"
        ]
    ]
    .isna()
    .sum()
)


# =====================================================
# STEP C4. 2025-07~12만 추출
# =====================================================

a1_eval = (
    a1_month[
        (
            a1_month["기준연월"]
            >= pd.Period("2025-06", freq="M")
        )
        &
        (
            a1_month["기준연월"]
            <= pd.Period("2025-12", freq="M")
        )
    ]
    .copy()
)


# =====================================================
# STEP C5. 전월 대비 절대 차이
# 2025-06은 7월 변화 계산용
# =====================================================

a1_eval = a1_eval.sort_values(
    [
        "행정동코드",
        "기준연월"
    ]
)

for col in [
    "low_weekday_outing_ratio",
    "very_low_outing_ratio"
]:

    a1_eval[f"{col}_diff"] = (
        a1_eval
        .groupby("행정동코드")[col]
        .diff()
    )


# 7~12월 평가기간
a1_eval = a1_eval[
    a1_eval["기준연월"]
    >= pd.Period("2025-07", freq="M")
].copy()


# =====================================================
# STEP C6. 월 공통효과 제거 + Robust Z
# =====================================================

def monthly_robust_signal(df, diff_col, output_col):

    # 월별 중앙값
    month_median = (
        df
        .groupby("기준연월")[diff_col]
        .transform("median")
    )

    residual = (
        df[diff_col]
        -
        month_median
    )

    # 월별 MAD
    month_mad = (
        residual
        .groupby(df["기준연월"])
        .transform(
            lambda x:
            np.median(
                np.abs(
                    x.dropna()
                    -
                    np.median(x.dropna())
                )
            )
        )
    )

    scale = 1.4826 * month_mad

    df[output_col] = (
        residual / scale
    )

    return df


a1_eval = monthly_robust_signal(
    a1_eval,
    "low_weekday_outing_ratio_diff",
    "low_weekday_outing_signal"
)

a1_eval = monthly_robust_signal(
    a1_eval,
    "very_low_outing_ratio_diff",
    "very_low_outing_signal"
)


# =====================================================
# STEP C7. 검증
# =====================================================

print("\n=== 수정 A1 Signal ===")

print(
    "shape:",
    a1_eval.shape
)

print(
    "low_weekday signal 결측:",
    a1_eval[
        "low_weekday_outing_signal"
    ].isna().sum()
)

print(
    "very_low signal 결측:",
    a1_eval[
        "very_low_outing_signal"
    ].isna().sum()
)


# =====================================================
# STEP C8. 개포3동 2025 결과
# =====================================================

gaepo3_a1 = (
    a1_eval[
        a1_eval["행정동코드"] == 1123074
    ][
        [
            "행정동명",
            "기준연월",
            "low_weekday_outing_ratio",
            "low_weekday_outing_signal",
            "very_low_outing_ratio",
            "very_low_outing_signal"
        ]
    ]
)

print(
    "\n=== 개포3동 수정 A1 ==="
)

print(
    gaepo3_a1
    .round(4)
    .to_string(index=False)
)
```

```python
# =====================================================
# STEP C22. A1 Core 분모 수정
# =====================================================

a1_month_correct = (
    interest_fix
    .groupby(
        [
            "행정동코드",
            "행정동명",
            "기준연월"
        ],
        as_index=False
    )
    .agg(
        total_population_sum=(
            "총인구",
            "sum"
        ),
        single_household_sum=(
            "1인가구수",
            "sum"
        ),

        low_weekday_outing_sum=(
            "평일 외출이 적은 집단",
            lambda x: x.sum(min_count=1)
        ),

        very_low_outing_sum=(
            "외출이 매우 적은 집단(전체)",
            "sum"
        )
    )
)


# -----------------------------------------------------
# 올바른 분모 적용
# -----------------------------------------------------

# 1인가구 대상
a1_month_correct[
    "low_weekday_outing_ratio"
] = (
    a1_month_correct[
        "low_weekday_outing_sum"
    ]
    /
    a1_month_correct[
        "single_household_sum"
    ]
)

# 전체 시민 대상
a1_month_correct[
    "very_low_outing_ratio"
] = (
    a1_month_correct[
        "very_low_outing_sum"
    ]
    /
    a1_month_correct[
        "total_population_sum"
    ]
)


print("=== 수정된 A1 동×월 ===")
print("shape:", a1_month_correct.shape)

print(
    "low_weekday 결측:",
    a1_month_correct[
        "low_weekday_outing_ratio"
    ].isna().sum()
)

print(
    "very_low 결측:",
    a1_month_correct[
        "very_low_outing_ratio"
    ].isna().sum()
)


print("\n=== 비율 분포 ===")

print(
    a1_month_correct[
        [
            "low_weekday_outing_ratio",
            "very_low_outing_ratio"
        ]
    ]
    .describe()
    .round(4)
    .to_string()
)


# =====================================================
# STEP C23. 2025-06~12 추출
# =====================================================

a1_correct_eval = (
    a1_month_correct[
        (
            a1_month_correct["기준연월"]
            >= pd.Period("2025-06", freq="M")
        )
        &
        (
            a1_month_correct["기준연월"]
            <= pd.Period("2025-12", freq="M")
        )
    ]
    .copy()
)

a1_correct_eval = (
    a1_correct_eval
    .sort_values(
        [
            "행정동코드",
            "기준연월"
        ]
    )
)


# =====================================================
# STEP C24. 전월 대비 차이
# =====================================================

for col in [
    "low_weekday_outing_ratio",
    "very_low_outing_ratio"
]:

    a1_correct_eval[
        f"{col}_diff"
    ] = (
        a1_correct_eval
        .groupby("행정동코드")[col]
        .diff()
    )


# 평가기간 7~12월
a1_correct_eval = (
    a1_correct_eval[
        a1_correct_eval["기준연월"]
        >= pd.Period("2025-07", freq="M")
    ]
    .copy()
)


# =====================================================
# STEP C25. 월 공통효과 제거 + Robust Z
# 기존 함수 재사용
# =====================================================

a1_correct_eval = monthly_robust_signal(
    a1_correct_eval,
    "low_weekday_outing_ratio_diff",
    "low_weekday_outing_signal"
)

a1_correct_eval = monthly_robust_signal(
    a1_correct_eval,
    "very_low_outing_ratio_diff",
    "very_low_outing_signal"
)


# =====================================================
# STEP C26. 검증
# =====================================================

print("\n=== 올바른 분모 적용 후 Signal ===")

print(
    "shape:",
    a1_correct_eval.shape
)

print(
    "low_weekday 결측:",
    a1_correct_eval[
        "low_weekday_outing_signal"
    ].isna().sum()
)

print(
    "very_low 결측:",
    a1_correct_eval[
        "very_low_outing_signal"
    ].isna().sum()
)


# =====================================================
# STEP C27. 대치1동 / 개포3동 확인
# =====================================================

print("\n=== 대치1동 수정 결과 ===")

print(
    a1_correct_eval[
        a1_correct_eval["행정동코드"] == 1123060
    ][
        [
            "기준연월",
            "low_weekday_outing_ratio",
            "low_weekday_outing_signal",
            "very_low_outing_ratio",
            "very_low_outing_signal"
        ]
    ]
    .round(4)
    .to_string(index=False)
)


print("\n=== 개포3동 수정 결과 ===")

print(
    a1_correct_eval[
        a1_correct_eval["행정동코드"] == 1123074
    ][
        [
            "기준연월",
            "low_weekday_outing_ratio",
            "low_weekday_outing_signal",
            "very_low_outing_ratio",
            "very_low_outing_signal"
        ]
    ]
    .round(4)
    .to_string(index=False)
)
```

```python
# =====================================================
# STEP C28. 올바른 A1 + 기존방식 A2 최종 결합
# =====================================================

final_core = (
    a1_correct_eval[
        [
            "행정동코드",
            "행정동명",
            "기준연월",
            "low_weekday_outing_signal",
            "very_low_outing_signal"
        ]
    ]
    .merge(
        a2_short[
            [
                "행정동코드",
                "기준연월",
                "outside_activity_axis",
                "home_stay_axis",
                "communication_axis"
            ]
        ],
        on=[
            "행정동코드",
            "기준연월"
        ],
        how="left",
        validate="one_to_one"
    )
)


# =====================================================
# STEP C29. Core 조건
# =====================================================

# A1:
# - 일반 동: weekday OR very_low
# - 개포3동: weekday는 NaN이므로 very_low만 자동 사용

final_core["a1_core_present"] = (
    (final_core["low_weekday_outing_signal"] >= 1)
    |
    (final_core["very_low_outing_signal"] >= 1)
)


# A2:
# outside OR home
final_core["a2_core_present"] = (
    (final_core["outside_activity_axis"] >= 1)
    |
    (final_core["home_stay_axis"] >= 1)
)


final_core["core_coherent_month_final"] = (
    final_core["a1_core_present"]
    &
    final_core["a2_core_present"]
).astype(int)


# =====================================================
# STEP C30. 기존 FACT와 비교
# =====================================================

final_compare = (
    fact_month
    .merge(
        final_core[
            [
                "행정동코드",
                "기준연월",
                "core_coherent_month_final"
            ]
        ],
        on=[
            "행정동코드",
            "기준연월"
        ],
        how="left",
        validate="one_to_one"
    )
)


final_compare["changed"] = (
    final_compare["core_coherent_month"]
    !=
    final_compare["core_coherent_month_final"]
)


# 기존 분석과 동일한 비교범위 8~12월
final_compare_0812 = final_compare[
    final_compare["기준연월"]
    >= pd.Period("2025-08", freq="M")
].copy()


print("=== 2025-08~12 최종 비교 ===")

print(
    "기존 coherent:",
    final_compare_0812[
        "core_coherent_month"
    ].sum()
)

print(
    "최종 coherent:",
    final_compare_0812[
        "core_coherent_month_final"
    ].sum()
)


print("\n=== 기존 대비 변경된 관측 ===")

print(
    final_compare_0812[
        final_compare_0812["changed"]
    ].to_string(index=False)
)


# =====================================================
# STEP C31. 최종 coherent 목록
# =====================================================

final_coherent = final_core[
    (
        final_core["기준연월"]
        >= pd.Period("2025-08", freq="M")
    )
    &
    (
        final_core["core_coherent_month_final"] == 1
    )
].copy()


print("\n=== 최종 coherent 동×월 ===")

print(
    final_coherent[
        [
            "행정동명",
            "기준연월",
            "low_weekday_outing_signal",
            "very_low_outing_signal",
            "outside_activity_axis",
            "home_stay_axis"
        ]
    ]
    .sort_values(
        [
            "행정동명",
            "기준연월"
        ]
    )
    .round(3)
    .to_string(index=False)
)


# =====================================================
# STEP C32. 동별 최종 coherent 횟수
# =====================================================

final_dong_count = (
    final_core[
        final_core["기준연월"]
        >= pd.Period("2025-08", freq="M")
    ]
    .groupby(
        [
            "행정동코드",
            "행정동명"
        ],
        as_index=False
    )
    .agg(
        coherent_month_count=(
            "core_coherent_month_final",
            "sum"
        )
    )
)


print("\n=== 동별 최종 coherent 횟수 ===")

print(
    final_dong_count[
        final_dong_count[
            "coherent_month_count"
        ] > 0
    ]
    .sort_values(
        [
            "coherent_month_count",
            "행정동명"
        ],
        ascending=[
            False,
            True
        ]
    )
    .to_string(index=False)
)


# =====================================================
# STEP C33. 7월은 별도 신규 평가
# =====================================================

print("\n=== 장기자료로 새롭게 평가 가능한 2025-07 ===")

print(
    final_core[
        (
            final_core["기준연월"]
            == pd.Period("2025-07", freq="M")
        )
        &
        (
            final_core["core_coherent_month_final"] == 1
        )
    ][
        [
            "행정동명",
            "기준연월",
            "low_weekday_outing_signal",
            "very_low_outing_signal",
            "outside_activity_axis",
            "home_stay_axis"
        ]
    ]
    .round(3)
    .to_string(index=False)
)
```

```python
# =====================================================
# STEP C34. 최종 15건 × Historical Baseline 결합
# =====================================================

final_validation = (
    final_coherent
    .merge(
        hist_wide[
            [
                "행정동코드",
                "기준연월",
                "hist_outside_activity_axis",
                "hist_home_stay_axis",
                "hist_communication_axis"
            ]
        ],
        on=[
            "행정동코드",
            "기준연월"
        ],
        how="left",
        validate="one_to_one"
    )
)


print("=== 최종 Historical Validation ===")
print("shape:", final_validation.shape)

print(
    "결측:",
    final_validation[
        [
            "hist_outside_activity_axis",
            "hist_home_stay_axis",
            "hist_communication_axis"
        ]
    ].isna().sum().sum()
)


# =====================================================
# STEP C35. 방향 일치 여부
# =====================================================

final_validation[
    "hist_outside_positive"
] = (
    final_validation[
        "hist_outside_activity_axis"
    ] > 0
)

final_validation[
    "hist_home_positive"
] = (
    final_validation[
        "hist_home_stay_axis"
    ] > 0
)

final_validation[
    "hist_both_positive"
] = (
    final_validation[
        "hist_outside_positive"
    ]
    &
    final_validation[
        "hist_home_positive"
    ]
)


print("\n=== 최종 15건 Historical 방향 일치 ===")

print(
    "외부활동 감소 방향:",
    final_validation[
        "hist_outside_positive"
    ].sum(),
    "/ 15"
)

print(
    "집 체류 증가 방향:",
    final_validation[
        "hist_home_positive"
    ].sum(),
    "/ 15"
)

print(
    "두 축 동시 일치:",
    final_validation[
        "hist_both_positive"
    ].sum(),
    "/ 15"
)


# =====================================================
# STEP C36. 최종 15건 상세표
# =====================================================

print("\n=== 최종 15건 상세 ===")

print(
    final_validation[
        [
            "행정동명",
            "기준연월",

            "low_weekday_outing_signal",
            "very_low_outing_signal",

            "outside_activity_axis",
            "home_stay_axis",

            "hist_outside_activity_axis",
            "hist_home_stay_axis",

            "hist_both_positive"
        ]
    ]
    .sort_values(
        [
            "행정동명",
            "기준연월"
        ]
    )
    .round(3)
    .to_string(index=False)
)


# =====================================================
# STEP C37. 동별 Historical 검증 요약
# =====================================================

dong_validation = (
    final_validation
    .groupby(
        [
            "행정동코드",
            "행정동명"
        ],
        as_index=False
    )
    .agg(
        coherent_months=(
            "기준연월",
            "count"
        ),
        historical_both_positive_months=(
            "hist_both_positive",
            "sum"
        ),
        hist_outside_mean=(
            "hist_outside_activity_axis",
            "mean"
        ),
        hist_home_mean=(
            "hist_home_stay_axis",
            "mean"
        )
    )
)


print("\n=== 동별 최종 Historical 검증 ===")

print(
    dong_validation
    .sort_values(
        [
            "coherent_months",
            "행정동명"
        ],
        ascending=[
            False,
            True
        ]
    )
    .round(3)
    .to_string(index=False)
)
```
