```python
import pandas as pd

files = {
    "2025-07": "2025.7월_29개 통신정보.xlsx",
    "2025-08": "2025.8월_29개 통신정보.xlsx",
    "2025-09": "2025.9월_29개 통신정보.xlsx",
    "2025-10": "2025.10월_29개 통신정보.xlsx",
    "2025-11": "2025.11월_29개 통신정보.xlsx",
    "2025-12": "2025.12월_29개 통신정보.xlsx"
}

dfs = {}

for month, filename in files.items():
    df = pd.read_excel(filename)
    dfs[month] = df

    print(f"\n[{month}]")
    print("shape:", df.shape)
    print("컬럼 수:", len(df.columns))

print("\n===== 컬럼 구조 비교 =====")
base_columns = list(dfs["2025-07"].columns)

for month, df in dfs.items():
    print(
        month,
        "7월과 컬럼 동일:",
        list(df.columns) == base_columns
    )

print("\n===== 7월 컬럼 목록 =====")
for i, col in enumerate(base_columns, start=1):
    print(i, col)
```

```python
telecom_list = []

for month, df in dfs.items():
    temp = df.copy()
    temp["기준연월"] = month
    telecom_list.append(temp)

telecom_all = pd.concat(
    telecom_list,
    ignore_index=True
)

print("통합 shape:", telecom_all.shape)

print("\n월별 행 수:")
print(
    telecom_all["기준연월"]
    .value_counts()
    .sort_index()
)

print("\n앞부분 확인:")
display(telecom_all.head())
```

```python
# 기본 분류 변수 확인
basic_cols = ["행정동코드", "자치구", "행정동명", "성별", "연령대"]

print("===== 기본 컬럼 존재 여부 =====")
for col in basic_cols:
    print(col, ":", col in telecom_all.columns)

print("\n===== 기본 구조 =====")
print("행정동 수:", telecom_all["행정동코드"].nunique())
print("자치구 수:", telecom_all["자치구"].nunique())
print("성별:", sorted(telecom_all["성별"].dropna().unique()))
print("연령대:", sorted(telecom_all["연령대"].dropna().unique()))

# 월 × 행정동 × 성별 × 연령대 중복 확인
key_cols = ["기준연월", "행정동코드", "성별", "연령대"]

duplicate_count = telecom_all.duplicated(
    subset=key_cols
).sum()

print("\n===== 중복 검사 =====")
print("중복 key 수:", duplicate_count)

# 각 월의 행정동 수
print("\n===== 월별 행정동 수 =====")
print(
    telecom_all.groupby("기준연월")["행정동코드"]
    .nunique()
)

# 행정동별 월별 관측치 수가 동일한지 확인
panel_check = (
    telecom_all
    .groupby(["기준연월", "행정동코드"])
    .size()
)

print("\n===== 행정동 × 월 관측치 수 =====")
print(panel_check.describe())
print("고유한 관측치 수:", sorted(panel_check.unique()))
```

```python
interest = pd.read_csv(
    "gangnam_interest_groups_2025_07_12.csv",
    encoding="utf-8-sig"
)

print("관심집단 shape:", interest.shape)
print("통신정보 shape:", gangnam_telecom.shape)

print("\n관심집단 컬럼:")
print(interest.columns.tolist())
```

```python
key_cols = [
    "기준연월",
    "행정동코드",
    "성별",
    "연령대"
]

# 각 데이터 내부의 key 중복 확인
interest_dup = interest.duplicated(subset=key_cols).sum()
telecom_dup = gangnam_telecom.duplicated(subset=key_cols).sum()

print("===== 내부 중복 확인 =====")
print("관심집단 중복 key:", interest_dup)
print("통신정보 중복 key:", telecom_dup)


# key만 추출
interest_keys = interest[key_cols].drop_duplicates()
telecom_keys = gangnam_telecom[key_cols].drop_duplicates()

print("\n===== 고유 key 수 =====")
print("관심집단:", len(interest_keys))
print("통신정보:", len(telecom_keys))


# 두 데이터 key 비교
key_check = interest_keys.merge(
    telecom_keys,
    on=key_cols,
    how="outer",
    indicator=True
)

print("\n===== key 대응 결과 =====")
print(key_check["_merge"].value_counts())


# 서로 대응되지 않는 key 확인
unmatched = key_check[
    key_check["_merge"] != "both"
]

print("\n대응되지 않는 key 수:", len(unmatched))

if len(unmatched) > 0:
    display(unmatched.head(20))
else:
    print("모든 key가 정확히 1:1 대응합니다.")
```

```python
# 비교에 필요한 컬럼만 추출
interest_check = interest[
    key_cols + ["행정동명", "총인구", "1인가구수"]
].copy()

telecom_check = gangnam_telecom[
    key_cols + ["행정동", "총인구수", "1인가구수"]
].copy()

# 이름 충돌 방지를 위해 변경
interest_check = interest_check.rename(columns={
    "행정동명": "행정동_interest",
    "총인구": "총인구_interest",
    "1인가구수": "1인가구수_interest"
})

telecom_check = telecom_check.rename(columns={
    "행정동": "행정동_telecom",
    "총인구수": "총인구_telecom",
    "1인가구수": "1인가구수_telecom"
})

# 1:1 병합
common_check = interest_check.merge(
    telecom_check,
    on=key_cols,
    how="inner",
    validate="one_to_one"
)

print("비교 행 수:", len(common_check))


# 행정동명 비교
dong_match = (
    common_check["행정동_interest"]
    == common_check["행정동_telecom"]
)

print("\n===== 행정동명 =====")
print("일치:", dong_match.sum())
print("불일치:", (~dong_match).sum())


# 총인구 비교
population_diff = (
    common_check["총인구_interest"]
    - common_check["총인구_telecom"]
).abs()

print("\n===== 총인구 =====")
print("완전 일치:", (population_diff == 0).sum())
print("불일치:", (population_diff != 0).sum())
print("최대 차이:", population_diff.max())


# 1인가구수 비교
single_diff = (
    common_check["1인가구수_interest"]
    - common_check["1인가구수_telecom"]
).abs()

print("\n===== 1인가구수 =====")
print("완전 일치:", (single_diff == 0).sum())
print("불일치:", (single_diff != 0).sum())
print("최대 차이:", single_diff.max())


# 불일치 사례가 있다면 확인
mismatch = common_check[
    (~dong_match)
    | (population_diff > 0)
    | (single_diff > 0)
]

print("\n===== 하나라도 불일치한 행 =====")
print("행 수:", len(mismatch))

if len(mismatch) > 0:
    display(mismatch.head(20))
else:
    print("공통 기본정보가 모두 정확히 일치합니다.")
```

```python
# 통신정보에서 중복 기본정보 제외
telecom_exclude = [
    "자치구",
    "행정동",
    "총인구수",
    "1인가구수"
]

telecom_feature_cols = [
    col for col in gangnam_telecom.columns
    if col not in key_cols + telecom_exclude
]

print("추가할 통신정보 컬럼 수:", len(telecom_feature_cols))


# 관심집단 + 통신정보 병합
citizen_life = interest.merge(
    gangnam_telecom[
        key_cols + telecom_feature_cols
    ],
    on=key_cols,
    how="left",
    validate="one_to_one"
)

print("\n===== 병합 결과 =====")
print("shape:", citizen_life.shape)
print("행 수:", len(citizen_life))
print("컬럼 수:", len(citizen_life.columns))
print("전체 결측값 수:", citizen_life.isna().sum().sum())

print("\n===== 기본 구조 =====")
print("행정동 수:", citizen_life["행정동코드"].nunique())
print("월 수:", citizen_life["기준연월"].nunique())
print("성별:", sorted(citizen_life["성별"].unique()))
print("연령대:", sorted(citizen_life["연령대"].unique()))

print("\n===== key 중복 =====")
print(
    citizen_life.duplicated(
        subset=key_cols
    ).sum()
)
```

```python
# 1인가구 기반 관심집단
single_interest_map = {
    "커뮤니케이션이 적은 집단": "low_comm_ratio",
    "평일 외출이 적은 집단": "low_weekday_outing_ratio",
    "휴일 외출이 적은 집단": "low_holiday_outing_ratio",
    "출근소요시간 및 근무시간이 많은 집단": "high_worktime_ratio",
    "동영상서비스 이용이 많은 집단": "high_video_ratio",
    "생활서비스 이용이 많은 집단": "high_life_service_ratio",
    "재정상태에 대한 관심집단": "financial_interest_ratio"
}

for source_col, feature_col in single_interest_map.items():
    citizen_life[feature_col] = (
        citizen_life[source_col]
        / citizen_life["1인가구수"]
        * 100
    )


# 전체 인구 기반 관심집단
whole_interest_map = {
    "외출이 매우 적은 집단(전체)": "very_low_outing_ratio",
    "외출-커뮤니케이션이 모두 적은 집단(전체)": "very_low_outing_comm_ratio"
}

for source_col, feature_col in whole_interest_map.items():
    citizen_life[feature_col] = (
        citizen_life[source_col]
        / citizen_life["총인구"]
        * 100
    )


# 생성된 Feature 확인
ratio_cols = (
    list(single_interest_map.values())
    + list(whole_interest_map.values())
)

print("생성된 비율 Feature 수:", len(ratio_cols))

ratio_summary = citizen_life[ratio_cols].agg(
    ["min", "median", "mean", "max"]
).T

display(ratio_summary.round(2))

print(
    "\n100% 초과 값 수:",
    (citizen_life[ratio_cols] > 100).sum().sum()
)

print(
    "결측값 수:",
    citizen_life[ratio_cols].isna().sum().sum()
)
```

```python
monthly_interest = (
    citizen_life
    .groupby("기준연월")[ratio_cols]
    .mean()
    .round(2)
)

print("===== 월별 관심집단 비율 평균 =====")
display(monthly_interest)

print("\n===== 월별 변화 범위 =====")

monthly_range = pd.DataFrame({
    "min": monthly_interest.min(),
    "max": monthly_interest.max()
})

monthly_range["range"] = (
    monthly_range["max"]
    - monthly_range["min"]
)

display(
    monthly_range
    .sort_values("range", ascending=False)
    .round(2)
)
```

```python
# 동 × 월 단위로 원자료 합산
dong_month_interest = (
    citizen_life
    .groupby(
        ["행정동코드", "행정동명", "기준연월"],
        as_index=False
    )
    .agg({
        "총인구": "sum",
        "1인가구수": "sum",
        "커뮤니케이션이 적은 집단": "sum",
        "평일 외출이 적은 집단": "sum",
        "휴일 외출이 적은 집단": "sum",
        "출근소요시간 및 근무시간이 많은 집단": "sum",
        "동영상서비스 이용이 많은 집단": "sum",
        "생활서비스 이용이 많은 집단": "sum",
        "재정상태에 대한 관심집단": "sum",
        "외출이 매우 적은 집단(전체)": "sum",
        "외출-커뮤니케이션이 모두 적은 집단(전체)": "sum"
    })
)

# 1인가구 기반 7개 비율
for source_col, feature_col in single_interest_map.items():
    dong_month_interest[feature_col] = (
        dong_month_interest[source_col]
        / dong_month_interest["1인가구수"]
        * 100
    )

# 전체 인구 기반 2개 비율
for source_col, feature_col in whole_interest_map.items():
    dong_month_interest[feature_col] = (
        dong_month_interest[source_col]
        / dong_month_interest["총인구"]
        * 100
    )

print("동×월 shape:", dong_month_interest.shape)
print(
    "동×월 key 중복:",
    dong_month_interest.duplicated(
        ["행정동코드", "기준연월"]
    ).sum()
)

# 각 동의 6개월 표준편차와 범위 계산
dong_variation = (
    dong_month_interest
    .groupby(["행정동코드", "행정동명"])[ratio_cols]
    .agg(["mean", "std", "min", "max"])
)

# 전체적으로 월간 변동성이 어느 정도인지 요약
variation_summary = []

for col in ratio_cols:
    variation_summary.append({
        "feature": col,
        "평균비율": dong_variation[(col, "mean")].mean(),
        "동별_평균_std": dong_variation[(col, "std")].mean(),
        "동별_최대_std": dong_variation[(col, "std")].max(),
        "동별_평균_range": (
            dong_variation[(col, "max")]
            - dong_variation[(col, "min")]
        ).mean(),
        "동별_최대_range": (
            dong_variation[(col, "max")]
            - dong_variation[(col, "min")]
        ).max()
    })

variation_summary = pd.DataFrame(variation_summary)

display(
    variation_summary
    .sort_values("동별_평균_std", ascending=False)
    .round(2)
)
```

```python
# 외출이 매우 많은 집단의 동×월 원값 추가
very_high_outing = (
    citizen_life
    .groupby(
        ["행정동코드", "행정동명", "기준연월"],
        as_index=False
    )["외출이 매우 많은 집단"]
    .sum()
)

dong_month_interest = dong_month_interest.merge(
    very_high_outing,
    on=["행정동코드", "행정동명", "기준연월"],
    how="left",
    validate="one_to_one"
)

# 분석용 동×월 Feature Table
interest_month_feature = dong_month_interest[
    [
        "행정동코드",
        "행정동명",
        "기준연월",
        "총인구",
        "1인가구수",
        *ratio_cols,
        "외출이 매우 많은 집단"
    ]
].copy()

print("Feature shape:", interest_month_feature.shape)

print(
    "key 중복:",
    interest_month_feature.duplicated(
        ["행정동코드", "기준연월"]
    ).sum()
)

print(
    "결측값:",
    interest_month_feature.isna().sum().sum()
)

display(interest_month_feature.head())
```

```python
# 9개 관심집단 비율의 동별 평균 + 표준편차
interest_dong_agg = (
    interest_month_feature
    .groupby(
        ["행정동코드", "행정동명"]
    )[ratio_cols]
    .agg(["mean", "std"])
)

# MultiIndex 컬럼을 일반 컬럼으로 변환
interest_dong_agg.columns = [
    f"{feature}_{stat}"
    for feature, stat in interest_dong_agg.columns
]

interest_dong_feature = (
    interest_dong_agg
    .reset_index()
)

print("Feature shape:", interest_dong_feature.shape)

print(
    "행정동 수:",
    interest_dong_feature["행정동코드"].nunique()
)

print(
    "key 중복:",
    interest_dong_feature["행정동코드"].duplicated().sum()
)

print(
    "결측값:",
    interest_dong_feature.isna().sum().sum()
)

print("\n===== 컬럼 목록 =====")
for col in interest_dong_feature.columns:
    print(col)

display(interest_dong_feature.head())
```

```python
zero_relation = []

for variable, status_col in true_unestimated_map.items():

    effective_n = (
        citizen_life["총인구"]
        - citizen_life[status_col]
    )

    value = citizen_life[variable]

    zero_relation.append({
        "통신정보": variable,

        # 대표값 자체가 0인 행
        "값0_행": (value == 0).sum(),

        # 유효 추정인구가 0인 행
        "유효인구0_행": (effective_n == 0).sum(),

        # 유효인구=0인데 대표값도 0
        "유효인구0_값0": (
            (effective_n == 0) &
            (value == 0)
        ).sum(),

        # 유효인구가 있는데 대표값=0
        "유효인구양수_값0": (
            (effective_n > 0) &
            (value == 0)
        ).sum(),

        # 유효인구=0인데 대표값이 0이 아님
        "유효인구0_값비0": (
            (effective_n == 0) &
            (value != 0)
        ).sum()
    })

zero_relation = pd.DataFrame(zero_relation)

display(
    zero_relation.sort_values(
        "유효인구0_행",
        ascending=False
    )
)
```

```python
no_status_vars = [
    "최근 3개월 내 요금 연체 비율",
    "평균 통화량",
    "평균 문자량",
    "평균 통화대상자 수",
    "평균 문자대상자 수",
    "데이터 사용량"
]

rows = []

for (dong_code, dong_name, month), g in citizen_life.groupby(
    ["행정동코드", "행정동명", "기준연월"]
):
    row = {
        "행정동코드": dong_code,
        "행정동명": dong_name,
        "기준연월": month
    }

    for col in no_status_vars:

        # 성별·연령대 셀의 단순평균
        simple_mean = g[col].mean()

        # 총인구 가중평균
        weighted_mean = (
            (g[col] * g["총인구"]).sum()
            / g["총인구"].sum()
        )

        row[f"{col}_simple"] = simple_mean
        row[f"{col}_weighted"] = weighted_mean
        row[f"{col}_diff"] = weighted_mean - simple_mean

    rows.append(row)

no_status_compare = pd.DataFrame(rows)

print("shape:", no_status_compare.shape)
print(
    "key 중복:",
    no_status_compare.duplicated(
        ["행정동코드", "기준연월"]
    ).sum()
)

comparison_summary = []

for col in no_status_vars:
    diff = no_status_compare[f"{col}_diff"]

    comparison_summary.append({
        "통신정보": col,
        "평균_절대차이": diff.abs().mean(),
        "최대_절대차이": diff.abs().max()
    })

comparison_summary = pd.DataFrame(comparison_summary)

display(
    comparison_summary
    .sort_values("평균_절대차이", ascending=False)
    .round(4)
)
```

```python
rows = []

for keys, g in citizen_life.groupby(
    ["행정동코드", "행정동명", "기준연월"]
):
    dong_code, dong_name, month = keys

    row = {
        "행정동코드": dong_code,
        "행정동명": dong_name,
        "기준연월": month
    }

    total_pop = g["총인구"].sum()

    # -------------------------
    # 1. 소액결제 사용횟수
    # -------------------------
    payment_users = (
        g["총인구"]
        - g["소액결재 비사용 인구수"]
    )

    row["소액결재_사용자비율"] = (
        payment_users.sum()
        / total_pop * 100
    )

    if payment_users.sum() > 0:
        row["소액결재 사용횟수 평균"] = np.average(
            g["소액결재 사용횟수 평균"],
            weights=payment_users
        )
    else:
        row["소액결재 사용횟수 평균"] = np.nan

    # -------------------------
    # 2. 소액결제 사용금액
    # -------------------------
    payment_amount_pop = (
        g["총인구"]
        - g["소액결재 사용 금액 없는 인구수"]
    )

    row["소액결재_금액존재비율"] = (
        payment_amount_pop.sum()
        / total_pop * 100
    )

    if payment_amount_pop.sum() > 0:
        row["소액결재 사용금액 평균"] = np.average(
            g["소액결재 사용금액 평균"],
            weights=payment_amount_pop
        )
    else:
        row["소액결재 사용금액 평균"] = np.nan

    # -------------------------
    # 3. SNS
    # -------------------------
    # SNS 값 자체는 Z-score이므로 총인구 가중
    row["SNS 사용횟수"] = np.average(
        g["SNS 사용횟수"],
        weights=g["총인구"]
    )

    # 카카오톡 비사용은 별도 비율 Feature
    row["카카오톡_비사용비율"] = (
        g["카카오톡 비사용 인구수"].sum()
        / total_pop * 100
    )

    rows.append(row)

telecom_special_feature = pd.DataFrame(rows)

print("shape:", telecom_special_feature.shape)

print(
    "key 중복:",
    telecom_special_feature.duplicated(
        ["행정동코드", "기준연월"]
    ).sum()
)

print(
    "결측값:",
    telecom_special_feature.isna().sum().sum()
)

display(
    telecom_special_feature[
        [
            "소액결재_사용자비율",
            "소액결재 사용횟수 평균",
            "소액결재_금액존재비율",
            "소액결재 사용금액 평균",
            "SNS 사용횟수",
            "카카오톡_비사용비율"
        ]
    ].describe().T.round(3)
)
```

```python
# 비교 키
key_cols = [
    "행정동코드",
    "행정동명",
    "기준연월"
]

# -----------------------------------
# 1. 각 테이블 기본 구조 확인
# -----------------------------------
print("관심집단:", interest_month_feature.shape)
print("통신정보:", telecom_core_month_feature.shape)

print(
    "관심집단 key 중복:",
    interest_month_feature
    .duplicated(["행정동코드", "기준연월"])
    .sum()
)

print(
    "통신정보 key 중복:",
    telecom_core_month_feature
    .duplicated(["행정동코드", "기준연월"])
    .sum()
)

# -----------------------------------
# 2. 행정동×월 키 일치 여부 확인
# -----------------------------------
key_check = (
    interest_month_feature[key_cols]
    .merge(
        telecom_core_month_feature[key_cols],
        on=key_cols,
        how="outer",
        indicator=True
    )
)

print("\n키 비교 결과")
print(key_check["_merge"].value_counts())

# -----------------------------------
# 3. 한쪽에만 존재하는 키 확인
# -----------------------------------
unmatched = key_check[
    key_check["_merge"] != "both"
]

print(
    "\n불일치 key 수:",
    len(unmatched)
)

if len(unmatched) > 0:
    display(unmatched)
```

```python
# 관심집단에서 사용할 9개 비율 Feature
interest_ratio_cols = [
    "low_comm_ratio",
    "low_weekday_outing_ratio",
    "low_holiday_outing_ratio",
    "high_worktime_ratio",
    "high_video_ratio",
    "high_life_service_ratio",
    "financial_interest_ratio",
    "very_low_outing_ratio",
    "very_low_outing_comm_ratio"
]

# -----------------------------------
# 1. 두 테이블 임시 결합
# -----------------------------------
relation_check = (
    interest_month_feature[
        key_cols + interest_ratio_cols
    ]
    .merge(
        telecom_core_month_feature[
            key_cols + core_cols
        ],
        on=key_cols,
        how="inner",
        validate="one_to_one"
    )
)

print("임시 결합 shape:", relation_check.shape)
print("결측값:", relation_check.isna().sum().sum())

# -----------------------------------
# 2. 관심집단 × 통신정보 교차상관
# -----------------------------------
cross_corr = relation_check[
    interest_ratio_cols + core_cols
].corr().loc[
    interest_ratio_cols,
    core_cols
]

# |r| >= 0.7인 관계만 추출
cross_pairs = []

for interest_var in interest_ratio_cols:
    for telecom_var in core_cols:

        r = cross_corr.loc[
            interest_var,
            telecom_var
        ]

        if abs(r) >= 0.7:
            cross_pairs.append({
                "관심집단_Feature": interest_var,
                "통신정보_Feature": telecom_var,
                "상관계수": r
            })

cross_pairs = pd.DataFrame(cross_pairs)

if len(cross_pairs) > 0:

    cross_pairs["절대상관"] = (
        cross_pairs["상관계수"].abs()
    )

    cross_pairs = (
        cross_pairs
        .sort_values(
            "절대상관",
            ascending=False
        )
        .drop(columns="절대상관")
        .reset_index(drop=True)
    )

print(
    "|r| >= 0.7 교차 관계 수:",
    len(cross_pairs)
)

display(cross_pairs.round(3))
```

```python
# 22개 행정동의 6개월 평균
relation_dong = (
    relation_check
    .groupby(
        ["행정동코드", "행정동명"]
    )[interest_ratio_cols + core_cols]
    .mean()
    .reset_index()
)

print("동 단위 shape:", relation_dong.shape)

# 관심집단 × 통신정보 교차상관
cross_corr_22 = (
    relation_dong[
        interest_ratio_cols + core_cols
    ]
    .corr()
    .loc[
        interest_ratio_cols,
        core_cols
    ]
)

cross_pairs_22 = []

for interest_var in interest_ratio_cols:
    for telecom_var in core_cols:

        r = cross_corr_22.loc[
            interest_var,
            telecom_var
        ]

        if abs(r) >= 0.7:
            cross_pairs_22.append({
                "관심집단_Feature": interest_var,
                "통신정보_Feature": telecom_var,
                "상관계수": r
            })

cross_pairs_22 = pd.DataFrame(
    cross_pairs_22
)

if len(cross_pairs_22) > 0:

    cross_pairs_22["절대상관"] = (
        cross_pairs_22[
            "상관계수"
        ].abs()
    )

    cross_pairs_22 = (
        cross_pairs_22
        .sort_values(
            "절대상관",
            ascending=False
        )
        .drop(columns="절대상관")
        .reset_index(drop=True)
    )

print(
    "|r| >= 0.7 동 단위 교차 관계 수:",
    len(cross_pairs_22)
)

display(
    cross_pairs_22.round(3)
)
```
